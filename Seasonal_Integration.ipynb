```python
# Melt onset + Growth Initiation (GI) detector — relaxed CHL gate
# with diagnostics, padded export, and plotting utilities (merged cell)

# ===== imports =====
import os, glob
import numpy as np
import xarray as xr
import pandas as pd
import gsw
from scipy.signal import argrelextrema as lmm
import scipy.signal as signal
import matplotlib.pyplot as plt

# ===== small helpers =====
def safe_interp_1d(arr):
    """Linearly interpolate NaNs only if there are >=2 good points; otherwise return as-is."""
    out = np.asarray(arr, dtype=float).copy()
    nans = np.isnan(out)
    good = ~nans
    if good.sum() >= 2:
        idx = np.flatnonzero
        out[nans] = np.interp(idx(nans), idx(good), out[good])
    return out

def safe_filtfilt(B, A, arr):
    """Apply filtfilt only if data length is sufficient and finite; else return input."""
    a = np.asarray(arr, dtype=float)
    if np.all(np.isfinite(a)) and a.size > 6:  # N=1 butter requires padlen safety
        try:
            return signal.filtfilt(B, A, a, method="pad")
        except Exception:
            return a
    return a

def pick_var(ds, base):
    """Prefer *_ADJUSTED if present, else base."""
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base]

def good_qc_mask(ds, base, qc_ok=("1","2")):
    """Return boolean mask where QC is in qc_ok (e.g., '1','2')."""
    qname = base + "_ADJUSTED_QC" if base + "_ADJUSTED_QC" in ds.variables else base + "_QC"
    if qname not in ds.variables:
        return None
    qc = ds[qname].astype("U1")
    return qc.isin(list(qc_ok))

def get_time_index(ds):
    """Robust Pandas DatetimeIndex from CF TIME."""
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    ti = t.to_index() if hasattr(t, "to_index") else pd.to_datetime(t.values)
    if hasattr(ti, "to_datetimeindex"):
        return ti.to_datetimeindex()
    return pd.DatetimeIndex(ti)

def gi_fallback_from_increase(chls):
    """
    Fallback GI pick:
      - interpolate if possible
      - simple 3-pt moving average
      - pick first index where we see two consecutive positive increments
        AND value >= (min + 0.5*std) threshold (if std available)
      - else first positive increment
    Returns global index 'i' or None.
    """
    a = np.asarray(chls, float)
    if np.isfinite(a).sum() < 5:
        return None
    a = safe_interp_1d(a)
    # simple moving average (3)
    if len(a) >= 3:
        kernel = np.ones(3)/3.0
        a_sm = np.convolve(a, kernel, mode="same")
    else:
        a_sm = a.copy()
    d = np.diff(a_sm)
    # threshold
    finite = np.isfinite(a_sm)
    if finite.sum() >= 2:
        std = np.nanstd(a_sm)
        amin = np.nanmin(a_sm)
        thresh = amin + 0.5*std if np.isfinite(std) else amin
    else:
        thresh = np.nanmin(a_sm[finite]) if finite.any() else 0.0
    # two-step rise crossing threshold
    for i in range(len(d)-1):
        if d[i] > 0 and d[i+1] > 0 and a_sm[i+1] >= thresh:
            return i+1
    # single positive step fallback
    for i in range(len(d)):
        if d[i] > 0:
            return i+1
    return None

# ===== main detector (diagnostics-capable) =====
def save_meltbloom2(
    rtc=0.5,                    # T - T_freeze <= rtc marks under-ice
    bl=0,                       # extra levels below MLD for CHL mean
    DIR="",
    qc_ok=("0","1","2","8"),    # relaxed for optics
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,     # prefer CHL within 0–100 m
    deep_gate_depth_m=200.0,    # fallback up to 200 m (logged)
    collect_diag=True
):
    def find_nearest(array, value):
        idx = np.nanargmin(np.abs(array - value))
        return int(idx)

    def frac_finite(arr):
        arr = np.asarray(arr)
        return 100.0 * np.isfinite(arr).sum() / arr.size if arr.size else 0.0

    flist = glob.glob(os.path.join(DIR, "*.nc"))
    print("No. of files:", len(flist))

    # aggregated outputs
    bloomdays  = np.array([]); bloomweeks = np.array([]); bloomweeks_m = np.array([])
    bloomyears = np.array([]); meltweeks  = np.array([]); meltdays     = np.array([])
    lons = np.array([]); lats = np.array([]); files = np.array([]); mld_gi = np.array([])
    mld_chl_pos = np.array([]); timing_pg = np.array([]); grth_rate_gm = np.array([])
    peak_chl = np.array([]); len_ts = np.array([]); melt_week = np.array([]); chl_sog = np.array([])
    gia_week = np.array([])

    # diagnostics container
    diag = {}

    # filters
    N = 1; Wn, Wn2 = 0.1, 0.2
    B, A  = signal.butter(N, Wn,  output="ba")
    B2, A2 = signal.butter(N, Wn2, output="ba")

    for f, path in enumerate(flist):
        print(f"\n=== File {f} { os.path.basename(path) } ===")
        ds = xr.open_dataset(path)

        # coords + variables
        if not all(v in ds.variables for v in ("TIME","LONGITUDE","LATITUDE")):
            print("  STOP: missing TIME/LONGITUDE/LATITUDE")
            ds.close(); continue
        t  = pick_var(ds, "TEMP")
        s  = pick_var(ds, "PSAL")
        pr = pick_var(ds, "PRES")
        if ("CPHL" in ds.variables) or ("CPHL_ADJUSTED" in ds.variables):
            chl = pick_var(ds, "CPHL")
        else:
            print("  STOP: no CPHL/CPHL_ADJUSTED")
            ds.close(); continue
        lon = ds["LONGITUDE"]; lat = ds["LATITUDE"]
        time = get_time_index(ds)

        lon_dim = lon.dims[0]; lat_dim = lat.dims[0]
        if "TIME" not in t.sizes:
            print("  STOP: TEMP has no TIME dim")
            ds.close(); continue
        nprof = int(t.sizes["TIME"])
        if lon.sizes[lon_dim] != nprof or lat.sizes[lat_dim] != nprof:
            print(f"  STOP: lon/lat length mismatch vs TIME ({lon.sizes[lon_dim]}/{lat.sizes[lat_dim]} vs {nprof})")
            ds.close(); continue

        # QC + screens
        for da, base in [(t,"TEMP"),(s,"PSAL"),(pr,"PRES"),(chl,"CPHL")]:
            m = good_qc_mask(ds, base, qc_ok=qc_ok)
            if m is not None and da.dims == m.dims:
                da[:] = da.where(m)
        s   = s.where(s > 28.0)
        chl = chl.where(chl >= 0)

        # timing diagnostics
        if nprof >= 2:
            step_days = np.diff(time.values) / np.timedelta64(1,"D")
            tsmax = float(np.nanmax(step_days)) if len(step_days) else np.nan
            span_days = float((time[-1] - time[0]).days)
        else:
            tsmax = np.nan; span_days = np.nan
        print(f"  profiles: {nprof}, time span days: {span_days:.1f}, max gap days: {tsmax if np.isfinite(tsmax) else np.nan:.1f}")
        print(f"  finite% (after QC/screens): TEMP {frac_finite(t.values):.1f}  PSAL {frac_finite(s.values):.1f}  PRES {frac_finite(pr.values):.1f}  CPHL {frac_finite(chl.values):.1f}")

        long_enough = (
            (np.isfinite(tsmax) and tsmax > 20 and nprof > 48) or
            (np.isfinite(tsmax) and tsmax <= 20 and nprof >= 24)
        )
        if not long_enough:
            print("  STOP: record too short or too gappy for seasonal analysis")
            ds.close(); continue

        # series we build
        rl = [None] * nprof
        pi = np.zeros(nprof)
        ts = np.array([]); ss = np.array([]); chls = np.array([]); mlds = np.array([])
        tdiff_series = []      # near-sfc T - Tf
        tns_series   = []      # near-sfc Temp (averaged)
        sns_series   = []      # near-sfc Salinity (averaged)
        tf_ns_series = []      # near-sfc freezing temp at i0

        # CHL gate counters
        used_shallow = used_deep = no_chl = 0

        for p in range(nprof):
            lonp = float(lon.isel({lon_dim: p}).values)
            latp = float(lat.isel({lat_dim: p}).values)
            pp   = np.asarray(pr.isel(TIME=p).values, dtype=float)
            chlp = np.asarray(chl.isel(TIME=p).values, dtype=float)

            dp = gsw.z_from_p(pp, latp); zpos = -dp  # +depth (m)
            if np.isnan(np.nanmean(dp)):
                dp = gsw.z_from_p(pp, float(lat.mean().values)); zpos = -dp

            tp = np.asarray(t.isel(TIME=p).values, dtype=float)
            sp = np.asarray(s.isel(TIME=p).values, dtype=float)
            spa = gsw.SA_from_SP(sp, pp, lonp, latp)
            if np.isnan(np.nanmean(spa)):
                spa = gsw.SA_from_SP(sp, pp, float(lon.mean().values), float(lat.mean().values))
            ct = gsw.CT_from_t(spa, tp, pp)
            n2, pmid = gsw.Nsquared(spa, ct, pp)
            dmid = gsw.z_from_p(pmid, latp)
            if np.isnan(np.nanmean(dmid)):
                dmid = gsw.z_from_p(pmid, float(lat.mean().values))

            # freezing temp (air-free) on the pressure grid; use near-sfc index
            tfreeze = gsw.t_freezing(spa, 0.0, 0.0)  # p=0 dbar
            try:
                i0 = int(np.nanargmin(pp))
            except Exception:
                i0 = 0

            try:
                mld = dmid[int(np.nanargmax(n2))]
                mlds = np.append(mlds, mld)

                # near-sfc 10 dbar window, fallback 20 dbar by pressure difference
                ddl = dp[i0] - 10.0
                ddi = int(np.nanargmin(np.abs(dp - ddl)))
                lo = min(ddi, i0); hi = max(ddi, i0) + 1
                seg = slice(lo, hi)
                vT = np.nanmean(tp[seg]); vS = np.nanmean(sp[seg])
                if np.isnan(vT) or np.isnan(vS):
                    p0 = pp[i0]; near_idx = np.where((pp - p0) <= 20.0)[0]
                    if near_idx.size:
                        if np.isnan(vT): vT = np.nanmean(tp[near_idx])
                        if np.isnan(vS): vS = np.nanmean(sp[near_idx])
                ts = np.append(ts, vT); ss = np.append(ss, vS)
                tns_series.append(vT); sns_series.append(vS)

                # CHL around MLD with relaxed gate
                valid_idx = np.where(~np.isnan(chlp))[0]
                if valid_idx.size == 0:
                    chls = np.append(chls, np.nan); no_chl += 1
                else:
                    shallow_idx = valid_idx[zpos[valid_idx] <= chl_gate_depth_m]
                    if shallow_idx.size > 0:
                        ic = int(shallow_idx.max()); used_shallow += 1
                    else:
                        deep_idx = valid_idx[zpos[valid_idx] <= deep_gate_depth_m]
                        ic = int((deep_idx if deep_idx.size else valid_idx).max()); used_deep += 1
                    ddic = int(np.nanargmin(np.abs(dp - mld)))
                    start = max(0, min(ddic - bl, ic)); stop = min(len(chlp), max(ddic - bl, ic) + 1)
                    vC = np.nanmean(chlp[start:stop])
                    chls = np.append(chls, vC)
            except Exception:
                mlds = np.append(mlds, np.nan); ts = np.append(ts, np.nan); ss = np.append(ss, np.nan); chls = np.append(chls, np.nan)
                tns_series.append(np.nan); sns_series.append(np.nan)

            # near-sfc freezing temp and T-Tf at i0
            try:
                tf_ns = float(tfreeze[i0])
            except Exception:
                tf_ns = np.nan
            tf_ns_series.append(tf_ns)
            try:
                tdiff = float(vT - tf_ns)
            except Exception:
                tdiff = np.inf
            tdiff_series.append(tdiff)

            # under-ice flag
            if np.isfinite(tdiff) and (tdiff <= rtc):
                rl[p] = 1; pi[p] = 1
            else:
                rl[p] = 0

        # seasonal axes
        week = pd.Index(time).isocalendar().week.to_numpy()
        day  = pd.Index(time).dayofyear.to_numpy()
        year = pd.Index(time).year.to_numpy()
        week = np.where(week <= 13, week + 52, week)
        day  = np.where(day  <= 89,  day  + 365, day)

        # prints
        def _ff(x): return f"{frac_finite(x):.1f}"
        print(f"  built series len: ts={len(ts)} ss={len(ss)} chl={len(chls)}  finite% ts:{_ff(ts)} ss:{_ff(ss)} chl:{_ff(chls)}")
        print(f"  CHL gating counts: shallow<= {int(chl_gate_depth_m)}m: {used_shallow}  deeper<= {int(deep_gate_depth_m)}m: {used_deep}  no-CHL: {no_chl}")
        if used_deep > 0 and used_shallow == 0:
            print(f"  INFO: No shallow (≤{int(chl_gate_depth_m)} m) CHL in this file; used deeper CHL up to {int(deep_gate_depth_m)} m where available.")
        elif used_deep > 0:
            print(f"  INFO: Some profiles lacked shallow CHL; used deeper CHL up to {int(deep_gate_depth_m)} m for {used_deep} profiles.")

        # safe interp + filters
        ts_i   = safe_interp_1d(ts);   ss_i   = safe_interp_1d(ss);   chls_i = safe_interp_1d(chls)
        chlsf = safe_filtfilt(B, A, chls_i); tsf = safe_filtfilt(B, A, ts_i); ssf = safe_filtfilt(B, A, ss_i)
        ssf2  = safe_filtfilt(B2, A2, ss_i)

        # ===== melt onset
        mis = np.array([], dtype=int)
        uice_ct = int(np.sum(pi == 1))
        print(f"  under-ice profiles (T-Tf <= {rtc}): {uice_ct}")
        tv = np.asarray([x for x in tdiff_series if np.isfinite(x)])
        if tv.size:
            q = np.nanpercentile(tv, [0,25,50,75,100])
            print(f"  T-Tf stats (°C): min:{q[0]:.3f} Q1:{q[1]:.3f} med:{q[2]:.3f} Q3:{q[3]:.3f} max:{q[4]:.3f}")
        if uice_ct >= 3:
            dpi = pi[1:] - pi[0:-1]
            mi  = np.where(dpi == -1)[0]
            print(f"  melt candidates (under-ice→open): {len(mi)}")
            keep = []
            for idx in mi:
                nundi = int(np.sum(pi[max(0, idx - 15): idx] == 1))
                if nundi >= 2: keep.append(idx)
            mi = np.array(keep, dtype=int)
            print(f"   after '≥2 prior under-ice' check: {len(mi)}")
            if mi.size > 1:
                dmi = np.abs(mi[1:] - mi[0:-1]); kept2 = 0
                for nm, melts in enumerate(mi):
                    salt = ssf2[melts:melts + 4]
                    dsalt = np.nanmean(salt[1:] - salt[0:-1])
                    cond = (nm < len(dmi) and dmi[nm] > 12 and dsalt < 0.0) or (nm == len(dmi) and dsalt < 0.0)
                    if cond:
                        mis = np.append(mis, melts); kept2 += 1
                print(f"   after salinity-drop & spacing check: {kept2}")
                if mis.size > 1:
                    mis = np.sort(mis.astype(int))
                    dmi_n = np.abs(mis[1:] - mis[0:-1])
                    mis = np.delete(mis, np.where(dmi_n < 12)[0])
            elif mi.size == 1:
                m0 = int(mi[0])
                salt = ssf2[m0:m0+4]
                dsalt = np.nanmean(salt[1:] - salt[0:-1])
                if dsalt < 0.0:
                    mis = np.append(mis, m0)
            print(f"  FINAL melt picks: {len(mis)}")
        else:
            print("  NOTE: fewer than 3 under-ice profiles → melt detection disabled")

        # ===== GI detection
        finite_chls = np.isfinite(chls_i)
        ch_std = np.nanstd(chls_i) if finite_chls.sum() >= 2 else 0.0
        chl_anom = chls_i - (ch_std if np.isfinite(ch_std) else 0.0)

        lmax = lmm(chlsf, np.greater)[0]
        lmin = lmm(chlsf, np.less)[0]
        print(f"  GI extrema counts: lmin={len(lmin)} lmax={len(lmax)}  finite% chl(after interp)={frac_finite(chls_i):.1f}")

        bloomday_f = np.array([]); bloomweek_f = np.array([]); bloomweek_m_f = np.array([])
        blm_inds = np.array([]); mld_a = np.array([]); mld_fg = np.array([]); pgweek = np.array([]); gia_a = np.array([])
        gi_found_segments = 0

        if len(lmin) > 0 and len(lmax) > 0:
            try:
                dlmin = len(chlsf) - lmin[-1]
            except IndexError:
                dlmin, lmin = lmax[0], np.array([0])
            try:
                dlmax = lmax[0]
            except IndexError:
                dlmax, lmax = len(chlsf)-1, np.array([len(chlsf)-1])

            nlmax, nlmin = len(lmax), len(lmin)
            if nlmin > nlmax and dlmin < 6: lmin = np.delete(lmin, -1)
            elif nlmin > nlmax and dlmin >= 6:
                chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe)
            elif nlmin < nlmax and dlmax < 9: lmax = np.delete(lmax, 0)
            elif nlmin < nlmax and dlmax >= 9: lmin = np.concatenate(([0], lmin))
            elif nlmin == nlmax and dlmin >= 6:
                dlmm = lmax - lmin
                if np.sum(dlmm < 0) > 0:
                    chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                    lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe); lmax = np.delete(lmax, 0)
            elif nlmin == nlmax and dlmin < 6:
                lmax = np.delete(lmax, 0); lmin = np.delete(lmin, -1)

            for l in range(len(lmax)):
                imax, imin = int(lmax[l]), int(lmin[l])
                if (imin < imax) and (abs(imax - imin) > 3):
                    chlmm = chls_i[imin:imax]
                    peak_chl = np.append(peak_chl, np.nanmax(chlmm))
                    weekmm = week[imin:imax]; daymm = day[imin:imax]
                    dchl = chlmm[1:] - chlmm[0:-1]
                    dchl = np.where(dchl > 0.0, dchl, np.nan)
                    medchl = np.nanmedian(dchl)
                    blmind = int(np.nanargmax(dchl > medchl))
                    blmind_m = int(np.nanargmax(dchl))
                    chl_fi = int(np.nanargmax(dchl > 0.0))

                    blm_inds = np.append(blm_inds, imin + blmind)
                    bloomweek_f = np.append(bloomweek_f, weekmm[blmind])
                    bloomweek_m_f = np.append(bloomweek_m_f, weekmm[blmind_m])
                    bloomday_f  = np.append(bloomday_f,  daymm[blmind])

                    mldmm = mlds[imin:imax]
                    mld_a = np.append(mld_a, mldmm[blmind])
                    mld_fg = np.append(mld_fg, mldmm[chl_fi])
                    pgweek = np.append(pgweek, weekmm[chl_fi])

                    # alt anomaly pick, saved per-segment
                    try:
                        gia_a = np.append(gia_a, weekmm[np.min(np.where(chl_anom[imin:imax] > 0))])
                    except ValueError:
                        gia_a = np.append(gia_a, np.nan)

                    gi_found_segments += 1

        used_fallback = False
        if gi_found_segments == 0:
            gi_idx = gi_fallback_from_increase(chls_i)
            used_fallback = gi_idx is not None
            print(f"  GI fallback used: {used_fallback}  idx:{gi_idx if gi_idx is not None else 'None'}")
            if gi_idx is not None:
                bloomday_f   = np.append(bloomday_f,  day[gi_idx])
                bloomweek_f  = np.append(bloomweek_f, week[gi_idx])
                bloomweek_m_f = np.append(bloomweek_m_f, week[gi_idx])
                if len(mlds) > gi_idx: mld_a = np.append(mld_a, mlds[gi_idx])
                else:                   mld_a = np.append(mld_a, np.nan)
                pgweek = np.append(pgweek, week[gi_idx])
                blm_inds = np.append(blm_inds, gi_idx)
                gia_week = np.append(gia_week, week[gi_idx])
        print(f"  GI picks: {len(bloomweek_f)}  (segments:{gi_found_segments}, fallback:{used_fallback})")

        # ===== pairing & export =====
        if (mis.size == 0) and gi_only_if_no_ice and (len(bloomweek_f) > 0):
            print("  Exporting GI-only rows (no melt detected)")
            for e in range(len(bloomweek_f)):
                bloomdays  = np.append(bloomdays,  bloomday_f[e])
                bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                bloomyears = np.append(bloomyears, year[0])  # coarse; same-year assumption per segment
                meltweeks  = np.append(meltweeks, np.nan)
                meltdays   = np.append(meltdays,  np.nan)
                lons       = np.append(lons, float(lon.isel({lon_dim: int(blm_inds[e])}).values))
                lats       = np.append(lats, float(lat.isel({lat_dim: int(blm_inds[e])}).values))
                files      = np.append(files, f)
                mld_gi     = np.append(mld_gi, mld_a[e])
                mld_chl_pos = np.append(mld_chl_pos, np.nan)
                timing_pg  = np.append(timing_pg, pgweek[e] if len(pgweek) > e else np.nan)
        else:
            meltday = np.array([day[m] for m in mis.astype(int)])
            pair_ct = 0
            for e in range(min(mis.size, len(bloomday_f))):
                if abs(float(meltday[e]) - float(bloomday_f[e])) <= 90:
                    pair_ct += 1
            print(f"  Potential GI–melt pairs within 90 days: {pair_ct}")

            for e in range(min(len(meltday), len(bloomday_f))):
                dd = abs(meltday[e] - bloomday_f[e])
                if dd <= 90:
                    bloomdays  = np.append(bloomdays,  bloomday_f[e])
                    bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                    bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                    bloomyears = np.append(bloomyears, year[0])
                    meltweeks  = np.append(meltweeks,  int(np.array([week[m] for m in mis.astype(int)])[e]))
                    meltdays   = np.append(meltdays,   meltday[e])
                    idx_for_loc = int(blm_inds[e])
                    lons = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                    lats = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                    files = np.append(files, f)
                    mld_gi = np.append(mld_gi, mld_a[e])
                    mld_chl_pos = np.append(mld_chl_pos, np.nan)
                    timing_pg   = np.append(timing_pg,   pgweek[e] if e < len(pgweek) else np.nan)

        len_ts = np.append(len_ts, len(chls_i))

        # ---- collect diagnostics per file ----
        if collect_diag:
            diag[f] = {
                "path": path,
                "time": time,
                "week": week, "day": day, "year": year,
                "lon": lon.values.astype(float),
                "lat": lat.values.astype(float),
                "ts_raw": ts, "ss_raw": ss, "chl_raw": chls,
                "ts_f": tsf, "ss_f": ssf, "chl_f": chlsf,
                "mld": mlds,                 # negative depth (m)
                "pi": pi.astype(int),        # under-ice flag
                "tdiff": np.asarray(tdiff_series, float),
                "tf_ns": np.asarray(tf_ns_series, float),
                "melt_idx": np.asarray(mis, int),
                "gi_idx": np.asarray(blm_inds, int) if len(blm_inds) else np.array([], dtype=int),
                "notes": {
                    "used_shallow": int(used_shallow),
                    "used_deep": int(used_deep),
                    "no_chl": int(no_chl),
                    "rtc": float(rtc),
                    "chl_gate_depth_m": float(chl_gate_depth_m),
                    "deep_gate_depth_m": float(deep_gate_depth_m),
                }
            }

        ds.close()

    # ---- summary prints ----
    print("\n=== Summary across files ===")
    print("Avg. length of time series:", np.nanmean(len_ts) if len(len_ts) else np.nan)
    print("Avg. peak chlorophyll:",     np.nanmean(peak_chl) if len(peak_chl) else np.nan)
    print("Avg. melt week:",            np.nanmean(melt_week) if len(melt_week) else np.nan)
    print("Avg. chla at GI:",           np.nanmean(chl_sog) if len(chl_sog) else np.nan)
    print("Avg. MLD at GI:",            np.nanmean(mld_gi) if len(mld_gi) else np.nan)
    print("Avg. Growth Rate GI->Melt:", np.nanmean(grth_rate_gm) if len(grth_rate_gm) else np.nan)
    print("Avg. Timing of First Positive Chl:", np.nanmean(timing_pg) if len(timing_pg) else np.nan)
    print("Avg. MLD at First Positive Chl:",    np.nanmean(mld_chl_pos) if len(mld_chl_pos) else np.nan)

    return (
        bloomweeks, bloomdays, bloomweeks_m, bloomyears,
        meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag
    )

# ===== convenience: pad vectors to same length =====
def _pad_to_len(a, L):
    a = np.asarray(a)
    n = a.size
    if n == L: return a
    if n > L:  return a[:L]
    pad = np.full(L - n, np.nan)
    return np.concatenate([a, pad])

# ===== run detector + write out.nc =====
DIR = "/glade/u/home/vyaasv/Floats"

(bloomweeks, bloomdays, bloomweeks_m, bloomyears,
 meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag) = save_meltbloom2(
    rtc=0.3,
    bl=0,
    DIR=DIR,
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
)

all_arrays = {
    "bloomweeks":     bloomweeks,
    "bloomdays":      bloomdays,
    "bloomweeks_max": bloomweeks_m,
    "bloomyears":     bloomyears,
    "meltweeks":      meltweeks,
    "meltdays":       meltdays,
    "mld_gi":         mld_gi,
    "lons":           lons,
    "lats":           lats,
    "files":          files,     # becomes float if padded with NaN; OK
    "gia_week":       gia_week,  # may be shorter; will be padded
}
L = max(len(v) for v in all_arrays.values()) if all_arrays else 0
all_arrays = {k: _pad_to_len(v, L) for k, v in all_arrays.items()}

dsb = xr.Dataset({k: (["e"], v) for k, v in all_arrays.items()})
out_path = os.path.join(DIR, "out.nc")
dsb.to_netcdf(out_path, format="NETCDF4_CLASSIC")
print("Wrote:", out_path)

# ===== diagnostics helpers =====
def list_diag_summary(diag_dict):
    """Quick table of which floats have picks."""
    rows = []
    for f, d in diag_dict.items():
        rows.append((f, os.path.basename(d["path"]), len(d["gi_idx"]), len(d["melt_idx"]),
                     d["notes"]["used_shallow"], d["notes"]["used_deep"]))
    rows = sorted(rows, key=lambda x: x[0])
    print("f | file | GI_picks | Melt_picks | CHL<=100m | CHL<=200m")
    for r in rows:
        print(f"{r[0]:2d} | {r[1]} | {r[2]:8d} | {r[3]:10d} | {r[4]:9d} | {r[5]:9d}")

from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import matplotlib.dates as mdates

def _proxy_lines():
    gi_proxy   = Line2D([0], [0], color="C3", lw=1.4, linestyle="-",  label="GI")
    melt_proxy = Line2D([0], [0], color="C3", lw=1.4, linestyle="--", label="Melt")
    return gi_proxy, melt_proxy

def _add_event_lines(ax, t, gi_idx, melt_idx, color="C3"):
    for i in gi_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="-")
    for i in melt_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="--")

def _apply_year_ticks(ax, label="Year"):
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    ax.set_xlabel(label)

def plot_float_panels(diag_dict, f, title=None):
    """
    Make THREE SEPARATE FIGURES (not subplots), each with its own x-label:
      1) CHL (raw + filtered) with GI (solid) and Melt (dashed) markers
      2) Near-surface T and T_freeze, right axis T − Tf, under-ice shading
      3) Near-surface S (left) and |MLD| (right)
    """
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")

    d = diag_dict[f]
    t = pd.DatetimeIndex(d["time"])
    gi_idx  = np.asarray(d["gi_idx"], int)
    melt_idx = np.asarray(d["melt_idx"], int)
    rtc = float(d["notes"].get("rtc", 0.5))
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"
    plt.rcParams.update({"figure.dpi": 130})

    # ---------- Fig 1: CHL ----------
    fig1 = plt.figure(figsize=(12, 4))
    ax1 = fig1.gca()
    l_chl_f, = ax1.plot(t, d["chl_f"], lw=1.4, label="CHL (filtered)")
    s_chl_r  = ax1.scatter(t, d["chl_raw"], s=12, alpha=0.7, label="CHL (raw)")
    _add_event_lines(ax1, t, gi_idx, melt_idx)
    gi_proxy, melt_proxy = _proxy_lines()

    ax1.set_ylabel("Chlorophyll (a.u.)")
    ax1.set_title(f"Mixed-layer CHL with GI (solid) and melt (dashed) — {title_stub}")
    ax1.grid(alpha=0.3)
    ax1.legend([l_chl_f, s_chl_r, gi_proxy, melt_proxy],
               [h.get_label() for h in [l_chl_f, s_chl_r, gi_proxy, melt_proxy]],
               loc="upper left")
    _apply_year_ticks(ax1)  # x-label on this figure
    fig1.tight_layout()
    plt.show()

    # ---------- Fig 2: Temperature / T_freeze with T − Tf and shading ----------
    fig2 = plt.figure(figsize=(12, 4))
    axT = fig2.gca()
    l_T,  = axT.plot(t, d["ts_f"], lw=1.4, label="T (near-sfc, filtered)")
    l_Tf, = axT.plot(t, d["tf_ns"], lw=1.2, label="T_freeze (near-sfc)")

    tdiff = np.asarray(d["tdiff"], float)
    under = np.where(np.isfinite(tdiff) & (tdiff <= rtc))[0]
    shade_handle = None
    if under.size > 0:
        for i in under:
            t0, t1 = t[i], t[min(i + 1, len(t) - 1)]
            axT.axvspan(t0, t1, alpha=0.15)
        shade_handle = Patch(alpha=0.15, label=f"Under-ice (T−Tf ≤ {rtc:g})")

    axTd = axT.twinx()
    l_dT, = axTd.plot(t, tdiff, lw=1.0, label="T − Tf")

    _add_event_lines(axT, t, gi_idx, melt_idx)

    axT.set_ylabel("Temperature (°C)")
    axTd.set_ylabel("T − Tf (°C)")
    axT.set_title(f"Near-surface T and freezing T with under-ice shading — {title_stub}")
    axT.grid(alpha=0.3)

    left_h, left_l   = axT.get_legend_handles_labels()
    right_h, right_l = axTd.get_legend_handles_labels()
    gi_proxy, melt_proxy = _proxy_lines()
    handles = left_h + right_h + [gi_proxy, melt_proxy]
    labels  = left_l + right_l + [gi_proxy.get_label(), melt_proxy.get_label()]
    if shade_handle is not None:
        handles.append(shade_handle); labels.append(shade_handle.get_label())
    axT.legend(handles, labels, loc="upper left")

    _apply_year_ticks(axT)  # x-label on this figure
    fig2.tight_layout()
    plt.show()

    # ---------- Fig 3: Salinity / |MLD| ----------
    fig3 = plt.figure(figsize=(12, 4))
    axS = fig3.gca()
    l_S,  = axS.plot(t, d["ss_f"], lw=1.4, label="S (near-sfc, filtered)")

    axM = axS.twinx()
    mld_mag = np.abs(np.asarray(d["mld"], float))
    l_MLD, = axM.plot(t, mld_mag, lw=1.0, label="MLD |depth|")

    _add_event_lines(axS, t, gi_idx, melt_idx)

    axS.set_ylabel("Salinity")
    axM.set_ylabel("MLD (m)")
    axS.set_title(f"Near-surface S with mixed-layer depth magnitude — {title_stub}")
    axS.grid(alpha=0.3)

    left_h, left_l   = axS.get_legend_handles_labels()
    right_h, right_l = axM.get_legend_handles_labels()
    gi_proxy, melt_proxy = _proxy_lines()
    axS.legend(left_h + right_h + [gi_proxy, melt_proxy],
               left_l + right_l + [gi_proxy.get_label(), melt_proxy.get_label()],
               loc="upper left")

    _apply_year_ticks(axS)  # x-label on this figure
    fig3.tight_layout()
    plt.show()



# ===== quick plotting suite for out.nc =====
def _arr(ds, name):
    a = np.asarray(ds[name].values, float)
    return a

def _valid_mask(*arrays):
    m = np.ones_like(arrays[0], dtype=bool)
    for a in arrays:
        m &= np.isfinite(a)
    return m

def _to_lon180(lon):
    lon = lon.copy()
    lon = np.where(lon > 180, lon - 360, lon)
    lon = np.where(lon < -180, lon + 360, lon)
    return lon

# ----- run a quick summary + a few figures (A–E). Fig F uses diagnostics -----
list_diag_summary(diag)

OUT_PATH = os.path.join(DIR, "out.nc")
ds_out = xr.open_dataset(OUT_PATH)

bloomweeks = _arr(ds_out, "bloomweeks")
bloomdays  = _arr(ds_out, "bloomdays")
meltweeks  = _arr(ds_out, "meltweeks")
meltdays   = _arr(ds_out, "meltdays")
mld_gi     = _arr(ds_out, "mld_gi")
lons       = _arr(ds_out, "lons")
lats       = _arr(ds_out, "lats")

plt.rcParams.update({"figure.dpi": 130})

# Fig A — Map of GI locations colored by GI week
mA = _valid_mask(bloomweeks, lons, lats)
if np.any(mA):
    lon180 = _to_lon180(lons[mA])
    plt.figure(figsize=(8,4.5))
    sc = plt.scatter(lon180, lats[mA], c=bloomweeks[mA], s=18, cmap="viridis", edgecolor="none")
    plt.colorbar(sc, label="GI week (ISO week shifted)")
    plt.xlabel("Longitude (°)")
    plt.ylabel("Latitude (°)")
    plt.title("GI event locations colored by GI week")
    plt.grid(alpha=0.3)
    plt.show()
else:
    print("Fig A: No valid GI+location data to plot.")

# Fig B — GI week vs Latitude (with linear fit)
mB = _valid_mask(bloomweeks, lats)
xB, yB = lats[mB], bloomweeks[mB]
plt.figure(figsize=(6,4.5))
plt.scatter(xB, yB, s=18, alpha=0.8)
if xB.size >= 2:
    p = np.polyfit(xB, yB, 1)
    xx = np.linspace(np.nanmin(xB), np.nanmax(xB), 100)
    yy = p[0]*xx + p[1]
    plt.plot(xx, yy, lw=2, alpha=0.8)
    plt.text(0.02, 0.98, f"slope={p[0]:.2f} wk/°", transform=plt.gca().transAxes,
             ha="left", va="top")
plt.xlabel("Latitude (°)")
plt.ylabel("GI week")
plt.title("GI timing vs latitude")
plt.grid(alpha=0.3)
plt.show()

# Fig C — GI week vs Melt week (1:1)
mC = _valid_mask(bloomweeks, meltweeks)
xC, yC = meltweeks[mC], bloomweeks[mC]
plt.figure(figsize=(5.2,5.2))
plt.scatter(xC, yC, s=22, alpha=0.85)
if xC.size:
    mn = np.nanmin(np.concatenate([xC, yC]))
    mx = np.nanmax(np.concatenate([xC, yC]))
    plt.plot([mn, mx], [mn, mx], ls="--", lw=1.2)
plt.xlabel("Melt week")
plt.ylabel("GI week")
plt.title("GI vs Melt week (1:1 dashed)")
plt.grid(alpha=0.3)
plt.axis("equal")
plt.show()

# Fig D — Δt = GI − Melt (days) histogram
mD = _valid_mask(bloomdays, meltdays)
if np.any(mD):
    dd = bloomdays[mD] - meltdays[mD]
    plt.figure(figsize=(7,4))
    bins = np.arange(-90, 91, 7)  # weekly bins over ±90 days
    plt.hist(dd, bins=bins, alpha=0.85)
    plt.axvline(0, color="k", ls="--", lw=1)
    plt.xlabel("GI − Melt (days)")
    plt.ylabel("Count")
    plt.title("Distribution of GI–Melt timing difference")
    plt.grid(axis="y", alpha=0.3)
    plt.show()
else:
    print("Fig D: No paired GI–melt events to plot.")

# Fig E — MLD at GI histogram (abs depth, meters)
mE = np.isfinite(mld_gi)
if np.any(mE):
    mld_mag = np.abs(mld_gi[mE])
    plt.figure(figsize=(7,4))
    plt.hist(mld_mag, bins=30, alpha=0.85)
    plt.xlabel("MLD at GI (m)")
    plt.ylabel("Count")
    plt.title("Distribution of mixed-layer depth at GI")
    plt.grid(axis="y", alpha=0.3)
    plt.show()
else:
    print("Fig E: No finite MLD-at-GI values to plot.")

# Fig F — Per-float 'Fig-2-style' time-series panels (requires `diag`)
have_diag = isinstance(diag, dict) and len(diag) > 0
if have_diag:
    picks = [(f, len(d["gi_idx"]) + len(d["melt_idx"])) for f, d in diag.items()]
    if picks:
        best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
        print(f"Plotting Fig F for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
        plot_float_panels(diag, f=best_f)
    else:
        print("Fig F: diag exists but no picks found.")
else:
    print("Fig F: Skipped (no `diag` in memory).")

# Delete out.nc if it exists
```

NO3 included below

```python
# ===== NCS (start of nitrate consumption) — add-on to existing pipeline =====
# This cell:
#   1) Re-runs melt+GI detector but ALSO extracts an upper-ocean nitrate series (same gate as CHL)
#   2) Detects NCS per float: first sustained decline after a local NO3 maximum, heading to the next minimum
#   3) Stores NO3 raw/filtered + NCS indices in `diag`
#   4) Adds a nitrate figure with NCS dash-dot lines to the per-float panels
#
# Notes:
# - It reuses the helpers: safe_interp_1d, safe_filtfilt, pick_var, good_qc_mask, get_time_index, gi_fallback_from_increase
# - CHL gate logic stays identical; nitrate uses the SAME indices as CHL averaging for fairness
# - No changes to earlier figures A–E; only the per-float time-series gets an extra nitrate figure

import numpy as np
import xarray as xr
import pandas as pd
import gsw
from scipy.signal import argrelextrema as lmm
import scipy.signal as signal
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import os

# ---------- nitrate helpers ----------
def _resolve_no3(ds):
    for base in ("NTAW", "NITRATE", "NO3"):
        adj = base + "_ADJUSTED"
        if adj in ds.variables:
            return ds[adj], base
        if base in ds.variables:
            return ds[base], base
    return None, None

def _no3_units_kind(units_attr):
    if not units_attr:
        return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    # treat mmol/m3, umol/l, etc. as per-volume
    return "mmol_m3"

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lon, lat):
    SA = gsw.SA_from_SP(psal, pres, lon, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    # µmol/kg * kg/m^3 = µmol/m^3 -> mmol/m^3
    return (no3_umolkg * rho) / 1000.0

# ---------- NCS detection (works on a single smoothed series) ----------
def _find_ncs_indices(no3f):
    ncs_idx = []
    if np.isfinite(no3f).sum() < 5:
        return np.array(ncs_idx, dtype=int)
    try:
        n_lmax = lmm(no3f, np.greater)[0]
        n_lmin = lmm(no3f, np.less)[0]
    except Exception:
        n_lmax, n_lmin = np.array([], int), np.array([], int)

    def _first_consumption_index(f, imax, imin):
        if not (0 <= imax < len(f)) or not (0 <= imin < len(f)) or imin <= imax:
            return None
        seg = f[imax:imin+1]
        d   = np.diff(seg)
        R   = seg[0] - seg[-1]
        pre = f[max(0, imax-5):imax+1]
        sigma = np.nanstd(pre) if np.isfinite(pre).sum() >= 2 else np.nanstd(seg)
        thresh = max(0.03 * (R if np.isfinite(R) else 0.0), 0.5 * (sigma if np.isfinite(sigma) else 0.0))
        for j in range(len(d)-1):
            if (d[j] < 0.0) and (d[j+1] < 0.0) and (seg[j+1] <= seg[0] - thresh):
                return imax + (j+1)
        # fallback: first negative step after imax
        for j in range(len(d)):
            if d[j] < 0.0:
                return imax + j + 1
        return None

    # pair each max with the next min
    for imax in n_lmax:
        nxt = n_lmin[n_lmin > imax]
        if nxt.size == 0:
            continue
        imin = int(nxt[0])
        k = _first_consumption_index(no3f, int(imax), int(imin))
        if k is not None:
            ncs_idx.append(int(k))

    # global fallback if nothing found
    if not ncs_idx:
        gmax = int(np.nanargmax(no3f))
        d = np.diff(no3f[gmax:])
        for j in range(len(d)):
            if d[j] < 0.0:
                ncs_idx.append(gmax + j + 1)
                break
    return np.array(ncs_idx, dtype=int)

# ---------- Detector with NO3 add-on ----------
def save_meltbloom2_with_ncs(
    rtc=0.5,
    bl=0,
    DIR="",
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
):
    # reuse your filter design
    N = 1; Wn, Wn2 = 0.1, 0.2
    B, A  = signal.butter(N, Wn,  output="ba")
    B2, A2 = signal.butter(N, Wn2, output="ba")

    flist = glob.glob(os.path.join(DIR, "*.nc"))
    print("No. of files:", len(flist))

    # outputs (same names as yours)
    bloomdays  = np.array([]); bloomweeks = np.array([]); bloomweeks_m = np.array([])
    bloomyears = np.array([]); meltweeks  = np.array([]); meltdays     = np.array([])
    lons = np.array([]); lats = np.array([]); files = np.array([]); mld_gi = np.array([])
    mld_chl_pos = np.array([]); timing_pg = np.array([]); grth_rate_gm = np.array([])
    peak_chl = np.array([]); len_ts = np.array([]); melt_week = np.array([]); chl_sog = np.array([])
    gia_week = np.array([])

    diag = {}

    for f, path in enumerate(flist):
        print(f"\n=== File {f} { os.path.basename(path) } ===")
        ds = xr.open_dataset(path)

        # basic vars (reuse your helpers)
        if not all(v in ds.variables for v in ("TIME","LONGITUDE","LATITUDE")):
            print("  STOP: missing TIME/LONGITUDE/LATITUDE"); ds.close(); continue
        t  = pick_var(ds, "TEMP")
        s  = pick_var(ds, "PSAL")
        pr = pick_var(ds, "PRES")
        if ("CPHL" in ds.variables) or ("CPHL_ADJUSTED" in ds.variables):
            chl = pick_var(ds, "CPHL")
        else:
            print("  STOP: no CPHL/CPHL_ADJUSTED"); ds.close(); continue

        # nitrate (optional)
        no3, no3_base = _resolve_no3(ds)
        have_no3 = no3 is not None

        lon = ds["LONGITUDE"]; lat = ds["LATITUDE"]
        time = get_time_index(ds)

        lon_dim = lon.dims[0]; lat_dim = lat.dims[0]
        if "TIME" not in t.sizes:
            print("  STOP: TEMP has no TIME dim"); ds.close(); continue
        nprof = int(t.sizes["TIME"])
        if lon.sizes[lon_dim] != nprof or lat.sizes[lat_dim] != nprof:
            print("  STOP: lon/lat length mismatch vs TIME"); ds.close(); continue

        # QC (reuse your QC mask)
        for da, base in [(t,"TEMP"),(s,"PSAL"),(pr,"PRES"),(chl,"CPHL")]:
            m = good_qc_mask(ds, base, qc_ok=qc_ok)
            if m is not None and da.dims == m.dims:
                da[:] = da.where(m)
        s   = s.where(s > 28.0)
        chl = chl.where(chl >= 0)

        if have_no3:
            m_no3 = good_qc_mask(ds, no3_base, qc_ok=("1","2","8","0"))
            if m_no3 is not None and no3.dims == m_no3.dims:
                no3 = no3.where(m_no3)

        # timing diag
        if nprof >= 2:
            step_days = np.diff(time.values) / np.timedelta64(1,"D")
            tsmax = float(np.nanmax(step_days)) if len(step_days) else np.nan
            span_days = float((time[-1] - time[0]).days)
        else:
            tsmax = np.nan; span_days = np.nan
        print(f"  profiles: {nprof}, time span days: {span_days:.1f}, max gap days: {tsmax if np.isfinite(tsmax) else np.nan:.1f}")

        # quick series & containers
        rl = [None] * nprof
        pi = np.zeros(nprof)
        ts = np.array([]); ss = np.array([]); chls = np.array([]); mlds = np.array([])
        tdiff_series = []; tns_series = []; sns_series = []; tf_ns_series = []
        no3s = np.array([])

        used_shallow = used_deep = no_chl = 0

        # per profile loop (same as yours; nitrate uses same indices as CHL gate)
        for p in range(nprof):
            lonp = float(lon.isel({lon_dim: p}).values)
            latp = float(lat.isel({lat_dim: p}).values)
            pp   = np.asarray(pr.isel(TIME=p).values, dtype=float)
            chlp = np.asarray(chl.isel(TIME=p).values, dtype=float)

            dp = gsw.z_from_p(pp, latp); zpos = -dp  # +depth
            if np.isnan(np.nanmean(dp)):
                dp = gsw.z_from_p(pp, float(lat.mean().values)); zpos = -dp

            tp = np.asarray(t.isel(TIME=p).values, dtype=float)
            sp = np.asarray(s.isel(TIME=p).values, dtype=float)
            spa = gsw.SA_from_SP(sp, pp, lonp, latp)
            if np.isnan(np.nanmean(spa)):
                spa = gsw.SA_from_SP(sp, pp, float(lon.mean().values), float(lat.mean().values))
            ct = gsw.CT_from_t(spa, tp, pp)
            n2, pmid = gsw.Nsquared(spa, ct, pp)
            dmid = gsw.z_from_p(pmid, latp)
            if np.isnan(np.nanmean(dmid)):
                dmid = gsw.z_from_p(pmid, float(lat.mean().values))

            tfreeze = gsw.t_freezing(spa, 0.0, 0.0)  # p=0 dbar
            try:
                i0 = int(np.nanargmin(pp))
            except Exception:
                i0 = 0

            try:
                mld = dmid[int(np.nanargmax(n2))]
                mlds = np.append(mlds, mld)

                # near-sfc windows (as in your code)
                ddl = dp[i0] - 10.0
                ddi = int(np.nanargmin(np.abs(dp - ddl)))
                lo = min(ddi, i0); hi = max(ddi, i0) + 1
                seg = slice(lo, hi)
                vT = np.nanmean(tp[seg]); vS = np.nanmean(sp[seg])
                if np.isnan(vT) or np.isnan(vS):
                    p0 = pp[i0]; near_idx = np.where((pp - p0) <= 20.0)[0]
                    if near_idx.size:
                        if np.isnan(vT): vT = np.nanmean(tp[near_idx])
                        if np.isnan(vS): vS = np.nanmean(sp[near_idx])
                ts = np.append(ts, vT); ss = np.append(ss, vS)
                tns_series.append(vT); sns_series.append(vS)

                # CHL gate (unchanged)
                valid_idx = np.where(~np.isnan(chlp))[0]
                if valid_idx.size == 0:
                    chls = np.append(chls, np.nan); no_chl += 1
                    # nitrate stays NaN when CHL gate cannot be formed
                    no3s = np.append(no3s, np.nan)
                else:
                    shallow_idx = valid_idx[zpos[valid_idx] <= chl_gate_depth_m]
                    if shallow_idx.size > 0:
                        ic = int(shallow_idx.max()); used_shallow += 1
                    else:
                        deep_idx = valid_idx[zpos[valid_idx] <= deep_gate_depth_m]
                        ic = int((deep_idx if deep_idx.size else valid_idx).max()); used_deep += 1
                    ddic = int(np.nanargmin(np.abs(dp - mld)))
                    start = max(0, min(ddic - bl, ic)); stop = min(len(chlp), max(ddic - bl, ic) + 1)
                    vC = np.nanmean(chlp[start:stop])
                    chls = np.append(chls, vC)

                    # nitrate using SAME start:stop window
                    if have_no3:
                        no3p = np.asarray(no3.isel(TIME=p).values, dtype=float)
                        uk = _no3_units_kind(no3.attrs.get("units",""))
                        if uk == "umol_kg":
                            no3v = _to_mmol_m3_from_umol_kg(no3p, pp, sp, tp, lonp, latp)
                        else:
                            no3v = no3p
                        vN = np.nanmean(no3v[start:stop])
                        no3s = np.append(no3s, vN)
                    else:
                        no3s = np.append(no3s, np.nan)

            except Exception:
                # keep parity with your code in the exception path
                mlds = np.append(mlds, np.nan); ts = np.append(ts, np.nan); ss = np.append(ss, np.nan); chls = np.append(chls, np.nan)
                tns_series.append(np.nan); sns_series.append(np.nan); no3s = np.append(no3s, np.nan)

            # T - Tf at near-sfc
            try: tf_ns = float(tfreeze[i0])
            except Exception: tf_ns = np.nan
            tf_ns_series.append(tf_ns)
            try: tdiff = float(vT - tf_ns)
            except Exception: tdiff = np.inf
            tdiff_series.append(tdiff)

            # under-ice flag
            if np.isfinite(tdiff) and (tdiff <= rtc):
                rl[p] = 1; pi[p] = 1
            else:
                rl[p] = 0

        # seasonal axes / prints exactly as yours
        week = pd.Index(time).isocalendar().week.to_numpy()
        day  = pd.Index(time).dayofyear.to_numpy()
        year = pd.Index(time).year.to_numpy()
        week = np.where(week <= 13, week + 52, week)
        day  = np.where(day  <= 89,  day  + 365, day)

        def _ff(x): return f"{100.0 * np.isfinite(np.asarray(x)).sum() / (len(x) if len(x) else 1):.1f}"
        print(f"  built series len: ts={len(ts)} ss={len(ss)} chl={len(chls)} no3={len(no3s)}  finite% ts:{_ff(ts)} ss:{_ff(ss)} chl:{_ff(chls)} no3:{_ff(no3s)}")
        print(f"  CHL gating counts: shallow<= {int(chl_gate_depth_m)}m: {used_shallow}  deeper<= {int(deep_gate_depth_m)}m: {used_deep}  no-CHL: {no_chl}")

        # interp + filters (reuse your filters)
        ts_i   = safe_interp_1d(ts);   ss_i   = safe_interp_1d(ss);   chls_i = safe_interp_1d(chls);  no3_i = safe_interp_1d(no3s)
        chlsf = safe_filtfilt(B, A, chls_i); tsf = safe_filtfilt(B, A, ts_i); ssf = safe_filtfilt(B, A, ss_i); no3f = safe_filtfilt(B, A, no3_i)
        ssf2  = safe_filtfilt(B2, A2, ss_i)

        # ---- melt detection (unchanged)
        mis = np.array([], dtype=int)
        uice_ct = int(np.sum(pi == 1))
        if uice_ct >= 3:
            dpi = pi[1:] - pi[0:-1]
            mi  = np.where(dpi == -1)[0]
            keep = []
            for idx in mi:
                nundi = int(np.sum(pi[max(0, idx - 15): idx] == 1))
                if nundi >= 2: keep.append(idx)
            mi = np.array(keep, dtype=int)
            if mi.size > 1:
                dmi = np.abs(mi[1:] - mi[0:-1])
                kept2 = 0
                for nm, melts in enumerate(mi):
                    salt = ssf2[melts:melts + 4]
                    dsalt = np.nanmean(salt[1:] - salt[0:-1])
                    cond = (nm < len(dmi) and dmi[nm] > 12 and dsalt < 0.0) or (nm == len(dmi) and dsalt < 0.0)
                    if cond:
                        mis = np.append(mis, melts); kept2 += 1
                if mis.size > 1:
                    mis = np.sort(mis.astype(int))
                    dmi_n = np.abs(mis[1:] - mis[0:-1])
                    mis = np.delete(mis, np.where(dmi_n < 12)[0])
            elif mi.size == 1:
                m0 = int(mi[0])
                salt = ssf2[m0:m0+4]
                dsalt = np.nanmean(salt[1:] - salt[0:-1])
                if dsalt < 0.0:
                    mis = np.append(mis, m0)

        # ---- GI detection (unchanged core)
        finite_chls = np.isfinite(chls_i)
        ch_std = np.nanstd(chls_i) if finite_chls.sum() >= 2 else 0.0
        chl_anom = chls_i - (ch_std if np.isfinite(ch_std) else 0.0)
        lmax = lmm(chlsf, np.greater)[0]
        lmin = lmm(chlsf, np.less)[0]

        bloomday_f = np.array([]); bloomweek_f = np.array([]); bloomweek_m_f = np.array([])
        blm_inds = np.array([]); mld_a = np.array([]); mld_fg = np.array([]); pgweek = np.array([]); gia_a = np.array([])
        gi_found_segments = 0

        if len(lmin) > 0 and len(lmax) > 0:
            try:    dlmin = len(chlsf) - lmin[-1]
            except: dlmin, lmin = lmax[0], np.array([0])
            try:    dlmax = lmax[0]
            except: dlmax, lmax = len(chlsf)-1, np.array([len(chlsf)-1])

            nlmax, nlmin = len(lmax), len(lmin)
            if nlmin > nlmax and dlmin < 6: lmin = np.delete(lmin, -1)
            elif nlmin > nlmax and dlmin >= 6:
                chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe)
            elif nlmin < nlmax and dlmax < 9: lmax = np.delete(lmax, 0)
            elif nlmin < nlmax and dlmax >= 9: lmin = np.concatenate(([0], lmin))
            elif nlmin == nlmax and dlmin >= 6:
                dlmm = lmax - lmin
                if np.sum(dlmm < 0) > 0:
                    chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                    lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe); lmax = np.delete(lmax, 0)
            elif nlmin == nlmax and dlmin < 6:
                lmax = np.delete(lmax, 0); lmin = np.delete(lmin, -1)

            for l in range(len(lmax)):
                imax, imin = int(lmax[l]), int(lmin[l])
                if (imin < imax) and (abs(imax - imin) > 3):
                    chlmm = chls_i[imin:imax]
                    peak_chl = np.append(peak_chl, np.nanmax(chlmm))
                    weekmm = week[imin:imax]; daymm = day[imin:imax]
                    dchl = chlmm[1:] - chlmm[0:-1]
                    dchl = np.where(dchl > 0.0, dchl, np.nan)
                    medchl = np.nanmedian(dchl)
                    blmind = int(np.nanargmax(dchl > medchl))
                    blmind_m = int(np.nanargmax(dchl))
                    chl_fi = int(np.nanargmax(dchl > 0.0))

                    blm_inds = np.append(blm_inds, imin + blmind)
                    bloomweek_f = np.append(bloomweek_f, weekmm[blmind])
                    bloomweek_m_f = np.append(bloomweek_m_f, weekmm[blmind_m])
                    bloomday_f  = np.append(bloomday_f,  daymm[blmind])

                    mldmm = mlds[imin:imax]
                    mld_a = np.append(mld_a, mldmm[blmind])
                    mld_fg = np.append(mld_fg, mldmm[chl_fi])
                    pgweek = np.append(pgweek, weekmm[chl_fi])

                    try:
                        gia_a = np.append(gia_a, weekmm[np.min(np.where(chl_anom[imin:imax] > 0))])
                    except ValueError:
                        gia_a = np.append(gia_a, np.nan)

                    gi_found_segments += 1

        used_fallback = False
        if gi_found_segments == 0:
            gi_idx = gi_fallback_from_increase(chls_i)
            used_fallback = gi_idx is not None
            if gi_idx is not None:
                bloomday_f   = np.append(bloomday_f,  day[gi_idx])
                bloomweek_f  = np.append(bloomweek_f, week[gi_idx])
                bloomweek_m_f = np.append(bloomweek_m_f, week[gi_idx])
                if len(mlds) > gi_idx: mld_a = np.append(mld_a, mlds[gi_idx])
                else:                   mld_a = np.append(mld_a, np.nan)
                pgweek = np.append(pgweek, week[gi_idx])
                blm_inds = np.append(blm_inds, gi_idx)
                gia_week = np.append(gia_week, week[gi_idx])

        # ---- NCS detection on no3f
        ncs_idx = _find_ncs_indices(no3f) if have_no3 else np.array([], dtype=int)
        print(f"  NCS picks: {len(ncs_idx)}")

        # ---- pairing & export (unchanged)
        if (mis.size == 0) and gi_only_if_no_ice and (len(bloomweek_f) > 0):
            for e in range(len(bloomweek_f)):
                bloomdays  = np.append(bloomdays,  bloomday_f[e])
                bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                bloomyears = np.append(bloomyears, year[0])
                meltweeks  = np.append(meltweeks, np.nan)
                meltdays   = np.append(meltdays,  np.nan)
                idx_for_loc = int(blm_inds[e])
                lons  = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                lats  = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                files = np.append(files, f)
                mld_gi = np.append(mld_gi, mld_a[e])
                mld_chl_pos = np.append(mld_chl_pos, np.nan)
                timing_pg  = np.append(timing_pg, pgweek[e] if len(pgweek) > e else np.nan)
        else:
            meltday = np.array([day[m] for m in mis.astype(int)])
            for e in range(min(len(meltday), len(bloomday_f))):
                dd = abs(meltday[e] - bloomday_f[e])
                if dd <= 90:
                    bloomdays  = np.append(bloomdays,  bloomday_f[e])
                    bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                    bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                    bloomyears = np.append(bloomyears, year[0])
                    meltweeks  = np.append(meltweeks,  int(np.array([week[m] for m in mis.astype(int)])[e]))
                    meltdays   = np.append(meltdays,   meltday[e])
                    idx_for_loc = int(blm_inds[e])
                    lons = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                    lats = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                    files = np.append(files, f)
                    mld_gi = np.append(mld_gi, mld_a[e])
                    mld_chl_pos = np.append(mld_chl_pos, np.nan)
                    timing_pg   = np.append(timing_pg,   pgweek[e] if e < len(pgweek) else np.nan)

        len_ts = np.append(len_ts, len(chls_i))

        # ---- diagnostics (same as yours + NO3/NCS) ----
        if collect_diag:
            diag[f] = {
                "path": path,
                "time": time,
                "week": week, "day": day, "year": year,
                "lon": lon.values.astype(float),
                "lat": lat.values.astype(float),
                "ts_raw": ts, "ss_raw": ss, "chl_raw": chls, "no3_raw": no3s,
                "ts_f": tsf, "ss_f": ssf, "chl_f": chlsf, "no3_f": no3f,
                "mld": mlds,
                "pi": pi.astype(int),
                "tdiff": np.asarray(tdiff_series, float),
                "tf_ns": np.asarray(tf_ns_series, float),
                "melt_idx": np.asarray(mis, int),
                "gi_idx": np.asarray(blm_inds, int) if len(blm_inds) else np.array([], dtype=int),
                "ncs_idx": np.asarray(ncs_idx, int),
                "notes": {
                    "used_shallow": int(used_shallow),
                    "used_deep": int(used_deep),
                    "no_chl": int(no_chl),
                    "rtc": float(rtc),
                    "chl_gate_depth_m": float(chl_gate_depth_m),
                    "deep_gate_depth_m": float(deep_gate_depth_m),
                }
            }

        ds.close()

    print("\n=== Summary across files (same metrics as before) ===")
    print("Avg. length of time series:", np.nanmean(len_ts) if len(len_ts) else np.nan)
    print("Avg. peak chlorophyll:",     np.nanmean(peak_chl) if len(peak_chl) else np.nan)
    print("Avg. MLD at GI:",            np.nanmean(mld_gi) if len(mld_gi) else np.nan)

    return (
        bloomweeks, bloomdays, bloomweeks_m, bloomyears,
        meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag
    )

# ---------- Plot: add a nitrate panel with NCS lines (dash-dot) ----------
def _proxy_lines_ncs():
    return Line2D([0], [0], color="C0", lw=1.4, linestyle="-.", label="NCS")

def _add_event_lines_ncs(ax, t, ncs_idx, color="C0"):
    for i in np.asarray(ncs_idx, int):
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="-.")

def plot_float_panels_nitrate(diag_dict, f, title=None):
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag_dict[f]
    t = pd.DatetimeIndex(d["time"])
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []), int)
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"

    if ("no3_f" not in d) or ("no3_raw" not in d) or (len(d["no3_raw"]) == 0):
        print("NO3 not available for this float.")
        return

    plt.rcParams.update({"figure.dpi": 130})
    figN = plt.figure(figsize=(12, 4))
    axN = figN.gca()
    l_no3_f, = axN.plot(t, d["no3_f"], lw=1.4, label="NO3 (filtered)")
    s_no3_r  = axN.scatter(t, d["no3_raw"], s=12, alpha=0.7, label="NO3 (raw)")

    # melt dashed, NCS dash-dot
    from matplotlib.dates import DateFormatter, YearLocator
    for i in melt_idx:
        if 0 <= i < len(t): axN.axvline(t[i], color="C3", lw=1.2, ls="--")
    _add_event_lines_ncs(axN, t, ncs_idx)

    ncs_proxy = _proxy_lines_ncs()
    melt_proxy = Line2D([0], [0], color="C3", lw=1.2, linestyle="--", label="Melt")

    axN.set_ylabel("NO3 (mmol m$^{-3}$)")
    axN.set_title(f"Upper-ocean nitrate with NCS (dash-dot) and melt (dashed) — {title_stub}")
    axN.grid(alpha=0.3)
    axN.legend([l_no3_f, s_no3_r, ncs_proxy, melt_proxy],
               [h.get_label() for h in [l_no3_f, s_no3_r, ncs_proxy, melt_proxy]],
               loc="upper left")
    axN.xaxis.set_major_locator(mdates.YearLocator())
    axN.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    axN.set_xlabel("Year")
    figN.tight_layout()
    plt.show()

# ---------- RUN (same call signature as yours; only function name changed) ----------
DIR = "/glade/u/home/vyaasv/Floats"

(bloomweeks, bloomdays, bloomweeks_m, bloomyears,
 meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag) = save_meltbloom2_with_ncs(
    rtc=0.3,
    bl=0,
    DIR=DIR,
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
)

# Quick per-float nitrate panel on the richest-picked float (like your Fig F logic)
if isinstance(diag, dict) and len(diag) > 0:
    picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", []))) for f, d in diag.items()]
    if picks:
        best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
        print(f"Plotting nitrate+NCS panel for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
        plot_float_panels_nitrate(diag, f=best_f)
    else:
        print("Nitrate panel: diag exists but no picks found.")
else:
    print("Nitrate panel: Skipped (no `diag`).")
```

```python
# ===== NCS (start of nitrate consumption) — robust to missing PSAL/SP/etc. =====
# Key changes vs previous version:
# - _pick_multi() tries multiple aliases (e.g., salinity: PSAL/SP/SAL/SALINITY)
# - Safe guards when S/T/P missing: MLD, Tf, and density conversion become NaN rather than error
# - Gate windows check start<stop before np.nanmean
# - NO3 unit conversion to mmol/m^3 only when S/T/P exist; else NO3 left NaN for that profile

import numpy as np
import xarray as xr
import pandas as pd
import gsw
from scipy.signal import argrelextrema as lmm
import scipy.signal as signal
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import os
import warnings
warnings.filterwarnings("ignore", message="Mean of empty slice")

# ---------- helpers (reuse your existing ones if already defined above) ----------
def _pick_multi(ds, bases):
    """Return (DataArray, picked_base) trying *_ADJUSTED then base for a list of candidate names."""
    for b in bases:
        adj = f"{b}_ADJUSTED"
        if adj in ds.variables:
            return ds[adj], b
    for b in bases:
        if b in ds.variables:
            return ds[b], b
    return None, None

def _resolve_no3(ds):
    return _pick_multi(ds, ("NTAW","NITRATE","NO3"))

def _no3_units_kind(units_attr):
    if not units_attr: return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    return "mmol_m3"

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lon, lat):
    SA = gsw.SA_from_SP(psal, pres, lon, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    return (no3_umolkg * rho) / 1000.0  # -> mmol/m^3

def _find_ncs_indices(no3f):
    ncs_idx = []
    if np.isfinite(no3f).sum() < 5:
        return np.array(ncs_idx, dtype=int)
    try:
        n_lmax = lmm(no3f, np.greater)[0]
        n_lmin = lmm(no3f, np.less)[0]
    except Exception:
        n_lmax, n_lmin = np.array([], int), np.array([], int)

    def _first_consumption_index(f, imax, imin):
        if not (0 <= imax < len(f)) or not (0 <= imin < len(f)) or imin <= imax:
            return None
        seg = f[imax:imin+1]
        d   = np.diff(seg)
        R   = seg[0] - seg[-1]
        pre = f[max(0, imax-5):imax+1]
        sigma = np.nanstd(pre) if np.isfinite(pre).sum() >= 2 else np.nanstd(seg)
        thresh = max(0.03 * (R if np.isfinite(R) else 0.0), 0.5 * (sigma if np.isfinite(sigma) else 0.0))
        for j in range(len(d)-1):
            if (d[j] < 0.0) and (d[j+1] < 0.0) and (seg[j+1] <= seg[0] - thresh):
                return imax + (j+1)
        for j in range(len(d)):
            if d[j] < 0.0:
                return imax + j + 1
        return None

    for imax in n_lmax:
        nxt = n_lmin[n_lmin > imax]
        if nxt.size == 0: continue
        imin = int(nxt[0])
        k = _first_consumption_index(no3f, int(imax), int(imin))
        if k is not None: ncs_idx.append(int(k))

    if not ncs_idx:
        gmax = int(np.nanargmax(no3f))
        d = np.diff(no3f[gmax:])
        for j in range(len(d)):
            if d[j] < 0.0:
                ncs_idx.append(gmax + j + 1); break
    return np.array(ncs_idx, dtype=int)

# Expect these helpers from your earlier cell:
# - safe_interp_1d, safe_filtfilt, good_qc_mask, get_time_index, gi_fallback_from_increase
# - If any are missing in your runtime, paste them above.

def save_meltbloom2_with_ncs(
    rtc=0.5,
    bl=0,
    DIR="",
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
):
    N = 1; Wn, Wn2 = 0.1, 0.2
    B, A  = signal.butter(N, Wn,  output="ba")
    B2, A2 = signal.butter(N, Wn2, output="ba")

    flist = glob.glob(os.path.join(DIR, "*.nc"))
    print("No. of files:", len(flist))

    bloomdays  = np.array([]); bloomweeks = np.array([]); bloomweeks_m = np.array([])
    bloomyears = np.array([]); meltweeks  = np.array([]); meltdays     = np.array([])
    lons = np.array([]); lats = np.array([]); files = np.array([]); mld_gi = np.array([])
    mld_chl_pos = np.array([]); timing_pg = np.array([]); grth_rate_gm = np.array([])
    peak_chl = np.array([]); len_ts = np.array([]); melt_week = np.array([]); chl_sog = np.array([])
    gia_week = np.array([])

    diag = {}

    for f, path in enumerate(flist):
        print(f"\n=== File {f} { os.path.basename(path) } ===")
        ds = xr.open_dataset(path)

        # --- variable resolution (ROBUST) ---
        if not all(v in ds.variables for v in ("TIME","LONGITUDE","LATITUDE")):
            print("  STOP: missing TIME/LONGITUDE/LATITUDE"); ds.close(); continue

        t, t_base   = _pick_multi(ds, ("TEMP","TEMP_CTD","TEMPERATURE"))
        s, s_base   = _pick_multi(ds, ("PSAL","SP","SAL","SALINITY"))
        pr, p_base  = _pick_multi(ds, ("PRES","PRESSURE"))
        chl, chl_b  = _pick_multi(ds, ("CPHL","CHLA","CHL","CHLA_NN","CPHL_NN"))

        if chl is None:
            print("  STOP: no CPHL/CHL-like variable"); ds.close(); continue

        no3, no3_base = _resolve_no3(ds)
        have_no3 = no3 is not None

        lon = ds["LONGITUDE"]; lat = ds["LATITUDE"]
        time = get_time_index(ds)

        lon_dim = lon.dims[0]; lat_dim = lat.dims[0]
        if (t is None) or ("TIME" not in t.sizes):
            print("  STOP: TEMP has no TIME dim or missing"); ds.close(); continue
        nprof = int(t.sizes["TIME"])
        if lon.sizes[lon_dim] != nprof or lat.sizes[lat_dim] != nprof:
            print("  STOP: lon/lat length mismatch vs TIME"); ds.close(); continue

        # --- QC screen (only when var exists) ---
        for da, base in [(t,t_base),(s,s_base),(pr,p_base),(chl,chl_b)]:
            if da is None: continue
            m = good_qc_mask(ds, base, qc_ok=qc_ok)
            if m is not None and da.dims == m.dims:
                da[:] = da.where(m)
        if s is not None: s = s.where(s > 28.0)
        chl = chl.where(chl >= 0)

        if have_no3:
            m_no3 = good_qc_mask(ds, no3_base, qc_ok=("1","2","8","0"))
            if m_no3 is not None and no3.dims == m_no3.dims:
                no3 = no3.where(m_no3)

        # --- timing diag ---
        if nprof >= 2:
            step_days = np.diff(time.values) / np.timedelta64(1,"D")
            tsmax = float(np.nanmax(step_days)) if len(step_days) else np.nan
            span_days = float((time[-1] - time[0]).days)
        else:
            tsmax = np.nan; span_days = np.nan
        print(f"  profiles: {nprof}, time span days: {span_days:.1f}, max gap days: {tsmax if np.isfinite(tsmax) else np.nan:.1f}")

        rl = [None] * nprof
        pi = np.zeros(nprof)
        ts = np.array([]); ss = np.array([]); chls = np.array([]); mlds = np.array([])
        tdiff_series = []; tf_ns_series = []
        no3s = np.array([])

        used_shallow = used_deep = no_chl = 0
        density_warned = False

        for p in range(nprof):
            lonp = float(lon.isel({lon_dim: p}).values)
            latp = float(lat.isel({lat_dim: p}).values)
            # pressure and CHL slice
            pp   = np.asarray(pr.isel(TIME=p).values, dtype=float) if pr is not None else np.array([], dtype=float)
            chlp = np.asarray(chl.isel(TIME=p).values, dtype=float)

            # depth coordinate
            if pp.size:
                dp = gsw.z_from_p(pp, latp); zpos = -dp
            else:
                # no pressure -> cannot form depth; mark NaN and make gates minimal
                dp = np.full_like(chlp, np.nan); zpos = np.full_like(chlp, np.nan)

            # temp/sal for near-surface window
            tp = np.asarray(t.isel(TIME=p).values, dtype=float)
            sp = np.asarray(s.isel(TIME=p).values, dtype=float) if s is not None else None

            # Compute SA/CT/N^2 only if S, T, P exist
            have_STP = (sp is not None) and (sp.size == tp.size == pp.size) and (pp.size > 1)
            if have_STP:
                try:
                    spa = gsw.SA_from_SP(sp, pp, lonp, latp)
                    ct  = gsw.CT_from_t(spa, tp, pp)
                    n2, pmid = gsw.Nsquared(spa, ct, pp)
                    dmid = gsw.z_from_p(pmid, latp)
                except Exception:
                    spa = None; n2 = None; dmid = np.full_like(pp, np.nan)
            else:
                spa = None; n2 = None; dmid = np.full_like(pp, np.nan)

            # freezing temperature @ p=0; requires SA; if missing, use NaN
            if spa is not None and spa.size:
                tfreeze = gsw.t_freezing(spa, 0.0, 0.0)
            else:
                tfreeze = np.full_like(tp, np.nan)

            # near-surface window (allow missing P by falling back to a tiny window around top index)
            try:
                i0 = int(np.nanargmin(pp)) if pp.size else 0
            except Exception:
                i0 = 0

            # MLD from N^2 max if available
            if (n2 is not None) and np.isfinite(n2).any():
                try:
                    mld = dmid[int(np.nanargmax(n2))]
                except Exception:
                    mld = np.nan
            else:
                mld = np.nan
            mlds = np.append(mlds, mld)

            # build near-sfc averages
            if pp.size:
                ddl = dp[i0] - 10.0
                ddi = int(np.nanargmin(np.abs(dp - ddl)))
                lo = min(ddi, i0); hi = max(ddi, i0) + 1
                seg = slice(lo, hi)
                vT = np.nanmean(tp[seg])
                vS = np.nanmean(sp[seg]) if sp is not None else np.nan
                if (np.isnan(vT) or (sp is not None and np.isnan(vS))) and pp.size:
                    p0 = pp[i0]
                    near_idx = np.where((pp - p0) <= 20.0)[0]
                    if near_idx.size:
                        if np.isnan(vT): vT = np.nanmean(tp[near_idx])
                        if sp is not None and np.isnan(vS): vS = np.nanmean(sp[near_idx])
            else:
                # no pressure -> take first 1–2 levels
                vT = np.nanmean(tp[:2]) if tp.size else np.nan
                vS = np.nanmean(sp[:2]) if (sp is not None and sp.size) else np.nan

            ts = np.append(ts, vT); ss = np.append(ss, vS)

            # CHL gating (unchanged logic, with checks)
            valid_idx = np.where(~np.isnan(chlp))[0]
            if valid_idx.size == 0:
                chls = np.append(chls, np.nan); no_chl += 1
                no3s = np.append(no3s, np.nan)
            else:
                shallow_idx = valid_idx[zpos[valid_idx] <= chl_gate_depth_m] if np.isfinite(zpos).any() else np.array([], int)
                if shallow_idx.size > 0:
                    ic = int(shallow_idx.max()); used_shallow += 1
                else:
                    deep_idx = valid_idx[zpos[valid_idx] <= deep_gate_depth_m] if np.isfinite(zpos).any() else valid_idx
                    ic = int((deep_idx if deep_idx.size else valid_idx).max()); used_deep += (1 if deep_idx.size and shallow_idx.size==0 else 0)

                # ddic: nearest to MLD if finite, else use ic
                if np.isfinite(mld) and pp.size:
                    ddic = int(np.nanargmin(np.abs(dp - mld)))
                else:
                    ddic = ic

                start = max(0, min(ddic - bl, ic))
                stop  = min(len(chlp), max(ddic - bl, ic) + 1)
                if stop <= start:
                    vC = np.nan
                else:
                    vC = np.nanmean(chlp[start:stop])
                chls = np.append(chls, vC)

                # nitrate using SAME start:stop window (only if we have any NO3)
                if have_no3:
                    no3p = np.asarray(no3.isel(TIME=p).values, dtype=float)
                    uk = _no3_units_kind(no3.attrs.get("units",""))
                    if uk == "umol_kg":
                        if have_STP:
                            try:
                                no3v = _to_mmol_m3_from_umol_kg(no3p, pp, (sp if sp is not None else np.full_like(pp, np.nan)), tp, lonp, latp)
                            except Exception:
                                no3v = np.full_like(no3p, np.nan)
                        else:
                            if not density_warned:
                                print("  NOTE: NO3 in µmol/kg but S/T/P incomplete → leaving NO3 as NaN for these profiles.")
                                density_warned = True
                            no3v = np.full_like(no3p, np.nan)
                    else:
                        no3v = no3p
                    vN = np.nan if (stop <= start) else np.nanmean(no3v[start:stop])
                    no3s = np.append(no3s, vN)
                else:
                    no3s = np.append(no3s, np.nan)

            # T - Tf at near-sfc (Tf may be NaN if S missing)
            tf_ns = float(tfreeze[i0]) if (np.isfinite(i0) and i0 < len(tfreeze)) else np.nan
            tf_ns_series.append(tf_ns)
            tdiff = vT - tf_ns if (np.isfinite(vT) and np.isfinite(tf_ns)) else np.inf
            tdiff_series.append(tdiff)

            # under-ice flag (disabled if no S → tf_ns NaN → tdiff inf)
            if np.isfinite(tdiff) and (tdiff <= rtc):
                rl[p] = 1; pi[p] = 1
            else:
                rl[p] = 0

        # seasonal axes
        week = pd.Index(time).isocalendar().week.to_numpy()
        day  = pd.Index(time).dayofyear.to_numpy()
        year = pd.Index(time).year.to_numpy()
        week = np.where(week <= 13, week + 52, week)
        day  = np.where(day  <= 89,  day  + 365, day)

        def _ff(x):
            x = np.asarray(x);
            return f"{100.0 * np.isfinite(x).sum() / (len(x) if len(x) else 1):.1f}"
        print(f"  built series len: ts={len(ts)} ss={len(ss)} chl={len(chls)} no3={len(no3s)}  finite% ts:{_ff(ts)} ss:{_ff(ss)} chl:{_ff(chls)} no3:{_ff(no3s)}")
        print(f"  CHL gating counts: shallow<= {int(chl_gate_depth_m)}m: {used_shallow}  deeper<= {int(deep_gate_depth_m)}m: {used_deep}  no-CHL: {no_chl}")

        # filters
        ts_i   = safe_interp_1d(ts);   ss_i   = safe_interp_1d(ss);   chls_i = safe_interp_1d(chls);  no3_i = safe_interp_1d(no3s)
        chlsf = safe_filtfilt(B, A, chls_i); tsf = safe_filtfilt(B, A, ts_i); ssf = safe_filtfilt(B, A, ss_i); no3f = safe_filtfilt(B, A, no3_i)
        ssf2  = safe_filtfilt(B2, A2, ss_i)

        # melt detection (unchanged logic, but may be disabled if tf_ns all NaN)
        mis = np.array([], dtype=int)
        uice_ct = int(np.sum(pi == 1))
        if uice_ct >= 3:
            dpi = pi[1:] - pi[0:-1]
            mi  = np.where(dpi == -1)[0]
            keep = []
            for idx in mi:
                nundi = int(np.sum(pi[max(0, idx - 15): idx] == 1))
                if nundi >= 2: keep.append(idx)
            mi = np.array(keep, dtype=int)
            if mi.size > 1:
                dmi = np.abs(mi[1:] - mi[0:-1])
                kept2 = 0
                for nm, melts in enumerate(mi):
                    salt = ssf2[melts:melts + 4]
                    dsalt = np.nanmean(salt[1:] - salt[0:-1])
                    cond = (nm < len(dmi) and dmi[nm] > 12 and dsalt < 0.0) or (nm == len(dmi) and dsalt < 0.0)
                    if cond:
                        mis = np.append(mis, melts); kept2 += 1
                if mis.size > 1:
                    mis = np.sort(mis.astype(int))
                    dmi_n = np.abs(mis[1:] - mis[0:-1])
                    mis = np.delete(mis, np.where(dmi_n < 12)[0])
            elif mi.size == 1:
                m0 = int(mi[0])
                salt = ssf2[m0:m0+4]
                dsalt = np.nanmean(salt[1:] - salt[0:-1])
                if dsalt < 0.0:
                    mis = np.append(mis, m0)

        # GI detection (unchanged)
        finite_chls = np.isfinite(chls_i)
        ch_std = np.nanstd(chls_i) if finite_chls.sum() >= 2 else 0.0
        chl_anom = chls_i - (ch_std if np.isfinite(ch_std) else 0.0)
        lmax = lmm(chlsf, np.greater)[0]
        lmin = lmm(chlsf, np.less)[0]

        bloomday_f = np.array([]); bloomweek_f = np.array([]); bloomweek_m_f = np.array([])
        blm_inds = np.array([]); mld_a = np.array([]); pgweek = np.array([]); gia_a = np.array([])
        gi_found_segments = 0

        if len(lmin) > 0 and len(lmax) > 0:
            try:    dlmin = len(chlsf) - lmin[-1]
            except: dlmin, lmin = lmax[0], np.array([0])
            try:    dlmax = lmax[0]
            except: dlmax, lmax = len(chlsf)-1, np.array([len(chlsf)-1])

            nlmax, nlmin = len(lmax), len(lmin)
            if nlmin > nlmax and dlmin < 6: lmin = np.delete(lmin, -1)
            elif nlmin > nlmax and dlmin >= 6:
                chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe)
            elif nlmin < nlmax and dlmax < 9: lmax = np.delete(lmax, 0)
            elif nlmin < nlmax and dlmax >= 9: lmin = np.concatenate(([0], lmin))
            elif nlmin == nlmax and dlmin >= 6:
                dlmm = lmax - lmin
                if np.sum(dlmm < 0) > 0:
                    chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                    lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe); lmax = np.delete(lmax, 0)
            elif nlmin == nlmax and dlmin < 6:
                lmax = np.delete(lmax, 0); lmin = np.delete(lmin, -1)

            for l in range(len(lmax)):
                imax, imin = int(lmax[l]), int(lmin[l])
                if (imin < imax) and (abs(imax - imin) > 3):
                    chlmm = chls_i[imin:imax]
                    peak_chl = np.append(peak_chl, np.nanmax(chlmm))
                    weekmm = week[imin:imax]; daymm = day[imin:imax]
                    dchl = chlmm[1:] - chlmm[0:-1]
                    dchl = np.where(dchl > 0.0, dchl, np.nan)
                    medchl = np.nanmedian(dchl)
                    blmind = int(np.nanargmax(dchl > medchl))
                    chl_fi = int(np.nanargmax(dchl > 0.0))

                    blm_inds = np.append(blm_inds, imin + blmind)
                    bloomweek_f = np.append(bloomweek_f, weekmm[blmind])
                    bloomweek_m_f = np.append(bloomweek_m_f, weekmm[int(np.nanargmax(dchl))])
                    bloomday_f  = np.append(bloomday_f,  daymm[blmind])

                    mld_a = np.append(mld_a, mlds[imin:imax][blmind])
                    pgweek = np.append(pgweek, weekmm[chl_fi])

                    try:
                        gia_a = np.append(gia_a, weekmm[np.min(np.where(chl_anom[imin:imax] > 0))])
                    except ValueError:
                        gia_a = np.append(gia_a, np.nan)

                    gi_found_segments += 1

        if gi_found_segments == 0:
            gi_idx = gi_fallback_from_increase(chls_i)
            if gi_idx is not None:
                bloomday_f   = np.append(bloomday_f,  day[gi_idx])
                bloomweek_f  = np.append(bloomweek_f, week[gi_idx])
                bloomweek_m_f = np.append(bloomweek_m_f, week[gi_idx])
                mld_a = np.append(mld_a, mlds[gi_idx] if len(mlds) > gi_idx else np.nan)
                pgweek = np.append(pgweek, week[gi_idx])
                blm_inds = np.append(blm_inds, gi_idx)
                gia_week = np.append(gia_week, week[gi_idx])

        # ---- NCS (NO3) ----
        ncs_idx = _find_ncs_indices(no3f) if have_no3 else np.array([], dtype=int)
        print(f"  NCS picks: {len(ncs_idx)}")

        # ---- Export rows (unchanged pairing logic) ----
        if (mis.size == 0) and gi_only_if_no_ice and (len(bloomweek_f) > 0):
            for e in range(len(bloomweek_f)):
                bloomdays  = np.append(bloomdays,  bloomday_f[e])
                bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                bloomyears = np.append(bloomyears, year[0])
                meltweeks  = np.append(meltweeks, np.nan)
                meltdays   = np.append(meltdays,  np.nan)
                idx_for_loc = int(blm_inds[e])
                lons  = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                lats  = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                files = np.append(files, f)
                mld_gi = np.append(mld_gi, mld_a[e])
                mld_chl_pos = np.append(mld_chl_pos, np.nan)
                timing_pg  = np.append(timing_pg, pgweek[e] if len(pgweek) > e else np.nan)
        else:
            meltday = np.array([day[m] for m in mis.astype(int)])
            for e in range(min(len(meltday), len(bloomday_f))):
                dd = abs(meltday[e] - bloomday_f[e])
                if dd <= 90:
                    bloomdays  = np.append(bloomdays,  bloomday_f[e])
                    bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                    bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                    bloomyears = np.append(bloomyears, year[0])
                    meltweeks  = np.append(meltweeks,  int(np.array([week[m] for m in mis.astype(int)])[e]))
                    meltdays   = np.append(meltdays,   meltday[e])
                    idx_for_loc = int(blm_inds[e])
                    lons = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                    lats = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                    files = np.append(files, f)
                    mld_gi = np.append(mld_gi, mld_a[e])
                    mld_chl_pos = np.append(mld_chl_pos, np.nan)
                    timing_pg   = np.append(timing_pg,   pgweek[e] if e < len(pgweek) else np.nan)

        len_ts = np.append(len_ts, len(chls_i))

        # ---- diagnostics (add NO3/NCS) ----
        if collect_diag:
            diag[f] = {
                "path": path,
                "time": time,
                "week": week, "day": day, "year": year,
                "lon": lon.values.astype(float),
                "lat": lat.values.astype(float),
                "ts_raw": ts, "ss_raw": ss, "chl_raw": chls, "no3_raw": no3s,
                "ts_f": tsf, "ss_f": ssf, "chl_f": chlsf, "no3_f": no3f,
                "mld": mlds,
                "pi": pi.astype(int),
                "tdiff": np.asarray(tdiff_series, float),
                "tf_ns": np.asarray(tf_ns_series, float),
                "melt_idx": np.asarray(mis, int),
                "gi_idx": np.asarray(blm_inds, int) if len(blm_inds) else np.array([], dtype=int),
                "ncs_idx": np.asarray(ncs_idx, int),
                "notes": {
                    "used_shallow": int(used_shallow),
                    "used_deep": int(used_deep),
                    "no_chl": int(no_chl),
                    "rtc": float(rtc),
                    "chl_gate_depth_m": float(chl_gate_depth_m),
                    "deep_gate_depth_m": float(deep_gate_depth_m),
                    "t_base": t_base, "s_base": s_base, "p_base": p_base,
                }
            }

        ds.close()

    print("\n=== Summary across files (same metrics as before) ===")
    print("Avg. length of time series:", np.nanmean(len_ts) if len(len_ts) else np.nan)
    print("Avg. peak chlorophyll:",     np.nanmean(peak_chl) if len(peak_chl) else np.nan)
    print("Avg. MLD at GI:",            np.nanmean(mld_gi) if len(mld_gi) else np.nan)

    return (
        bloomweeks, bloomdays, bloomweeks_m, bloomyears,
        meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag
    )

# ---------- nitrate panel (unchanged from your version) ----------
def _proxy_lines_ncs():
    return Line2D([0], [0], color="C0", lw=1.4, linestyle="-.", label="NCS")

def _add_event_lines_ncs(ax, t, ncs_idx, color="C0"):
    for i in np.asarray(ncs_idx, int):
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="-.")

def plot_float_panels_nitrate(diag_dict, f, title=None):
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag_dict[f]
    t = pd.DatetimeIndex(d["time"])
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []), int)
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"

    if ("no3_f" not in d) or ("no3_raw" not in d) or (len(d["no3_raw"]) == 0):
        print("NO3 not available for this float.")
        return

    plt.rcParams.update({"figure.dpi": 130})
    figN = plt.figure(figsize=(12, 4))
    axN = figN.gca()
    l_no3_f, = axN.plot(t, d["no3_f"], lw=1.4, label="NO3 (filtered)")
    s_no3_r  = axN.scatter(t, d["no3_raw"], s=12, alpha=0.7, label="NO3 (raw)")

    for i in melt_idx:
        if 0 <= i < len(t): axN.axvline(t[i], color="C3", lw=1.2, ls="--")
    _add_event_lines_ncs(axN, t, ncs_idx)

    ncs_proxy = _proxy_lines_ncs()
    melt_proxy = Line2D([0], [0], color="C3", lw=1.2, linestyle="--", label="Melt")

    axN.set_ylabel("NO3 (mmol m$^{-3}$)")
    axN.set_title(f"Upper-ocean nitrate with NCS (dash-dot) and melt (dashed) — {title_stub}")
    axN.grid(alpha=0.3)
    axN.legend([l_no3_f, s_no3_r, ncs_proxy, melt_proxy],
               [h.get_label() for h in [l_no3_f, s_no3_r, ncs_proxy, melt_proxy]],
               loc="upper left")
    import matplotlib.dates as mdates
    axN.xaxis.set_major_locator(mdates.YearLocator())
    axN.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    axN.set_xlabel("Year")
    figN.tight_layout()
    plt.show()

# ---------- RUN (point to the directory you want to test) ----------
DIR = "/glade/u/home/vyaasv/NO3_Floats"  # <-- change freely

(bloomweeks, bloomdays, bloomweeks_m, bloomyears,
 meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag) = save_meltbloom2_with_ncs(
    rtc=0.3,
    bl=0,
    DIR=DIR,
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
)

# Optional quick nitrate panel on a "rich" float
if isinstance(diag, dict) and len(diag) > 0:
    picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", []))) for f, d in diag.items()]
    if picks:
        best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
        print(f"Plotting nitrate+NCS panel for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
        plot_float_panels_nitrate(diag, f=best_f)
    else:
        print("Nitrate panel: diag exists but no picks found.")
else:
    print("Nitrate panel: Skipped (no `diag`).")
```

```python
# === NCS winter-only plot + driver ===
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.lines import Line2D
import os

# ---- Configure which months count as "winter" (default Aug–Jan) ----
WINTER_MONTHS = (8, 9, 10, 11, 12, 1)

def _proxy_lines_ncs():
    return Line2D([0], [0], color="C0", lw=1.4, linestyle="-.", label=f"NCS")

def _add_event_lines_ncs(ax, t, ncs_idx, months=WINTER_MONTHS, color="C0"):
    months_set = set(int(m) for m in months)
    T = pd.DatetimeIndex(t)
    for i in np.asarray(ncs_idx, dtype=int):
        if 0 <= i < len(T) and int(T[i].month) in months_set:
            ax.axvline(T[i], color=color, lw=1.2, ls="-.")

def plot_float_panels_nitrate(diag_dict, f, title=None):
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag_dict[f]
    t = pd.DatetimeIndex(d["time"])
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []), int)
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"

    if ("no3_f" not in d) or ("no3_raw" not in d) or (len(d["no3_raw"]) == 0):
        print("NO3 not available for this float.")
        return

    plt.rcParams.update({"figure.dpi": 130})
    figN = plt.figure(figsize=(12, 4))
    axN = figN.gca()
    l_no3_f, = axN.plot(t, d["no3_f"], lw=1.4, label="NO3 (filtered)")
    s_no3_r  = axN.scatter(t, d["no3_raw"], s=12, alpha=0.7, label="NO3 (raw)")

    # Melt (dashed) & NCS (dash-dot only in WINTER_MONTHS)
    for i in melt_idx:
        if 0 <= i < len(t): axN.axvline(t[i], color="C3", lw=1.2, ls="--")
    _add_event_lines_ncs(axN, t, ncs_idx, months=WINTER_MONTHS)

    ncs_proxy  = _proxy_lines_ncs()
    melt_proxy = Line2D([0], [0], color="C3", lw=1.2, linestyle="--", label="Melt")

    axN.set_ylabel("NO3 (mmol m$^{-3}$)")
    axN.set_title(f"Upper-ocean nitrate with winter-only NCS (dash-dot) — {title_stub}")
    axN.grid(alpha=0.3)
    axN.legend([l_no3_f, s_no3_r, ncs_proxy, melt_proxy],
               [h.get_label() for h in [l_no3_f, s_no3_r, ncs_proxy, melt_proxy]],
               loc="upper left")
    axN.xaxis.set_major_locator(mdates.YearLocator())
    axN.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    axN.set_xlabel("Year")
    figN.tight_layout()
    plt.show()


# ---- DRIVER: pick a float and plot immediately ----
def _ensure_diag_and_plot(diag_obj=None, run_builder_if_missing=True, builder_kwargs=None):
    """
    If diag is provided and valid, plot the 'richest' float.
    Otherwise (optionally) call save_meltbloom2_with_ncs(**builder_kwargs) to build diag, then plot.
    """
    global diag
    if isinstance(diag_obj, dict) and len(diag_obj) > 0:
        diag = diag_obj  # use provided
    elif run_builder_if_missing:
        # Fallback to your builder if available in this notebook
        if "save_meltbloom2_with_ncs" not in globals():
            raise RuntimeError("No 'diag' found and save_meltbloom2_with_ncs(...) is not defined in this notebook.")
        kw = dict(
            rtc=0.3, bl=0,
            DIR=builder_kwargs.get("DIR", "/glade/u/home/vyaasv/NO3_Floats") if builder_kwargs else "/glade/u/home/vyaasv/NO3_Floats",
            qc_ok=("0","1","2","8"),
            gi_only_if_no_ice=True,
            debug=True,
            chl_gate_depth_m=100.0,
            deep_gate_depth_m=200.0,
            collect_diag=True
        )
        print("Building diag via save_meltbloom2_with_ncs(...) for plotting...")
        (_bw, _bd, _bwm, _by, _mw, _md, _lo, _la, _fi, _mld, _gia, diag) = save_meltbloom2_with_ncs(**kw)
    else:
        raise RuntimeError("No 'diag' provided and builder execution disabled.")

    # Choose the float with most picks (GI + Melt + NCS)
    picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", [])))
             for f, d in diag.items()]
    if not picks:
        print("No floats with picks to plot.")
        return
    best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
    print(f"Plotting nitrate panel (winter-only NCS) for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
    plot_float_panels_nitrate(diag, f=best_f)

# Call the driver:
# - If you already have a `diag` from earlier in the notebook, this will use it.
# - Otherwise it will run the builder with the default DIR (adjust below if needed).
_ensure_diag_and_plot(
    diag_obj=globals().get("diag", None),
    run_builder_if_missing=True,
    builder_kwargs={"DIR": "/glade/u/home/vyaasv/NO3_Floats"}  # <— change DIR here if you want
)
```

```python
# === NCS summary plots A–D (winter-only) + optional CHL panel overlay with NCS ===
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# Configure winter window (Aug–Jan)
WINTER_MONTHS = (8, 9, 10, 11, 12, 1)

# Helpers (reuse if present)
def _to_lon180(lon):
    lon = np.asarray(lon, float).copy()
    lon = np.where(lon > 180, lon - 360, lon)
    lon = np.where(lon < -180, lon + 360, lon)
    return lon

def _valid_mask(*arrays):
    m = np.ones_like(np.asarray(arrays[0], float), dtype=bool)
    for a in arrays:
        m &= np.isfinite(np.asarray(a, float))
    return m

def _collect_ncs_events(diag, months=WINTER_MONTHS, max_pair_days=90):
    """
    Walks diag and collects per-event NCS timing, location, and nearest melt timing.
    Returns dict of numpy arrays: ncsweeks, ncsdays, years, lons, lats, meltweeks, meltdays
    Only events whose datetime month is in `months` are kept.
    """
    months_set = set(int(m) for m in months)
    ncs_w = []; ncs_d = []; yrs = []; LON = []; LAT = []
    melt_w_at = []; melt_d_at = []

    for f, d in diag.items():
        # required fields
        if not {"time", "ncs_idx", "week", "day", "year", "lon", "lat"}.issubset(d.keys()):
            continue
        t = pd.DatetimeIndex(d["time"])
        ncs_idx = np.asarray(d.get("ncs_idx", []), dtype=int)
        melt_idx = np.asarray(d.get("melt_idx", []), dtype=int)

        # precompute seasonal coordinates
        week = np.asarray(d["week"], dtype=float)
        day  = np.asarray(d["day"],  dtype=float)
        year = np.asarray(d["year"], dtype=float)
        lon  = np.asarray(d["lon"],  dtype=float)
        lat  = np.asarray(d["lat"],  dtype=float)

        for i in ncs_idx:
            if not (0 <= i < len(t)):
                continue
            if int(t[i].month) not in months_set:
                continue

            # basic event info
            ncs_w.append(float(week[i]))
            ncs_d.append(float(day[i]))
            yrs.append(float(year[i]))
            LON.append(float(lon[i] if i < len(lon) else np.nan))
            LAT.append(float(lat[i] if i < len(lat) else np.nan))

            # pair with nearest melt within ±max_pair_days
            if melt_idx.size:
                # choose melt with minimum |Δday|
                mdays = np.asarray([day[j] for j in melt_idx if 0 <= j < len(day)], dtype=float)
                if mdays.size:
                    diffs = np.abs(mdays - day[i])
                    j = int(np.nanargmin(diffs))
                    if np.isfinite(diffs[j]) and diffs[j] <= max_pair_days:
                        melt_d_at.append(mdays[j])
                        # melt week at that same melt index
                        mweeks = np.asarray([week[jj] for jj in melt_idx if 0 <= jj < len(week)], dtype=float)
                        melt_w_at.append(mweeks[j])
                    else:
                        melt_d_at.append(np.nan); melt_w_at.append(np.nan)
                else:
                    melt_d_at.append(np.nan); melt_w_at.append(np.nan)
            else:
                melt_d_at.append(np.nan); melt_w_at.append(np.nan)

    out = dict(
        ncsweeks=np.asarray(ncs_w, float),
        ncsdays =np.asarray(ncs_d, float),
        years   =np.asarray(yrs, float),
        lons    =np.asarray(LON, float),
        lats    =np.asarray(LAT, float),
        meltweeks=np.asarray(melt_w_at, float),
        meltdays =np.asarray(melt_d_at, float),
    )
    return out

def _linear_fit_scatter(ax, x, y, xlabel, ylabel, title):
    ax.scatter(x, y, s=18, alpha=0.85)
    if np.isfinite(x).sum() >= 2 and np.isfinite(y).sum() >= 2:
        m = np.isfinite(x) & np.isfinite(y)
        p = np.polyfit(x[m], y[m], 1)
        xx = np.linspace(np.nanmin(x[m]), np.nanmax(x[m]), 100)
        yy = p[0]*xx + p[1]
        ax.plot(xx, yy, lw=2, alpha=0.9)
        ax.text(0.02, 0.98, f"slope={p[0]:.2f}", transform=ax.transAxes, ha="left", va="top")
    ax.set_xlabel(xlabel); ax.set_ylabel(ylabel); ax.set_title(title); ax.grid(alpha=0.3)

def plot_chl_panel_with_ncs(diag, f, title=None):
    """
    Re-creates your CHL figure (filtered + raw) and adds:
      - GI vertical lines (solid)
      - Melt vertical lines (dashed)
      - NCS vertical lines (dash-dot, winter-only)
    Uses only what is already stored in `diag`.
    """
    if f not in diag:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag[f]
    if not {"time","chl_f","chl_raw"}.issubset(d.keys()):
        print("CHL series not available in diag for this float.")
        return
    t = pd.DatetimeIndex(d["time"])
    gi_idx   = np.asarray(d.get("gi_idx", []), int)
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []), int)
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"

    plt.rcParams.update({"figure.dpi": 130})
    fig = plt.figure(figsize=(12, 4))
    ax = fig.gca()
    l_chl_f, = ax.plot(t, d["chl_f"], lw=1.4, label="CHL (filtered)")
    s_chl_r  = ax.scatter(t, d["chl_raw"], s=12, alpha=0.7, label="CHL (raw)")

    # GI solid, Melt dashed
    for i in gi_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color="C3", lw=1.2, ls="-")
    for i in melt_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color="C3", lw=1.2, ls="--")

    # NCS winter-only dash-dot
    months_set = set(int(m) for m in WINTER_MONTHS)
    for i in ncs_idx:
        if 0 <= i < len(t) and int(t[i].month) in months_set:
            ax.axvline(t[i], color="C0", lw=1.2, ls="-.")

    from matplotlib.lines import Line2D
    gi_proxy   = Line2D([0],[0], color="C3", lw=1.2, ls="-",  label="GI")
    melt_proxy = Line2D([0],[0], color="C3", lw=1.2, ls="--", label="Melt")
    ncs_proxy  = Line2D([0],[0], color="C0", lw=1.2, ls="-.", label=f"NCS")

    ax.set_ylabel("Chlorophyll (a.u.)")
    ax.set_title(f"CHL with GI (solid), melt (dashed), winter-only NCS (dash-dot) — {title_stub}")
    ax.grid(alpha=0.3)
    ax.legend([l_chl_f, s_chl_r, gi_proxy, melt_proxy, ncs_proxy],
              [h.get_label() for h in [l_chl_f, s_chl_r, gi_proxy, melt_proxy, ncs_proxy]],
              loc="upper left")
    ax.xaxis.set_major_locator(mdates.YearLocator()); ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    ax.set_xlabel("Year")
    fig.tight_layout()
    plt.show()

# ===== Build and plot =====
if 'diag' not in globals() or not isinstance(diag, dict) or len(diag) == 0:
    raise RuntimeError("This cell expects a populated `diag` from your GI+melt(+NCS) run.")

evt = _collect_ncs_events(diag, months=WINTER_MONTHS, max_pair_days=90)

plt.rcParams.update({"figure.dpi": 130})

# Aₙ — Map of NCS locations colored by NCS week
mA = _valid_mask(evt["ncsweeks"], evt["lons"], evt["lats"])
if np.any(mA):
    lon180 = _to_lon180(evt["lons"][mA])
    plt.figure(figsize=(8,4.5))
    sc = plt.scatter(lon180, evt["lats"][mA], c=evt["ncsweeks"][mA], s=18, cmap="viridis", edgecolor="none")
    plt.colorbar(sc, label="NCS week (ISO week shifted)")
    plt.xlabel("Longitude (°)"); plt.ylabel("Latitude (°)")
    plt.title(f"NCS event locations (winter months {WINTER_MONTHS})")
    plt.grid(alpha=0.3); plt.show()
else:
    print("Aₙ: No valid NCS+location data to plot (winter-only).")

# Bₙ — NCS week vs Latitude
mB = _valid_mask(evt["lats"], evt["ncsweeks"])
if np.any(mB):
    fig = plt.figure(figsize=(6,4.5)); ax = fig.gca()
    _linear_fit_scatter(ax, evt["lats"][mB], evt["ncsweeks"][mB],
                        "Latitude (°)", "NCS week", "NCS timing vs latitude (winter-only)")
    plt.show()
else:
    print("Bₙ: No valid NCS week vs latitude data (winter-only).")

# Cₙ — NCS week vs Melt week (1:1)
mC = _valid_mask(evt["ncsweeks"], evt["meltweeks"])
xC, yC = evt["meltweeks"][mC], evt["ncsweeks"][mC]
if np.any(mC):
    fig = plt.figure(figsize=(5.2,5.2)); ax = fig.gca()
    ax.scatter(xC, yC, s=22, alpha=0.85)
    if xC.size:
        mn = np.nanmin(np.concatenate([xC, yC])); mx = np.nanmax(np.concatenate([xC, yC]))
        ax.plot([mn, mx], [mn, mx], ls="--", lw=1.2)
    ax.set_xlabel("Melt week"); ax.set_ylabel("NCS week")
    ax.set_title("NCS vs Melt week (winter-only NCS)"); ax.grid(alpha=0.3); ax.axis("equal")
    plt.show()
else:
    print("Cₙ: No paired NCS–melt weeks within ±90 days (winter-only NCS).")

# Dₙ — Δt = NCS − Melt (days) histogram
mD = _valid_mask(evt["ncsdays"], evt["meltdays"])
if np.any(mD):
    dd = evt["ncsdays"][mD] - evt["meltdays"][mD]
    plt.figure(figsize=(7,4))
    bins = np.arange(-90, 91, 7)  # weekly bins over ±90 days
    plt.hist(dd, bins=bins, alpha=0.85)
    plt.axvline(0, color="k", ls="--", lw=1)
    plt.xlabel("NCS − Melt (days)"); plt.ylabel("Count")
    plt.title("Distribution of NCS–Melt timing difference (winter-only NCS)")
    plt.grid(axis="y", alpha=0.3); plt.show()
else:
    print("Dₙ: No paired NCS–melt events to plot (winter-only NCS).")

# Optional: pick a float and overlay NCS on the CHL panel
# Choose the float with the most combined picks
picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", []))) for f, d in diag.items()]
if picks:
    best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
    print(f"Overlaying NCS on CHL panel for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
    plot_chl_panel_with_ncs(diag, f=best_f)
else:
    print("CHL+NCS panel: no floats with picks found.")
```

```python
# ==========================================
# Task 2 — GI (this season) vs next-season Max 30d-smoothed CHL
# Uses GI indices from other save_meltbloom2 (diag[f]["gi_idx"])
# Season = Jul→Jun. Discard first season if start after July.
# ==========================================

import numpy as np, pandas as pd, matplotlib.pyplot as plt, os

SEASON_START_MONTH = 7
DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY = True  # matches your earlier rule

def _season_year(dt):
    return dt.year if dt.month >= SEASON_START_MONTH else dt.year - 1

def _season_bounds(sy):
    """Return [Jul 1 sy, Jun 30 sy+1] as timestamps (inclusive)."""
    start = pd.Timestamp(year=int(sy),   month=7, day=1)
    end   = pd.Timestamp(year=int(sy+1), month=6, day=30, hour=23, minute=59, second=59)
    return start, end

def _season_doy(dt):
    start = pd.Timestamp(year=_season_year(dt), month=7, day=1)
    return int((dt.normalize() - start).days) + 1

def _smooth_30d_daily(times, values):
    """Daily time series with centered 30-day mean, like your smooth_30d."""
    t = pd.to_datetime(times)
    s = pd.Series(np.asarray(values, float), index=t).sort_index()
    s = s[~s.index.isna()]
    if s.empty:
        return s
    s = s.groupby(s.index.normalize()).mean()
    s = s.asfreq("1D")
    s = s.interpolate(method="time", limit_direction="both")
    return s.rolling("30D", center=True, min_periods=1).mean()

# ------------- build GI→NextPeak table -------------
rows = []

if not (isinstance(diag, dict) and len(diag)):
    raise RuntimeError("Task 2 needs `diag` from save_meltbloom2 to be in memory.")

for f, d in diag.items():
    # per-float inputs from your diagnostics
    t = pd.to_datetime(d["time"])
    chl_raw = np.asarray(d["chl_raw"], float)
    gi_idx  = np.asarray(d["gi_idx"], int)     # YOUR GI picks
    lat     = np.asarray(d["lat"], float)
    lon     = np.asarray(d["lon"], float)
    path    = d.get("path", f"{f}")

    if len(t) < 5 or len(gi_idx) == 0:
        continue

    # 30-day smoothed CHL on daily grid (from raw mixed-layer CHL)
    s30 = _smooth_30d_daily(t, chl_raw)
    if s30.empty:
        continue

    # season labeling for daily series
    s30_df = pd.DataFrame({"t": s30.index, "chl30": s30.values})
    s30_df["season"] = s30_df["t"].map(_season_year)

    # deployment month logic to drop first season if Aug–Dec start
    first_time = t.min()
    first_season = _season_year(first_time)
    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        s30_df = s30_df[s30_df["season"] != first_season]

    if s30_df.empty:
        continue

    # GI events → choose the earliest GI per season (if multiple)
    gi_times = t[gi_idx]
    gi_df = pd.DataFrame({"gi_time": gi_times})
    gi_df["season"] = gi_df["gi_time"].map(_season_year)

    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        gi_df = gi_df[gi_df["season"] != first_season]

    if gi_df.empty:
        continue

    # pick earliest GI in each season, compute GI day-of-season
    gi_by_season = (gi_df.sort_values("gi_time")
                        .groupby("season", as_index=False)
                        .first())
    gi_by_season["gi_doy_season"] = gi_by_season["gi_time"].map(_season_doy)

    # per-float mean lat/lon (for coloring/QA)
    lat_mean = float(np.nanmean(lat)) if np.isfinite(lat).any() else np.nan
    lon_mean = float(np.nanmean(lon)) if np.isfinite(lon).any() else np.nan
    float_file = os.path.basename(path)

    # for each GI(season s), find next season’s max 30d-smoothed CHL
    for _, r in gi_by_season.iterrows():
        s = int(r["season"])
        s_next = s + 1

        # next season window on the daily series
        start_next, end_next = _season_bounds(s_next)
        gnext = s30_df[(s30_df["t"] >= start_next) & (s30_df["t"] <= end_next)]

        if len(gnext) == 0 or np.all(~np.isfinite(gnext["chl30"].values)):
            continue

        peak_idx = int(np.nanargmax(gnext["chl30"].values))
        peak_val = float(gnext["chl30"].values[peak_idx])
        peak_time = pd.to_datetime(gnext["t"].values[peak_idx])

        rows.append(dict(
            float_file=float_file,
            float_index=int(f),
            season_start=int(s),
            season_next=int(s_next),
            gi_time=pd.to_datetime(r["gi_time"]),
            gi_doy_season=int(r["gi_doy_season"]),
            next_peak_time=peak_time,
            next_peak_chl30=peak_val,
            lat_mean=lat_mean,
            lon_mean=lon_mean
        ))

gi_nextpeak_df = pd.DataFrame(rows).sort_values(["float_file","season_start"]).reset_index(drop=True)
print(f"[Task 2] Built {len(gi_nextpeak_df)} GI→NextPeak pairs.")

# ------------- scatter: GI day (season s) vs next-season max CHL30 -------------
if len(gi_nextpeak_df):
    plt.figure(figsize=(6.6,5.2))
    sc = plt.scatter(gi_nextpeak_df["gi_doy_season"],
                     gi_nextpeak_df["next_peak_chl30"],
                     c=gi_nextpeak_df["lat_mean"], s=44, cmap="viridis")
    plt.colorbar(sc, label="Latitude (°)")
    plt.xlabel("GI day of season (days since Jul 1)")
    plt.ylabel("Next season max 30-day smoothed CHL")
    plt.title("Earlier GI vs Following-season Max CHL (color = latitude)")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # optional simple linear fit
    x = gi_nextpeak_df["gi_doy_season"].to_numpy(dtype=float)
    y = gi_nextpeak_df["next_peak_chl30"].to_numpy(dtype=float)
    m = np.isfinite(x) & np.isfinite(y)
    if m.sum() >= 3 and np.nanvar(y[m]) > 0:
        coef = np.polyfit(x[m], y[m], 1)
        r2 = 1 - (np.nansum((y[m] - (coef[0]*x[m] + coef[1]))**2) /
                  np.nansum((y[m] - np.nanmean(y[m]))**2))
        print(f"[Task 2] Fit: slope={coef[0]:.3f}  R²={r2:.2f}")
else:
    print("[Task 2] No valid rows to plot.")
```

```python
# ==========================================
# Task 3 — GI (this season) vs NEXT season's Annually Integrated Chl
# Uses YOUR GI picks from `diag` and the Jul→Jun seasonal table.
# Y-axis uses `annual_chl_sum` (true integration) from the next season.
# ==========================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SEASON_START_MONTH = 7
DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY = True  # same rule as before

def _season_year(dt):
    return dt.year if dt.month >= SEASON_START_MONTH else dt.year - 1

def _season_doy(dt):
    start = pd.Timestamp(year=_season_year(dt), month=7, day=1)
    return int((dt.normalize() - start).days) + 1

# ---- Load the seasonal metrics table (prefer RAM, else CSV) ----
def _load_seasonal_metrics():
    if 'metrics_seasonal' in globals() and isinstance(metrics_seasonal, pd.DataFrame) and len(metrics_seasonal):
        df = metrics_seasonal.copy()
        src = "RAM"
    else:
        csv_path = os.path.join(DIR, "seasonal_chl_ice_metrics_JulJun.csv")
        if not os.path.exists(csv_path):
            raise RuntimeError("Task 3 needs `metrics_seasonal` in RAM or the CSV "
                               "`seasonal_chl_ice_metrics_JulJun.csv` on disk.")
        df = pd.read_csv(csv_path)
        src = "CSV"
    # sanity: must have these columns
    need = {"float_file","season_start","annual_chl_sum","annual_chl_mean"}
    missing = need - set(df.columns)
    if missing:
        raise RuntimeError(f"Seasonal metrics table missing: {missing}")
    return df, src

# ---- Build GI (this season) → next-season annual integration pairings ----
if not (isinstance(diag, dict) and len(diag)):
    raise RuntimeError("Task 3 needs `diag` from your `save_meltbloom2` run.")

season_tbl, src_where = _load_seasonal_metrics()
print(f"[Task 3] Using seasonal metrics from {src_where} with {len(season_tbl)} rows.")

rows = []

for f, d in diag.items():
    # Pull per-float diagnostics you created
    t        = pd.to_datetime(d["time"])
    gi_idx   = np.asarray(d["gi_idx"], int)  # YOUR GI picks (indices into t)
    lat      = np.asarray(d["lat"], float)
    lon      = np.asarray(d["lon"], float)
    path     = d.get("path", f"{f}")
    if len(t) < 5 or gi_idx.size == 0:
        continue

    first_time   = t.min()
    first_season = _season_year(first_time)

    gi_times = t[gi_idx]
    gi_df = pd.DataFrame({"gi_time": gi_times})
    gi_df["season"] = gi_df["gi_time"].map(_season_year)

    # Drop first season if float deployed Aug–Dec (to keep clean Jul→Jun windows)
    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        gi_df = gi_df[gi_df["season"] != first_season]

    if gi_df.empty:
        continue

    # Earliest GI per season (your convention)
    gi_by_season = (gi_df.sort_values("gi_time")
                        .groupby("season", as_index=False)
                        .first())
    gi_by_season["gi_doy_season"] = gi_by_season["gi_time"].map(_season_doy)

    # Per-float mean location (useful for coloring/QC)
    lat_mean = float(np.nanmean(lat)) if np.isfinite(lat).any() else np.nan
    lon_mean = float(np.nanmean(lon)) if np.isfinite(lon).any() else np.nan
    float_file = os.path.basename(path)

    # Pair season s (GI) with next season s+1 annual integration from seasonal table
    # Prefer true integration: `annual_chl_sum`. (We also carry mean for reference.)
    st_float = season_tbl[season_tbl["float_file"] == float_file].copy()

    for _, r in gi_by_season.iterrows():
        s  = int(r["season"])
        s1 = s + 1  # next season
        nxt = st_float[st_float["season_start"] == s1]
        if len(nxt) == 0:
            continue
        ann_sum  = float(nxt["annual_chl_sum"].iloc[0])
        ann_mean = float(nxt["annual_chl_mean"].iloc[0])
        rows.append(dict(
            float_file=float_file,
            float_index=int(f),
            gi_time=pd.to_datetime(r["gi_time"]),
            gi_doy_season=int(r["gi_doy_season"]),
            season_start=int(s),
            season_next=int(s1),
            next_ann_chl_sum=ann_sum,     # <-- main Y variable (integrated)
            next_ann_chl_mean=ann_mean,   # optional reference
            lat_mean=lat_mean,
            lon_mean=lon_mean
        ))

gi_nextann_df = pd.DataFrame(rows).sort_values(
    ["float_file","season_start"]
).reset_index(drop=True)

print(f"[Task 3] Built {len(gi_nextann_df)} GI→NextSeason(annual) pairs.")

# ---- Scatter: GI day-of-season vs NEXT season annual integrated Chl ----
if len(gi_nextann_df):
    plt.figure(figsize=(6.8,5.2))
    sc = plt.scatter(gi_nextann_df["gi_doy_season"],
                     gi_nextann_df["next_ann_chl_sum"],
                     c=gi_nextann_df["lat_mean"], s=46, cmap="viridis")
    plt.colorbar(sc, label="Latitude (°)")
    plt.xlabel("GI day of season (days since Jul 1)")
    plt.ylabel("Next season annually integrated Chl (sum of daily 0–200 m)")
    plt.title("Earlier GI vs Following-season Annual Integrated Chl\n(color = latitude)")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Optional: simple linear fit
    x = gi_nextann_df["gi_doy_season"].to_numpy(dtype=float)
    y = gi_nextann_df["next_ann_chl_sum"].to_numpy(dtype=float)
    m = np.isfinite(x) & np.isfinite(y)
    if m.sum() >= 3 and np.nanvar(y[m]) > 0:
        coef = np.polyfit(x[m], y[m], 1)
        r2 = 1 - (np.nansum((y[m] - (coef[0]*x[m] + coef[1]))**2) /
                  np.nansum((y[m] - np.nanmean(y[m]))**2))
        print(f"[Task 3] Linear fit (GI day → next annual sum): slope={coef[0]:.3f}, R²={r2:.2f}")
else:
    print("[Task 3] No valid GI→next-season annual pairs to plot.")
```

```python
# ==========================================
# Task 5a — Compute POC from bbp (uses BBP700[_ADJUSTED])
# - POC (mg m^-3) = K_POC_BBP700 * bbp700 (m^-1)
# - Also computes 0–200 m depth-integrated POC (mg m^-2)
# - Writes one tidy CSV combining all floats
# ==========================================

import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw

DIR = "/glade/u/home/vyaasv/Floats"   # <- your directory
OUT_CSV = os.path.join(DIR, "poc_timeseries_0_200m.csv")

# ---- set your conversion (adjust if your group uses a different K)
K_POC_BBP700 = 13000.0   # mg m^-3 per m^-1  (POC ≈ 13,000 * bbp700)

# ---- helpers (reuse your style)
def _pick(ds, base):
    """Prefer *_ADJUSTED if present, else base."""
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base] if base in ds.variables else None

def _qc_mask(ds, base, ok=("0","1","2","8")):
    """Return boolean mask where QC char is in ok; None if no QC var."""
    for suf in ("_ADJUSTED_QC", "_QC"):
        name = base + suf
        if name in ds.variables:
            qc = ds[name].astype("U1")
            return qc.isin(list(ok))
    return None

def _decode_time(ds):
    """Robust TIME to pandas.DatetimeIndex."""
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        idx = t.to_index()
    except Exception:
        idx = pd.to_datetime(t.values)
    return pd.DatetimeIndex(idx)

def _integrate_0_200(depth_m, field):
    """
    depth_m: 1D positive-down (m), same length as field along depth
    field:   1D array (mg m^-3)
    returns mg m^-2 via trapezoid rule over 0..200 m
    """
    z = np.asarray(depth_m, float)
    v = np.asarray(field, float)
    m = np.isfinite(z) & np.isfinite(v)
    if m.sum() < 2:
        return np.nan
    use = m & (z <= 200.0)
    if use.sum() < 2:
        return np.nan
    return float(np.trapz(v[use], z[use]))

rows = []
paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
print(f"[POC] Found {len(paths)} files")

for path in paths:
    base = os.path.basename(path)
    try:
        ds = xr.open_dataset(path)

        # pick variables
        P  = _pick(ds, "PRES")
        T  = _pick(ds, "TEMP")
        S  = _pick(ds, "PSAL")
        BB = _pick(ds, "BBP700")
        lon = ds["LONGITUDE"] if "LONGITUDE" in ds else None
        lat = ds["LATITUDE"]  if "LATITUDE"  in ds else None
        if any(v is None for v in (P, T, S, BB, lon, lat)):
            print(f"[POC] SKIP (missing needed vars) {base}")
            ds.close(); continue

        # QC: keep 0/1/2/8
        for da, name in ((P,"PRES"),(T,"TEMP"),(S,"PSAL"),(BB,"BBP700")):
            qm = _qc_mask(ds, name, ok=("0","1","2","8"))
            if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                da[:] = da.where(qm)

        # screens
        S  = S.where(S > 28.0)      # polar surface-water sanity
        BB = BB.where(BB >= 0)      # non-negative bbp

        # time + arrays
        time = _decode_time(ds)
        nprof = int(BB.sizes["TIME"])
        DEPTH_DIM = [d for d in BB.dims if d != "TIME"][0]
        nlev  = int(BB.sizes[DEPTH_DIM])

        # compute POC per profile and 0–200 m integral
        for p in range(nprof):
            try:
                bbp = np.asarray(BB.isel(TIME=p).values, float)             # m^-1
                pres= np.asarray(P.isel(TIME=p).values, float)              # dbar
                la  = float(lat.isel({lat.dims[0]: p}).values)
                lo  = float(lon.isel({lon.dims[0]: p}).values)
            except Exception:
                # Some files use different first dim name for lon/lat
                la  = float(lat.isel({list(lat.dims)[0]: p}).values) if lat.ndim==1 else float(lat.values[p])
                lo  = float(lon.isel({list(lon.dims)[0]: p}).values) if lon.ndim==1 else float(lon.values[p])
                pres= np.asarray(P.isel(TIME=p).values, float)
                bbp = np.asarray(BB.isel(TIME=p).values, float)

            # pressure -> depth (positive-down)
            try:
                depth = -gsw.z_from_p(pres, la)
            except Exception:
                depth = np.full_like(pres, np.nan, float)

            # POC concentration (mg m^-3)
            poc = K_POC_BBP700 * bbp

            # 0–200 m integral (mg m^-2)
            poc_int = _integrate_0_200(depth, poc)

            rows.append(dict(
                float_file = base,
                time = time[p].isoformat() if p < len(time) else np.nan,
                lon = lo, lat = la,
                poc_int_0_200 = poc_int
            ))

        ds.close()
        print(f"[POC] OK {base}: {nprof} profiles")

    except Exception as e:
        print(f"[POC] SKIP {base}: {e}")

# write tidy CSV
poc_df = pd.DataFrame(rows)
if len(poc_df):
    poc_df.to_csv(OUT_CSV, index=False)
    print(f"[POC] Wrote: {OUT_CSV}")
else:
    print("[POC] No rows written.")
```

```python
# ==========================================
# Task 5b — Time–depth section of POC (mg m^-3)
# - Reads each *.nc in DIR
# - POC = K_POC_BBP700 * BBP700 (m^-1)
# - Interpolates to a uniform depth grid (0..300 m by 2 m)
# - Plots a Hovmöller (time vs depth) per file
# ==========================================

import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ---- reuse if already defined; else defaults
try:
    DIR
except NameError:
    DIR = "/glade/u/home/vyaasv/Floats"

try:
    K_POC_BBP700
except NameError:
    K_POC_BBP700 = 13000.0  # mg m^-3 per m^-1

# ---- small helpers (compatible with your earlier style)
def _pick(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base] if base in ds.variables else None

def _qc_mask(ds, base, ok=("0","1","2","8")):
    for suf in ("_ADJUSTED_QC", "_QC"):
        name = base + suf
        if name in ds.variables:
            qc = ds[name].astype("U1")
            return qc.isin(list(ok))
    return None

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        idx = t.to_index()
    except Exception:
        idx = pd.to_datetime(t.values)
    return pd.DatetimeIndex(idx)

def _interp_col(z_src, v_src, z_tgt):
    """1D linear interp with NaN-safe handling; returns values on z_tgt."""
    z = np.asarray(z_src, float)
    v = np.asarray(v_src, float)
    m = np.isfinite(z) & np.isfinite(v)
    if m.sum() < 2:
        return np.full_like(z_tgt, np.nan, float)
    # require monotonic z for np.interp
    order = np.argsort(z[m])
    return np.interp(z_tgt, z[m][order], v[m][order], left=np.nan, right=np.nan)

# ---- plotting parameters
Z_MAX = 300.0     # meters (positive down)
DZ    = 2.0
z_grid = np.arange(0.0, Z_MAX + DZ, DZ)

paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
print(f"[POC-Plot] Found {len(paths)} files in {DIR}")

for path in paths:
    base = os.path.basename(path)
    try:
        ds = xr.open_dataset(path)

        P  = _pick(ds, "PRES")
        BB = _pick(ds, "BBP700")
        LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None
        LON = ds["LONGITUDE"] if "LONGITUDE" in ds else None
        if any(v is None for v in (P, BB, LAT, LON)) or "TIME" not in ds:
            print(f"[POC-Plot] SKIP (missing BBP/PRES/LAT/LON/TIME): {base}")
            ds.close(); continue

        # QC + screens
        for da, name in ((P,"PRES"), (BB,"BBP700")):
            qm = _qc_mask(ds, name, ok=("0","1","2","8"))
            if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                da[:] = da.where(qm)
        BB = BB.where(BB >= 0)

        time = _decode_time(ds)
        nprof = int(BB.sizes["TIME"])
        dname = [d for d in BB.dims if d != "TIME"][0]

        # allocate (profiles x depth)
        POC_sec = np.full((nprof, z_grid.size), np.nan, float)

        for p in range(nprof):
            la = float(LAT.isel({LAT.dims[0]: p}).values)
            pres = np.asarray(P.isel(TIME=p).values, float)
            bbp  = np.asarray(BB.isel(TIME=p).values, float)  # m^-1
            # pressure -> depth (positive down)
            try:
                depth = -gsw.z_from_p(pres, la)
            except Exception:
                depth = np.full_like(pres, np.nan, float)
            poc = ((3.12*(10**4)*bbp)+3.04)/12  #K_POC_BBP700 * bbp  # mg m^-3
            # interpolate onto common z grid
            POC_sec[p, :] = _interp_col(depth, poc, z_grid)

        # robust color limits
        valid = np.isfinite(POC_sec)
        if valid.any():
            vmin = float(np.nanpercentile(POC_sec[valid], 5))
            vmax = float(np.nanpercentile(POC_sec[valid], 95))
            if not np.isfinite(vmin) or not np.isfinite(vmax) or vmin >= vmax:
                vmin, vmax = np.nanmin(POC_sec[valid]), np.nanmax(POC_sec[valid])
        else:
            print(f"[POC-Plot] No valid POC values in {base}")
            ds.close(); continue

        # plot (time x depth)
        Tnums = mdates.date2num(time.to_pydatetime())  # 1D
        T2D, Z2D = np.meshgrid(Tnums, z_grid, indexing="ij")  # (profiles x depth)

        plt.figure(figsize=(11, 4.6))
        pm = plt.pcolormesh(T2D, Z2D, POC_sec, shading="auto", vmin=vmin, vmax=vmax)
        cb = plt.colorbar(pm, label="POC (mg m$^{-3}$)")
        plt.gca().invert_yaxis()  # surface at top
        plt.ylim(Z_MAX, 0)
        plt.xlabel("Time")
        plt.ylabel("Depth (m)")
        plt.title(f"POC time–depth section (0–{int(Z_MAX)} m)\n{base}")
        plt.gca().xaxis.set_major_locator(mdates.YearLocator())
        plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
        plt.grid(alpha=0.25, which="major", axis="y")
        plt.tight_layout()
        plt.show()

        ds.close()
        print(f"[POC-Plot] OK: {base}")

    except Exception as e:
        print(f"[POC-Plot] SKIP {base}: {e}")
```

```python
# ==========================================
# Task 5c — Depth-integrated POC vs time, overlaid with depth-integrated Chl
# - POC = K_POC_BBP700 * BBP700; integrate 0–200 m
# - Chl integrated 0–200 m (same logic as earlier: trapz of non-negative values)
# - 30-day smoothing, twin-axis plot, peak markers, and lag diagnostics
# ==========================================

import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ---- reuse existing globals if defined; else set defaults
try:
    DIR
except NameError:
    # DIR = "/glade/u/home/vyaasv/Floats"
    DIR = "/glade/u/home/vyaasv/5905078"

try:
    K_POC_BBP700
except NameError:
    K_POC_BBP700 = 13000.0  # mg m^-3 per m^-1 (adjust if you have a preferred coefficient)

Z_MAX = 200.0  # integration limit (m)

# ---- helper shims (kept consistent with your prior cells)
def _pick(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else (ds[base] if base in ds.variables else None)

def _qc_mask(ds, base, ok=("0","1","2","8")):
    for suf in ("_ADJUSTED_QC", "_QC"):
        name = base + suf
        if name in ds.variables:
            qc = ds[name].astype("U1")
            return qc.isin(list(ok))
    return None

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        idx = t.to_index()
    except Exception:
        idx = pd.to_datetime(t.values)
    return pd.DatetimeIndex(idx)

def _integrate_profile_on_depth(value, pres, lat, zmax=Z_MAX, clip_zero=False):
    """Integrate one profile to zmax using pressure->depth; returns scalar integral."""
    v = np.asarray(value, float)
    p = np.asarray(pres, float)
    if not (np.isfinite(v).any() and np.isfinite(p).any()):
        return np.nan
    try:
        z = -np.asarray(gsw.z_from_p(p, float(lat)), float)  # positive down
    except Exception:
        return np.nan
    m = np.isfinite(v) & np.isfinite(z) & (z <= zmax)
    if m.sum() < 2:
        return np.nan
    vv = np.clip(v[m], 0.0, np.inf) if clip_zero else v[m]
    zz = z[m]
    return float(np.trapz(vv, zz))

def _smooth_30d(series_indexed):
    """Your same 30-day centered smoother on a daily grid."""
    t = pd.to_datetime(series_indexed.index, errors="coerce")
    s = pd.Series(np.asarray(series_indexed.values, float), index=t).sort_index()
    s = s[~s.index.isna()]
    if s.empty or s.notna().sum() == 0:
        return s
    s = s.groupby(s.index.normalize()).mean()
    s = s.asfreq("1D")
    s = s.interpolate(method="time", limit_direction="both")
    return s.rolling("30D", center=True, min_periods=1).mean()

def _find_peak_time(s):
    """Return timestamp of global max (ignoring NaNs), else None."""
    if s is None or len(s) == 0 or s.isna().all():
        return None
    i = s.values.argmax() if np.isfinite(s.values).any() else None
    if i is None:
        return None
    return s.index[i]

# ---- main loop over files
paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
print(f"[POC-vs-Chl] Found {len(paths)} files in {DIR}")

for path in paths:
    base = os.path.basename(path)
    try:
        ds = xr.open_dataset(path)

        P   = _pick(ds, "PRES")
        BBP = _pick(ds, "BBP700")
        # Chl aliases: prefer CPHL (your files have CPHL_ADJUSTED); fallback to CHLA if present
        CHL = _pick(ds, "CPHL") if _pick(ds, "CPHL") is not None else _pick(ds, "CHLA")
        LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None

        if any(v is None for v in (P, BBP, CHL, LAT)) or "TIME" not in ds:
            print(f"[POC-vs-Chl] SKIP (missing PRES/BBP/CHL/LAT/TIME): {base}")
            ds.close(); continue

        # QC + basic screens (allow relaxed optics QC like earlier)
        for da, name in ((P,"PRES"), (BBP,"BBP700"), (CHL, CHL.name.split("_")[0])):
            qm = _qc_mask(ds, name, ok=("0","1","2","8"))
            if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                da[:] = da.where(qm)
        BBP = BBP.where(BBP >= 0)
        CHL = CHL.where(CHL >= 0)

        time = _decode_time(ds)
        nprof = int(P.sizes["TIME"])
        dname = [d for d in P.dims if d != "TIME"][0]

        # per-profile depth-integrated series
        poc_int = np.full(nprof, np.nan, float)  # mg m^-2
        chl_int = np.full(nprof, np.nan, float)  # (Chl units) m * conc

        for i in range(nprof):
            la = float(LAT.isel({LAT.dims[0]: i}).values)
            pres_i = np.asarray(P.isel(TIME=i).values, float)
            bbp_i  = np.asarray(BBP.isel(TIME=i).values, float)
            chl_i  = np.asarray(CHL.isel(TIME=i).values, float)

            poc = K_POC_BBP700 * bbp_i  # mg m^-3
            poc_int[i] = _integrate_profile_on_depth(poc, pres_i, la, zmax=Z_MAX, clip_zero=True)
            chl_int[i] = _integrate_profile_on_depth(chl_i, pres_i, la, zmax=Z_MAX, clip_zero=True)

        # build indexed series and smooth 30D
        s_poc = pd.Series(poc_int, index=time).dropna()
        s_chl = pd.Series(chl_int, index=time).dropna()
        s_poc_sm = _smooth_30d(s_poc)
        s_chl_sm = _smooth_30d(s_chl)

        if s_poc_sm.empty or s_chl_sm.empty:
            print(f"[POC-vs-Chl] SKIP (no valid smoothed series): {base}")
            ds.close(); continue

        # find peaks (global); you can limit to windows if desired
        t_peak_poc = _find_peak_time(s_poc_sm)
        t_peak_chl = _find_peak_time(s_chl_sm)
        lag_days = (t_peak_poc - t_peak_chl).days if (t_peak_poc and t_peak_chl) else np.nan

        # ---- PLOT: both on one figure with twin y-axes
        fig, ax1 = plt.subplots(figsize=(11, 4.2))
        l1, = ax1.plot(s_chl_sm.index, s_chl_sm.values, label="Integrated Chl (0–200 m)", lw=1.6)
        ax1.set_ylabel("Integrated Chl (0–200 m)")
        ax1.grid(alpha=0.3)

        ax2 = ax1.twinx()
        l2, = ax2.plot(s_poc_sm.index, s_poc_sm.values, label="Integrated POC (0–200 m)", lw=1.6, linestyle="--")
        ax2.set_ylabel("Integrated POC (mg m$^{-2}$, 0–200 m)")

        # peak markers
        if t_peak_chl:
            ax1.axvline(t_peak_chl, color=l1.get_color(), lw=1.2, ls=":")
        if t_peak_poc:
            ax2.axvline(t_peak_poc, color=l2.get_color(), lw=1.2, ls=":")

        ax1.set_title(f"{base} — Depth-integrated time series (30-day mean)\n"
                      f"Peak lag (POC − Chl) ≈ {lag_days:.0f} days" if np.isfinite(lag_days) else f"{base} — Depth-integrated time series (30-day mean)")
        ax1.xaxis.set_major_locator(mdates.YearLocator())
        ax1.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
        fig.tight_layout()

        # legend combining both axes
        lines = [l1, l2]
        labels = [h.get_label() for h in lines]
        ax1.legend(lines, labels, loc="upper left")
        plt.show()

        # quick print diagnostics
        print(f"[POC-vs-Chl] {base}: peak Chl = {t_peak_chl}, peak POC = {t_peak_poc}, lag = {lag_days} days")

        ds.close()

    except Exception as e:
        print(f"[POC-vs-Chl] SKIP {base}: {e}")
```

```python
# ============================
# Under-ice (T - Tf ≤ RTC) vs POSITION_QC == '8' consistency check
# ============================

import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt

# --- expects DIR (folder with *.nc) and RTC already defined, e.g.:
# DIR = "/glade/u/home/vyaasv/Floats"
# RTC = 0.3  # or 0.5 — use the same threshold you used for melt/under-ice
RTC=0.5
def _decode_time(ds):
    # robust CF decode
    tvar = "TIME" if "TIME" in ds.variables else ("JULD" if "JULD" in ds.variables else None)
    if tvar is None: raise ValueError("No TIME/JULD in file.")
    try:
        dsd = xr.decode_cf(ds)
        t = dsd[tvar]
    except Exception:
        t = ds[tvar]
    try:
        ti = t.to_index()
    except Exception:
        ti = pd.to_datetime(t.values)
    return pd.DatetimeIndex(ti)

def _get_first(ds, names):
    for n in names:
        if n in ds.variables:
            return ds[n]
    return None

def _near_surface_TS_one_profile(tp, sp, pp, lonp, latp, window_dbar=20.0):
    """Mean T,S around shallowest pressure; Tf from SA on p-grid, then take near-surface index."""
    tp = np.asarray(tp, float); sp = np.asarray(sp, float); pp = np.asarray(pp, float)
    if not (np.isfinite(pp).any() and np.isfinite(tp).any() and np.isfinite(sp).any()):
        return np.nan, np.nan, np.nan
    i0 = int(np.nanargmin(pp)) if np.isfinite(pp).any() else 0
    near = np.where((pp - pp[i0]) <= window_dbar)[0]
    vT = float(np.nanmean(tp[near])) if near.size else np.nan
    vS = float(np.nanmean(sp[near])) if near.size else np.nan
    try:
        SA = gsw.SA_from_SP(sp, pp, lonp, latp)
        Tf_prof = gsw.t_freezing(SA, 0.0, 0.0)
        Tf_ns = float(Tf_prof[i0])
    except Exception:
        Tf_ns = np.nan
    return vT, vS, Tf_ns

def _under_ice_flags(T, S, P, lon, lat, rtc):
    """Return 0/1 under-ice flags per profile using near-surface T - Tf ≤ rtc."""
    n = T.shape[0]
    pi = np.zeros(n, dtype=int)
    for p in range(n):
        vT, vS, Tf_ns = _near_surface_TS_one_profile(T[p,:], S[p,:], P[p,:],
                                                     float(lon[p]), float(lat[p]))
        tdiff = vT - Tf_ns if np.isfinite(vT) and np.isfinite(Tf_ns) else np.inf
        if np.isfinite(tdiff) and (tdiff <= rtc):
            pi[p] = 1
    return pi

def _position_qc_series(ds):
    """Return per-profile boolean series for POSITION_QC == '8' (interpolated)."""
    if "POSITION_QC" not in ds.variables:
        return None
    q = ds["POSITION_QC"].astype("U1").values  # char per profile
    return (q == "8").astype(bool)

# -------- main pass over files --------
paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
print(f"[INFO] Checking {len(paths)} files in {DIR}")

summary = []
example_plot_done = False

for k, path in enumerate(paths, 1):
    base = os.path.basename(path)
    try:
        ds = xr.open_dataset(path)

        # pull fields (prefer *_ADJUSTED if present)
        Tda = _get_first(ds, ["TEMP_ADJUSTED","TEMP"])
        Sda = _get_first(ds, ["PSAL_ADJUSTED","PSAL"])
        Pda = _get_first(ds, ["PRES_ADJUSTED","PRES"])
        LON = _get_first(ds, ["LONGITUDE","Lon"])
        LAT = _get_first(ds, ["LATITUDE","Lat"])
        if any(v is None for v in [Tda,Sda,Pda,LON,LAT]):
            print(f"[SKIP] {base}: missing T/S/P/LON/LAT")
            ds.close(); continue

        # time
        try:
            t = _decode_time(ds)
        except Exception as e:
            print(f"[SKIP] {base}: time decode failed: {e}")
            ds.close(); continue

        # to numpy
        T = np.asarray(Tda.values, float)
        S = np.asarray(Sda.values, float)
        P = np.asarray(Pda.values, float)
        lon = np.asarray(LON.values, float)
        lat = np.asarray(LAT.values, float)

        m = min(T.shape[0], S.shape[0], P.shape[0], len(lon), len(lat), len(t))
        T, S, P = T[:m,:], S[:m,:], P[:m,:]
        lon, lat, t = lon[:m], lat[:m], t[:m]

        # flags
        pi = _under_ice_flags(T, S, P, lon, lat, rtc=RTC)
        pq8 = _position_qc_series(ds)
        if pq8 is None:
            print(f"[WARN] {base}: no POSITION_QC")
            ds.close(); continue
        pq8 = np.asarray(pq8[:m], bool)

        # metrics
        n = m
        ui = pi.astype(bool)
        both = ui & pq8
        ui_only = ui & ~pq8
        pq8_only = pq8 & ~ui

        frac_ui = ui.sum()/n if n else np.nan
        frac_pq8 = pq8.sum()/n if n else np.nan
        overlap = both.sum()/n if n else np.nan
        p_ui_given_pq8 = (both.sum()/pq8.sum()) if pq8.sum() else np.nan
        p_pq8_given_ui = (both.sum()/ui.sum()) if ui.sum() else np.nan

        summary.append(dict(
            file=base, n=n,
            under_ice_frac=frac_ui, posqc8_frac=frac_pq8,
            overlap_frac=overlap,
            P_ui_given_pq8=p_ui_given_pq8,
            P_pq8_given_ui=p_pq8_given_ui,
            ui_count=int(ui.sum()), pq8_count=int(pq8.sum()), both_count=int(both.sum()),
            ui_only=int(ui_only.sum()), pq8_only=int(pq8_only.sum())
        ))

        # quick overlay plot for the first file with any events
        if (not example_plot_done) and (ui.sum()+pq8.sum() > 0):
            example_plot_done = True
            plt.figure(figsize=(12,3.2))
            ax = plt.gca()
            # blue shading = under-ice (your convention)
            for i in np.where(ui)[0]:
                ax.axvspan(t[i], t[min(i+1, len(t)-1)], color="C0", alpha=0.20)
            # orange vertical ticks = POSITION_QC==8
            for i in np.where(pq8)[0]:
                ax.axvline(t[i], color="tab:orange", lw=1.2, alpha=0.8)
            ax.set_title(f"Under-ice shading vs POSITION_QC==8 — {base}")
            ax.set_ylabel("indicator")
            ax.set_yticks([])
            ax.grid(alpha=0.3)
            # simple legend proxies
            from matplotlib.patches import Patch
            from matplotlib.lines import Line2D
            ax.legend([Patch(color="C0", alpha=0.20), Line2D([0],[0], color="tab:orange", lw=1.2)],
                      ["Under-ice (T−Tf ≤ RTC)", "POSITION_QC == 8"], loc="upper left")
            plt.tight_layout()
            plt.show()

        ds.close()

    except Exception as e:
        print(f"[ERROR] {base}: {e}")

# print summary table
if summary:
    df_sum = pd.DataFrame(summary)
    with pd.option_context("display.max_rows", None, "display.float_format", "{:.3f}".format):
        print(df_sum.sort_values("overlap_frac", ascending=False))
else:
    print("No results to summarize.")
```

Performing the CHL operations with POC

```python
# POC analysis helpers (Jul->Jun), reusing old style
import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# directory with *.nc
try:
    DIR
except NameError:
    DIR = "/glade/u/home/vyaasv/Floats"

POC_TS_CSV   = os.path.join(DIR, "poc_timeseries_0_200m.csv")
POC_SEASONAL = os.path.join(DIR, "seasonal_poc_0_200m_JulJun.csv")

SEASON_START_MONTH = 7
DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY = True  # match your CHL rule

def _season_year(dt):
    # returns the season label for Jul->Jun
    dt = pd.to_datetime(dt)
    return dt.year if dt.month >= SEASON_START_MONTH else dt.year - 1

def _season_bounds(sy):
    start = pd.Timestamp(year=int(sy),   month=7, day=1)
    end   = pd.Timestamp(year=int(sy+1), month=6, day=30, hour=23, minute=59, second=59)
    return start, end

def _season_doy(dt):
    start = pd.Timestamp(year=_season_year(dt), month=7, day=1)
    return int((pd.to_datetime(dt).normalize() - start).days) + 1

def _smooth_30d_daily(times, values):
    # daily series with centered 30d mean (like your smooth_30d)
    t = pd.to_datetime(times)
    s = pd.Series(np.asarray(values, float), index=t).sort_index()
    s = s[~s.index.isna()]
    if s.empty:
        return s
    s = s.groupby(s.index.normalize()).mean()
    s = s.asfreq("1D")
    s = s.interpolate(method="time", limit_direction="both")
    return s.rolling("30D", center=True, min_periods=1).mean()
```

```python
# Build 0–200 m depth-integrated POC (mg m^-2) per profile and save tidy CSV if not present.
# Uses new POC formula with BBP700 or BBP700_ADJUSTED.

def _pick(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base] if base in ds.variables else None

def _qc_mask(ds, base, ok=("0","1","2","8")):
    for suf in ("_ADJUSTED_QC", "_QC"):
        name = base + suf
        if name in ds.variables:
            qc = ds[name].astype("U1")
            return qc.isin(list(ok))
    return None

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        idx = t.to_index()
    except Exception:
        idx = pd.to_datetime(t.values)
    return pd.DatetimeIndex(idx)

def _integrate_0_200(depth_m, field):
    z = np.asarray(depth_m, float)
    v = np.asarray(field, float)
    m = np.isfinite(z) & np.isfinite(v)
    if m.sum() < 2:
        return np.nan
    use = m & (z <= 200.0)
    if use.sum() < 2:
        return np.nan
    # trapz over positive-down z gives mg m^-2
    return float(np.trapz(v[use], z[use]))

def _poc_from_bbp700(bbp):
    # Your formula: POC mg m^-3
    return ((3.12*(10**4)*bbp) + 3.04) / 12.0

if os.path.exists(POC_TS_CSV):
    print(f"[POC] Found existing CSV: {POC_TS_CSV}")
    poc_df = pd.read_csv(POC_TS_CSV, parse_dates=["time"])
else:
    rows = []
    paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
    print(f"[POC] Scanning {len(paths)} files in {DIR}")
    for path in paths:
        base = os.path.basename(path)
        try:
            ds = xr.open_dataset(path)

            P   = _pick(ds, "PRES")
            BB  = _pick(ds, "BBP700")
            LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None
            LON = ds["LONGITUDE"] if "LONGITUDE" in ds else None
            if any(v is None for v in (P, BB, LAT, LON)) or "TIME" not in ds:
                print(f"[POC] SKIP (missing BBP/PRES/LAT/LON/TIME): {base}")
                ds.close(); continue

            for da, name in ((P,"PRES"), (BB,"BBP700")):
                qm = _qc_mask(ds, name, ok=("0","1","2","8"))
                if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                    da[:] = da.where(qm)
            BB = BB.where(BB >= 0)

            time = _decode_time(ds)
            nprof = int(BB.sizes["TIME"])
            dname = [d for d in BB.dims if d != "TIME"][0]

            for p in range(nprof):
                try:
                    la = float(LAT.isel({LAT.dims[0]: p}).values)
                    lo = float(LON.isel({LON.dims[0]: p}).values)
                except Exception:
                    la = float(LAT.values[p]) if LAT.ndim==1 else float(np.ravel(LAT.values)[p])
                    lo = float(LON.values[p]) if LON.ndim==1 else float(np.ravel(LON.values)[p])

                pres = np.asarray(P.isel(TIME=p).values, float)
                bbp  = np.asarray(BB.isel(TIME=p).values, float)

                # pressure -> depth positive-down
                try:
                    depth = -gsw.z_from_p(pres, la)
                except Exception:
                    depth = np.full_like(pres, np.nan, float)

                poc_conc = _poc_from_bbp700(bbp)  # mg m^-3
                poc_int  = _integrate_0_200(depth, poc_conc)  # mg m^-2

                rows.append(dict(
                    float_file = base,
                    time = time[p].isoformat() if p < len(time) else pd.NaT,
                    lon = lo, lat = la,
                    poc_int_0_200 = poc_int
                ))

            ds.close()
            print(f"[POC] OK {base}: {nprof} profiles")
        except Exception as e:
            print(f"[POC] SKIP {base}: {e}")

    poc_df = pd.DataFrame(rows)
    if len(poc_df):
        poc_df["time"] = pd.to_datetime(poc_df["time"], errors="coerce")
        poc_df.to_csv(POC_TS_CSV, index=False)
        print(f"[POC] Wrote: {POC_TS_CSV}")
    else:
        print("[POC] No rows written.")
```

```python
# Build seasonal POC table for Jul->Jun windows
# Uses the integrated 0–200 m POC time series CSV created above

if "poc_df" not in globals():
    poc_df = pd.read_csv(POC_TS_CSV, parse_dates=["time"])

# ensure datetime and clean rows
poc_df["time"] = pd.to_datetime(poc_df["time"], errors="coerce")
poc_df = poc_df.dropna(subset=["time", "float_file"])
poc_df = poc_df[np.isfinite(poc_df["poc_int_0_200"].values)]
poc_df = poc_df.sort_values("time")

# season labeling
poc_df["season_start"] = poc_df["time"].map(_season_year)

def _first_time_for_float(df):
    # always return a Timestamp (or NaT), never a string
    return pd.to_datetime(df["time"], errors="coerce").min()

rows = []
for ff, g in poc_df.groupby("float_file"):
    first_time = _first_time_for_float(g)
    if pd.isna(first_time):
        # nothing usable for this float
        continue

    first_season = _season_year(first_time)
    gg = g.copy()

    # guard: first_time is a Timestamp, safe to use .month now
    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        gg = gg[gg["season_start"] != first_season]
    if gg.empty:
        continue

    for sy, gs in gg.groupby("season_start"):
        s0, s1 = _season_bounds(sy)
        gs = gs[(gs["time"] >= s0) & (gs["time"] <= s1)]
        if gs.empty:
            continue

        daily = (gs.set_index("time")["poc_int_0_200"]
                   .sort_index()
                   .groupby(pd.Grouper(freq="1D")).mean())

        annual_sum  = float(np.nansum(daily.values))
        annual_mean = float(np.nanmean(daily.values)) if daily.size else np.nan

        rows.append(dict(
            float_file=ff,
            season_start=int(sy),
            annual_poc_sum=annual_sum,
            annual_poc_mean=annual_mean,
            profiles_in_season=int(gs.shape[0]),
            days_covered=int(daily.size),
            t_first=gs["time"].min(),
            t_last=gs["time"].max()
        ))

seasonal_poc = pd.DataFrame(rows).sort_values(["float_file","season_start"]).reset_index(drop=True)
print(f"[POC seasonal] rows: {len(seasonal_poc)}")
if len(seasonal_poc):
    seasonal_poc.to_csv(POC_SEASONAL, index=False)
    print(f"[POC seasonal] Wrote: {POC_SEASONAL}")
```

```python
# GI(this season) vs next-season max 30d-smoothed POC (from 0–200 m integrals)
# Requires your 'diag' dict from save_meltbloom2 with d["gi_idx"], d["time"], d["lat"], d["lon"]

assert isinstance(diag, dict) and len(diag), "Need your `diag` in memory."

# load integrated POC time series if not present
if "poc_df" not in globals():
    poc_df = pd.read_csv(POC_TS_CSV, parse_dates=["time"])

# helper: daily 30d smooth per float
def _daily30_for_float(ff):
    g = poc_df[poc_df["float_file"] == ff].copy()
    if g.empty:
        return pd.Series(dtype=float)
    return _smooth_30d_daily(g["time"].values, g["poc_int_0_200"].values)

rows = []
for f, d in diag.items():
    t = pd.to_datetime(d["time"])
    gi_idx = np.asarray(d["gi_idx"], int)
    if (len(t) < 5) or (gi_idx.size == 0):
        continue
    float_file = os.path.basename(d.get("path", str(f)))

    s30 = _daily30_for_float(float_file)
    if s30.empty:
        continue

    # season tag for daily series
    s30_df = pd.DataFrame({"t": s30.index, "poc30": s30.values})
    s30_df["season"] = s30_df["t"].map(_season_year)

    # drop first season if deployed Aug–Dec
    first_time = t.min()
    first_season = _season_year(first_time)
    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        s30_df = s30_df[s30_df["season"] != first_season]
    if s30_df.empty:
        continue

    gi_times = t[gi_idx]
    gi_df = pd.DataFrame({"gi_time": gi_times})
    gi_df["season"] = gi_df["gi_time"].map(_season_year)
    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        gi_df = gi_df[gi_df["season"] != first_season]
    if gi_df.empty:
        continue

    gi_by_season = (gi_df.sort_values("gi_time")
                        .groupby("season", as_index=False)
                        .first())
    gi_by_season["gi_doy_season"] = gi_by_season["gi_time"].map(_season_doy)

    lat_mean = float(np.nanmean(np.asarray(d["lat"], float))) if np.isfinite(np.asarray(d["lat"], float)).any() else np.nan

    for _, r in gi_by_season.iterrows():
        s  = int(r["season"])
        s1 = s + 1
        start_next, end_next = _season_bounds(s1)
        gnext = s30_df[(s30_df["t"] >= start_next) & (s30_df["t"] <= end_next)]
        if len(gnext) == 0 or np.all(~np.isfinite(gnext["poc30"].values)):
            continue
        peak_idx = int(np.nanargmax(gnext["poc30"].values))
        peak_val = float(gnext["poc30"].values[peak_idx])
        peak_time = pd.to_datetime(gnext["t"].values[peak_idx])

        rows.append(dict(
            float_file=float_file,
            float_index=int(f),
            season_start=int(s),
            season_next=int(s1),
            gi_time=pd.to_datetime(r["gi_time"]),
            gi_doy_season=int(r["gi_doy_season"]),
            next_peak_time=peak_time,
            next_peak_poc30=peak_val,
            lat_mean=lat_mean
        ))

gi_nextpeak_poc = pd.DataFrame(rows).sort_values(["float_file","season_start"]).reset_index(drop=True)
print(f"[POC Task A] GI->NextSeason max POC30 pairs: {len(gi_nextpeak_poc)}")

# scatter
if len(gi_nextpeak_poc):
    plt.figure(figsize=(6.6,5.2))
    sc = plt.scatter(gi_nextpeak_poc["gi_doy_season"],
                     gi_nextpeak_poc["next_peak_poc30"],
                     c=gi_nextpeak_poc["lat_mean"], s=44, cmap="viridis")
    plt.colorbar(sc, label="Latitude (deg)")
    plt.xlabel("GI day of season (days since Jul 1)")
    plt.ylabel("Next season max 30d smoothed POC (mg m$^{-2}$)")
    plt.title("Earlier GI vs following-season max POC (color = latitude)")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    x = gi_nextpeak_poc["gi_doy_season"].to_numpy(float)
    y = gi_nextpeak_poc["next_peak_poc30"].to_numpy(float)
    m = np.isfinite(x) & np.isfinite(y)
    if m.sum() >= 3 and np.nanvar(y[m]) > 0:
        coef = np.polyfit(x[m], y[m], 1)
        r2 = 1 - (np.nansum((y[m] - (coef[0]*x[m] + coef[1]))**2) /
                  np.nansum((y[m] - np.nanmean(y[m]))**2))
        print(f"[POC Task A] Fit: slope={coef[0]:.3f}  R^2={r2:.2f}")
else:
    print("[POC Task A] No valid pairs.")
```

```python
# Annually integrated CHL vs POC (Jul->Jun), sums and means
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Paths
try:
    DIR
except NameError:
    DIR = "/glade/u/home/vyaasv/Floats"

CHL_SEASONAL = os.path.join(DIR, "seasonal_chl_ice_metrics_JulJun.csv")
POC_SEASONAL = os.path.join(DIR, "seasonal_poc_0_200m_JulJun.csv")

def _safe_read(path):
    df = pd.read_csv(path)
    # parse datetime columns only if present
    for c in ("t_first", "t_last"):
        if c in df.columns:
            df[c] = pd.to_datetime(df[c], errors="coerce")
    return df

chl = _safe_read(CHL_SEASONAL)
poc = _safe_read(POC_SEASONAL)

# Required columns
need_chl = {"float_file","season_start","annual_chl_sum","annual_chl_mean"}
need_poc = {"float_file","season_start","annual_poc_sum","annual_poc_mean"}

missing = (need_chl - set(chl.columns)) | (need_poc - set(poc.columns))
if missing:
    raise RuntimeError(f"Missing required columns: {missing}")

# Inner merge on float_file and season_start
df = (chl[list(need_chl)]
      .merge(poc[list(need_poc)], on=["float_file","season_start"], how="inner"))

def _finite2(a, b):
    a = np.asarray(a, float); b = np.asarray(b, float)
    return np.isfinite(a) & np.isfinite(b)

m_sum = _finite2(df["annual_chl_sum"],  df["annual_poc_sum"])
m_mean= _finite2(df["annual_chl_mean"], df["annual_poc_mean"])

# Plot 1: seasonal sums
x = df.loc[m_sum, "annual_chl_sum"].to_numpy(float)
y = df.loc[m_sum, "annual_poc_sum"].to_numpy(float)

plt.figure(figsize=(6.8,5.2))
plt.scatter(x, y, s=40, alpha=0.85)
if x.size >= 3 and np.nanvar(y) > 0:
    coef = np.polyfit(x, y, 1)
    xx = np.linspace(np.nanmin(x), np.nanmax(x), 100)
    yy = coef[0]*xx + coef[1]
    r2 = 1 - (np.nansum((y - (coef[0]*x + coef[1]))**2) /
              np.nansum((y - np.nanmean(y))**2))
    plt.plot(xx, yy, lw=2, alpha=0.9)
    plt.text(0.02, 0.98, f"slope={coef[0]:.3g}, R^2={r2:.2f}",
             transform=plt.gca().transAxes, ha="left", va="top")
plt.xlabel("Annual integrated CHL (seasonal sum of daily 0–200 m)")
plt.ylabel("Annual integrated POC (seasonal sum of daily 0–200 m)")
plt.title("Annual CHL vs Annual POC — seasonal sums (Jul->Jun)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Plot 2: seasonal means
x2 = df.loc[m_mean, "annual_chl_mean"].to_numpy(float)
y2 = df.loc[m_mean, "annual_poc_mean"].to_numpy(float)

plt.figure(figsize=(6.8,5.2))
plt.scatter(x2, y2, s=40, alpha=0.85)
if x2.size >= 3 and np.nanvar(y2) > 0:
    coef2 = np.polyfit(x2, y2, 1)
    xx2 = np.linspace(np.nanmin(x2), np.nanmax(x2), 100)
    yy2 = coef2[0]*xx2 + coef2[1]
    r2b = 1 - (np.nansum((y2 - (coef2[0]*x2 + coef2[1]))**2) /
               np.nansum((y2 - np.nanmean(y2))**2))
    plt.plot(xx2, yy2, lw=2, alpha=0.9)
    plt.text(0.02, 0.98, f"slope={coef2[0]:.3g}, R^2={r2b:.2f}",
             transform=plt.gca().transAxes, ha="left", va="top")
plt.xlabel("Mean daily CHL (0–200 m) within season")
plt.ylabel("Mean daily POC (0–200 m) within season")
plt.title("Annual CHL vs Annual POC — seasonal means (Jul->Jun)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Merged rows: {len(df)}")
```

```python
# --- Task 4: POC/CHL ratio and plot for a test float (e.g., 5904183) ---
import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ---- config ----
# try:
#     DIR
# except NameError:
DIR = "/glade/u/home/vyaasv/Floats"

TARGET_WMO = "5904183"   # <- change if you need a different float
CHL_VAR    = "CPHL"      # picks *_ADJUSTED if present
BBP_VAR    = "BBP700"    # picks *_ADJUSTED if present

# POC (mg m^-3) from bbp700 (m^-1)
def poc_from_bbp700(bbp):
    return ((3.12*(10**4)*bbp) + 3.04) / 12.0

# ---- helpers (reuse your style) ----
def _pick(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base] if base in ds.variables else None

def _qc_mask(ds, base, ok=("1","2","8")):
    for suf in ("_ADJUSTED_QC", "_QC"):
        name = base + suf
        if name in ds.variables:
            qc = ds[name].astype("U1")
            return qc.isin(list(ok))
    return None

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        idx = t.to_index()
    except Exception:
        idx = pd.to_datetime(t.values)
    return pd.DatetimeIndex(idx)

def integrate_0_200(depth_m, field):
    z = np.asarray(depth_m, float)     # positive down (m)
    v = np.asarray(field, float)       # mg m^-3
    m = np.isfinite(z) & np.isfinite(v)
    if m.sum() < 2:
        return np.nan
    use = m & (z <= 200.0)
    if use.sum() < 2:
        return np.nan
    # trapezoid integral → mg m^-2 over 0–200 m
    return float(np.trapz(v[use], z[use]))

# ---- gather files for the target float ----
paths_all = sorted(glob.glob(os.path.join(DIR, "*.nc")))
paths = [p for p in paths_all if TARGET_WMO in os.path.basename(p)]
if not paths:
    raise RuntimeError(f"No files found for WMO {TARGET_WMO} in {DIR}")

rows = []
for path in paths:
    base = os.path.basename(path)
    try:
        ds = xr.open_dataset(path)

        P   = _pick(ds, "PRES")
        T   = _pick(ds, "TEMP")
        S   = _pick(ds, "PSAL")
        CHL = _pick(ds, CHL_VAR)
        BB  = _pick(ds, BBP_VAR)
        LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None
        LON = ds["LONGITUDE"] if "LONGITUDE" in ds else None

        need = (P, T, S, CHL, BB, LAT, LON)
        if any(v is None for v in need) or "TIME" not in ds:
            print(f"[WARN] Missing vars in {base}; skipping")
            ds.close(); continue

        # QC & screens (relaxed for optics: keep 0/1/2/8)
        for da, name in ((P,"PRES"), (T,"TEMP"), (S,"PSAL"), (CHL,CHL_VAR), (BB,BBP_VAR)):
            qm = _qc_mask(ds, name, ok=("1","2","8"))
            if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                da[:] = da.where(qm)
        S   = S.where(S > 28.0)
        CHL = CHL.where(CHL >= 0)
        BB  = BB.where(BB >= 0)

        time = _decode_time(ds)
        nprof = int(CHL.sizes["TIME"])

        for p in range(nprof):
            # coords
            try:
                la = float(LAT.isel({LAT.dims[0]: p}).values)
                lo = float(LON.isel({LON.dims[0]: p}).values)
            except Exception:
                la = float(np.ravel(LAT.values)[p])
                lo = float(np.ravel(LON.values)[p])

            pres = np.asarray(P.isel(TIME=p).values, float)
            chl  = np.asarray(CHL.isel(TIME=p).values, float)  # mg m^-3 typically
            bbp  = np.asarray(BB.isel(TIME=p).values, float)   # m^-1

            # pressure -> depth positive-down
            try:
                depth = -gsw.z_from_p(pres, la)
            except Exception:
                depth = np.full_like(pres, np.nan, float)

            # integrals over 0–200 m (mg m^-2)
            chl_int = integrate_0_200(depth, chl)
            poc_conc = poc_from_bbp700(bbp)
            poc_int = integrate_0_200(depth, poc_conc)

            rows.append(dict(
                file = base,
                time = time[p].to_pydatetime() if p < len(time) else pd.NaT,
                lon  = lo, lat = la,
                chl_int_0_200 = chl_int,
                poc_int_0_200 = poc_int
            ))

        ds.close()
    except Exception as e:
        print(f"[SKIP] {base}: {e}")

df = pd.DataFrame(rows).dropna(subset=["time"])
df["time"] = pd.to_datetime(df["time"], errors="coerce")
df = df.sort_values("time")

# safe ratio: POC/CHL; avoid zero/neg/NaN CHL
chl = np.asarray(df["chl_int_0_200"], float)
poc = np.asarray(df["poc_int_0_200"], float)
den  = np.where(np.isfinite(chl) & (chl > 0), chl, np.nan)
ratio = poc / den
df["poc_to_chl"] = ratio

# keep rows with at least CHL or ratio finite for plotting
m_any = np.isfinite(df["chl_int_0_200"].values) | np.isfinite(df["poc_to_chl"].values)
dfp = df[m_any].copy()

# ---- Plot: CHL (left y) and POC/CHL (right y) vs time, hollow circles ----
plt.rcParams.update({"figure.dpi": 130})
fig, ax1 = plt.subplots(figsize=(11, 4.6))

# CHL: hollow circles
ax1.scatter(dfp["time"], dfp["chl_int_0_200"], s=24, facecolors="none", edgecolors="C0", label="CHL (0–200 m, sum)")
ax1.set_ylabel("CHL integrated 0–200 m (mg m$^{-2}$)", color="C0")
ax1.tick_params(axis="y", labelcolor="C0")

# Ratio on right axis: hollow circles
ax2 = ax1.twinx()
ax2.scatter(dfp["time"], dfp["poc_to_chl"], s=24, facecolors="none", edgecolors="C3", label="POC / CHL")
ax2.set_ylabel("POC / CHL", color="C3")
ax2.tick_params(axis="y", labelcolor="C3")

# x-axis ticks at 1/1 and 7/1 every year
ax1.xaxis.set_major_locator(mdates.MonthLocator(bymonth=[1, 7], bymonthday=1))
# Try platform-agnostic formatter
try:
    ax1.xaxis.set_major_formatter(mdates.DateFormatter("%-m/%-d"))  # e.g., 1/1, 7/1 (Unix)
except ValueError:
    ax1.xaxis.set_major_formatter(mdates.DateFormatter("%m/%d"))    # Windows fallback

ax1.set_xlabel("Date (month/day)")
ax1.grid(alpha=0.3)

# Title with WMO
plt.title(f"WMO {TARGET_WMO}: CHL (left) and POC/CHL (right) — 0–200 m integrals")

# Build a combined legend with hollow markers
h1 = plt.Line2D([], [], marker="o", linestyle="None", markersize=6, markerfacecolor="none", markeredgecolor="C0", label="CHL (0–200 m, sum)")
h2 = plt.Line2D([], [], marker="o", linestyle="None", markersize=6, markerfacecolor="none", markeredgecolor="C3", label="POC / CHL")
ax1.legend(handles=[h1, h2], loc="upper left")

fig.tight_layout()
plt.show()
```

```python
# =========================
# Task 5 — Max summertime CHL vs Max summertime POC (after GI, same season)
# =========================
import os, glob
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt

# ---------- Config / paths ----------
try:
    DIR
except NameError:
    DIR = "/glade/u/home/vyaasv/Floats"

CHL_TS_CSV   = os.path.join(DIR, "chl_timeseries_0_200m.csv")          # built here if missing
POC_TS_CSV   = os.path.join(DIR, "poc_timeseries_0_200m.csv")          # reuse from earlier; build if missing
SEASON_START_MONTH = 7
DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY = True   # match your rule

# ---------- Helpers ----------
def _season_year(dt):
    dt = pd.to_datetime(dt)
    return dt.year if dt.month >= SEASON_START_MONTH else dt.year - 1

def _season_bounds(sy):
    start = pd.Timestamp(year=int(sy),   month=7, day=1)
    end   = pd.Timestamp(year=int(sy+1), month=6, day=30, hour=23, minute=59, second=59)
    return start, end

def _smooth_30d_daily(times, values):
    t = pd.to_datetime(times)
    s = pd.Series(np.asarray(values, float), index=t).sort_index()
    s = s[~s.index.isna()]
    if s.empty:
        return s
    s = s.groupby(s.index.normalize()).mean()
    s = s.asfreq("1D")
    s = s.interpolate(method="time", limit_direction="both")
    return s.rolling("30D", center=True, min_periods=1).mean()

def _pick(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base] if base in ds.variables else None

def _qc_mask(ds, base, ok=("0","1","2","8")):
    for suf in ("_ADJUSTED_QC", "_QC"):
        name = base + suf
        if name in ds.variables:
            qc = ds[name].astype("U1")
            return qc.isin(list(ok))
    return None

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        idx = t.to_index()
    except Exception:
        idx = pd.to_datetime(t.values)
    return pd.DatetimeIndex(idx)

def _integrate_0_200(depth_m, field):
    z = np.asarray(depth_m, float)
    v = np.asarray(field, float)
    m = np.isfinite(z) & np.isfinite(v)
    if m.sum() < 2:
        return np.nan
    use = m & (z <= 200.0)
    if use.sum() < 2:
        return np.nan
    return float(np.trapz(v[use], z[use]))

def _ensure_poc_csv():
    if os.path.exists(POC_TS_CSV):
        return pd.read_csv(POC_TS_CSV, parse_dates=["time"])
    # build POC CSV (uses BBP700 → POC formula)
    def _poc_from_bbp700(bbp):
        return ((3.12*(10**4)*bbp) + 3.04) / 12.0
    rows = []
    paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
    for path in paths:
        base = os.path.basename(path)
        try:
            ds = xr.open_dataset(path)
            P   = _pick(ds, "PRES")
            BB  = _pick(ds, "BBP700")
            LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None
            LON = ds["LONGITUDE"] if "LONGITUDE" in ds else None
            if any(v is None for v in (P, BB, LAT, LON)) or "TIME" not in ds:
                ds.close(); continue
            for da, name in ((P,"PRES"), (BB,"BBP700")):
                qm = _qc_mask(ds, name, ok=("0","1","2","8"))
                if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                    da[:] = da.where(qm)
            BB = BB.where(BB >= 0)
            time = _decode_time(ds)
            nprof = int(BB.sizes["TIME"])
            for p in range(nprof):
                la = float(np.ravel(LAT.values)[p]) if LAT.ndim != 1 else float(LAT.isel({LAT.dims[0]: p}).values)
                lo = float(np.ravel(LON.values)[p]) if LON.ndim != 1 else float(LON.isel({LON.dims[0]: p}).values)
                pres = np.asarray(P.isel(TIME=p).values, float)
                bbp  = np.asarray(BB.isel(TIME=p).values, float)
                try:
                    depth = -gsw.z_from_p(pres, la)
                except Exception:
                    depth = np.full_like(pres, np.nan, float)
                poc_conc = _poc_from_bbp700(bbp)      # mg m^-3
                poc_int  = _integrate_0_200(depth, poc_conc)  # mg m^-2
                rows.append(dict(float_file=base, time=time[p].isoformat(), lon=lo, lat=la, poc_int_0_200=poc_int))
            ds.close()
        except Exception:
            pass
    poc_df = pd.DataFrame(rows)
    if len(poc_df):
        poc_df["time"] = pd.to_datetime(poc_df["time"], errors="coerce")
        poc_df.to_csv(POC_TS_CSV, index=False)
    return poc_df

def _ensure_chl_csv():
    if os.path.exists(CHL_TS_CSV):
        return pd.read_csv(CHL_TS_CSV, parse_dates=["time"])
    # build CHL CSV (integrated 0–200 m from CPHL)
    rows = []
    paths = sorted(glob.glob(os.path.join(DIR, "*.nc")))
    for path in paths:
        base = os.path.basename(path)
        try:
            ds = xr.open_dataset(path)
            P   = _pick(ds, "PRES")
            CHL = _pick(ds, "CPHL")
            LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None
            LON = ds["LONGITUDE"] if "LONGITUDE" in ds else None
            if any(v is None for v in (P, CHL, LAT, LON)) or "TIME" not in ds:
                ds.close(); continue
            for da, name in ((P,"PRES"), (CHL,"CPHL")):
                qm = _qc_mask(ds, name, ok=("0","1","2","8"))
                if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                    da[:] = da.where(qm)
            CHL = CHL.where(CHL >= 0)
            time = _decode_time(ds)
            nprof = int(CHL.sizes["TIME"])
            for p in range(nprof):
                la = float(np.ravel(LAT.values)[p]) if LAT.ndim != 1 else float(LAT.isel({LAT.dims[0]: p}).values)
                lo = float(np.ravel(LON.values)[p]) if LON.ndim != 1 else float(LON.isel({LON.dims[0]: p}).values)
                pres = np.asarray(P.isel(TIME=p).values, float)
                chl  = np.asarray(CHL.isel(TIME=p).values, float)  # mg m^-3 (typical units)
                try:
                    depth = -gsw.z_from_p(pres, la)
                except Exception:
                    depth = np.full_like(pres, np.nan, float)
                chl_int = _integrate_0_200(depth, chl)  # mg m^-2
                rows.append(dict(float_file=base, time=time[p].isoformat(), lon=lo, lat=la, chl_int_0_200=chl_int))
            ds.close()
        except Exception:
            pass
    chl_df = pd.DataFrame(rows)
    if len(chl_df):
        chl_df["time"] = pd.to_datetime(chl_df["time"], errors="coerce")
        chl_df.to_csv(CHL_TS_CSV, index=False)
    return chl_df

# ---------- Require diag with GI picks ----------
assert isinstance(diag, dict) and len(diag), "Need your `diag` in memory from save_meltbloom2(..., collect_diag=True)."

# ---------- Load or build time series ----------
chl_df = _ensure_chl_csv()
poc_df = _ensure_poc_csv()

# basic guards
for df in (chl_df, poc_df):
    if "time" not in df.columns:
        df["time"] = pd.NaT
    df["time"] = pd.to_datetime(df["time"], errors="coerce")

# ---------- Per-float daily smoothed series builders ----------
def _daily30_for_float(df, float_file_col, target_ff, value_col):
    g = df[df[float_file_col] == target_ff].copy()
    if g.empty:
        return pd.Series(dtype=float)
    return _smooth_30d_daily(g["time"].values, g[value_col].values)

# ---------- Build per (float, season) maxima after GI ----------
rows = []
for f, d in diag.items():
    float_file = os.path.basename(d.get("path", str(f)))
    t = pd.to_datetime(d["time"])
    gi_idx = np.asarray(d["gi_idx"], int)
    if (len(t) < 5) or (gi_idx.size == 0):
        continue

    # optional: drop first season if deployed Aug–Dec
    first_time = t.min()
    first_season = _season_year(first_time)

    # daily smoothed series for this float
    chl30 = _daily30_for_float(chl_df, "float_file", float_file, "chl_int_0_200")
    poc30 = _daily30_for_float(poc_df, "float_file", float_file, "poc_int_0_200")
    if chl30.empty or poc30.empty:
        continue

    # season tags for daily indexes
    df_chl = pd.DataFrame({"t": chl30.index, "chl30": chl30.values})
    df_chl["season"] = df_chl["t"].map(_season_year)
    df_poc = pd.DataFrame({"t": poc30.index, "poc30": poc30.values})
    df_poc["season"] = df_poc["t"].map(_season_year)

    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        df_chl = df_chl[df_chl["season"] != first_season]
        df_poc = df_poc[df_poc["season"] != first_season]

    if df_chl.empty or df_poc.empty:
        continue

    # earliest GI per season (your convention)
    gi_times = t[gi_idx]
    gi_df = pd.DataFrame({"gi_time": gi_times})
    gi_df["season"] = gi_df["gi_time"].map(_season_year)
    if DISCARD_FIRST_SEASON_IF_DEPLOYED_AFTER_JULY and first_time.month > SEASON_START_MONTH:
        gi_df = gi_df[gi_df["season"] != first_season]
    if gi_df.empty:
        continue
    gi_by_season = (gi_df.sort_values("gi_time")
                        .groupby("season", as_index=False)
                        .first())

    # mean lat for color
    lat_mean = float(np.nanmean(np.asarray(d["lat"], float))) if np.isfinite(np.asarray(d["lat"], float)).any() else np.nan

    # for each season, get maxima AFTER GI within same season window
    for _, r in gi_by_season.iterrows():
        s = int(r["season"])
        gi_time = pd.to_datetime(r["gi_time"])
        s0, s1 = _season_bounds(s)

        # restrict to [GI time, season end]
        chl_win = df_chl[(df_chl["t"] >= gi_time) & (df_chl["t"] <= s1)]
        poc_win = df_poc[(df_poc["t"] >= gi_time) & (df_poc["t"] <= s1)]
        if chl_win.empty or poc_win.empty:
            continue

        # maxima and times
        try:
            iC = int(np.nanargmax(chl_win["chl30"].values))
            max_chl = float(chl_win["chl30"].values[iC])
            tC = pd.to_datetime(chl_win["t"].values[iC])
        except ValueError:
            continue
        try:
            iP = int(np.nanargmax(poc_win["poc30"].values))
            max_poc = float(poc_win["poc30"].values[iP])
            tP = pd.to_datetime(poc_win["t"].values[iP])
        except ValueError:
            continue

        rows.append(dict(
            float_file=float_file,
            float_index=int(f),
            season_start=int(s),
            gi_time=gi_time,
            max_chl_after_gi=max_chl,   # mg m^-2 (smoothed daily)
            max_poc_after_gi=max_poc,   # mg m^-2 (smoothed daily)
            t_max_chl=tC,
            t_max_poc=tP,
            lat_mean=lat_mean
        ))

pair_df = pd.DataFrame(rows).sort_values(["float_file","season_start"]).reset_index(drop=True)
print(f"[Task 5] Built {len(pair_df)} GI→Max (CHL, POC) pairs (same-season, post-GI).")

# ---------- Scatter: Max summertime CHL vs Max summertime POC ----------
if len(pair_df):
    plt.figure(figsize=(6.8,5.2))
    sc = plt.scatter(pair_df["max_chl_after_gi"],
                     pair_df["max_poc_after_gi"],
                     c=pair_df["lat_mean"], s=46, cmap="viridis")
    # # --- Plot regression line on top of scatter ---
    # x = pair_df["max_chl_after_gi"].to_numpy(dtype=float)
    # y = pair_df["max_poc_after_gi"].to_numpy(dtype=float)
    # m = np.isfinite(x) & np.isfinite(y)
    # if m.sum() >= 3 and np.nanvar(y[m]) > 0:
    #     coef = np.polyfit(x[m], y[m], 1)
    #     xx = np.linspace(np.nanmin(x[m]), np.nanmax(x[m]), 100)
    #     yy = coef[0] * xx + coef[1]
    #     r2 = 1 - (np.nansum((y[m] - (coef[0]*x[m] + coef[1]))**2) /
    #               np.nansum((y[m] - np.nanmean(y[m]))**2))
    #     plt.plot(xx, yy, color="k", lw=2.2, alpha=0.9, label=f"y = {coef[0]:.2f}x + {coef[1]:.2f}, R² = {r2:.2f}")
    #     plt.legend(frameon=False)

    plt.colorbar(sc, label="Latitude (°)")
    plt.xlabel("Max summertime CHL after GI (mg m$^{-2}$)")
    plt.ylabel("Max summertime POC after GI (mg m$^{-2}$)")
    plt.title("Max CHL vs Max POC after GI (same season; label says 'Max', 30d smoothing used)")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # simple linear fit
    x = pair_df["max_chl_after_gi"].to_numpy(dtype=float)
    y = pair_df["max_poc_after_gi"].to_numpy(dtype=float)
    m = np.isfinite(x) & np.isfinite(y)
    if m.sum() >= 3 and np.nanvar(y[m]) > 0:
        coef = np.polyfit(x[m], y[m], 1)
        r2 = 1 - (np.nansum((y[m] - (coef[0]*x[m] + coef[1]))**2) /
                  np.nansum((y[m] - np.nanmean(y[m]))**2))
        print(f"[Task 5] Linear fit: slope={coef[0]:.3f}, R^2={r2:.2f}")
else:
    print("[Task 5] No valid rows to plot (check that CHL/POC CSVs and diag exist).")
```

```python
# === ANCP (Johnson et al., 2017) — "official" version ===
# - No interpolation / gap filling / smoothing
# - Integrate only finite points with np.trapz
# - Pick winter/summer months by column-mean nitrate (0–200 m) rule
# - Month "mean" is the mean of per-profile integrals (no vertical regridding)
# - Outputs a tidy table and optional CSV

import os, glob, re, warnings
import numpy as np
import pandas as pd
import xarray as xr
import gsw

# ---------------- USER CONFIG ----------------
DIR = "/glade/u/home/vyaasv/NO3_Floats"
OUT_CSV = os.path.join(DIR, "ancp_johnson_0_200m.csv")  # set to None to skip writing
ZMAX = 50.0  #200.0
SPARSITY_WARN_N = 5            # warn if fewer than this many finite depth points used by np.trapz
SAL_NORMALIZE = False          # Johnson+2017 did not sal-normalize; keep False unless needed
S_REF = 35.0                   # only used if SAL_NORMALIZE=True
PRINT_QA = True
# ---------------------------------------------

warnings.filterwarnings("ignore", message="Mean of empty slice")

NO3_CANDIDATES = ["NTAW", "NITRATE", "NO3"]

def _resolve_var(ds, base_names):
    if isinstance(base_names, str):
        base_names = [base_names]
    for b in base_names:
        an = b + "_ADJUSTED"
        if an in ds.variables:
            return ds[an], b
    for b in base_names:
        if b in ds.variables:
            return ds[b], b
    return None, None

def _apply_qc(ds, da, base):
    for suf in ("_ADJUSTED_QC","_QC"):
        qn = base + suf
        if qn in ds.variables and tuple(ds[qn].dims) == tuple(da.dims):
            qc = ds[qn].astype("U1")
            return da.where(qc.isin(list("128")))
    return da

def _units_kind(units_attr):
    if not units_attr:
        return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    if "mmol/m3" in u or "mmol m-3" in u or "umol/l" in u or "µmol l-1" in u:
        return "mmol_m3"
    return "mmol_m3"

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        return t.to_index()
    except Exception:
        return pd.to_datetime(t.values)

def _lat_at(ds, i):
    if "LATITUDE" not in ds: return np.nan
    da = ds["LATITUDE"]
    if da.ndim == 0: return float(da.values)
    d0 = da.dims[0]
    return float(da.isel({d0: min(i, da.sizes[d0]-1)}).values)

def _depth_from_p(pres_1d, lat):
    p = np.asarray(pres_1d, float)
    try:
        return -gsw.z_from_p(p, float(lat))
    except Exception:
        return np.full_like(p, np.nan, float)

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lat):
    SA = gsw.SA_from_SP(psal, pres, 0.0, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    return (no3_umolkg * rho) / 1000.0  # µmol/kg → µmol/m^3 → mmol/m^3

# def _col_mean_0_200(z, v):
#     z = np.asarray(z, float); v = np.asarray(v, float)
#     m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
#     return float(np.nanmean(v[m])) if m.sum() >= 2 else np.nan

def _col_mean_0_200(z, v):
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
    if m.sum() < 2:
        return np.nan
    zz = z[m]; vv = v[m]
    o = np.argsort(zz)
    intgrl = np.trapz(vv[o], zz[o])  # mmol N m^-2 if v is mmol N m^-3
    return float(intgrl / ZMAX)       # mmol N m^-3 depth-weighted mean over 0–200 m


# def _integral_0_200(z, v):
#     """Return (integral mmol N m^-2, npoints_used)."""
#     z = np.asarray(z, float); v = np.asarray(v, float)
#     m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
#     if m.sum() < 2:
#         return np.nan, int(m.sum())
#     # IMPORTANT: we still integrate on the native (z[m], v[m]) without filling.
#     return float(np.trapz(v[m], z[m])), int(m.sum())

def _integral_0_200(z, v):
    """Return (integral mmol N m^-2, npoints_used)."""
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
    if m.sum() < 2:
        return np.nan, int(m.sum())
    # IMPORTANT: we still integrate on the native (z[m], v[m]) without filling.
    return float(np.trapz(v[m], z[m])), int(m.sum())



def _wmo_from_ds(ds, fname):
    if "PLATFORM_NUMBER" in ds.variables:
        try:
            raw = ds["PLATFORM_NUMBER"].values
            wmo = "".join(np.atleast_1d(raw)[0].astype(str)).strip()
            return re.sub(r"\s+", "", wmo)
        except Exception:
            pass
    m = re.search(r"(\d{6,})", os.path.basename(fname))
    return m.group(1) if m else os.path.basename(fname)

def _concat_time(files):
    """Concatenate along TIME with graceful fallback."""
    dsets = []
    for p in files:
        try: dsets.append(xr.open_dataset(p))
        except Exception: pass
    if not dsets:
        return None, []
    if len(dsets) == 1:
        return dsets[0], dsets
    try:
        ds = xr.concat(dsets, dim="TIME", data_vars="minimal", coords="minimal",
                       compat="override", join="outer")
        return ds, dsets
    except Exception:
        ds = dsets[0]
        for ex in dsets[1:]:
            try:
                ds = xr.concat([ds, ex], dim="TIME", data_vars="minimal",
                               coords="minimal", compat="override", join="outer")
            except Exception:
                pass
        return ds, dsets

def ancp_johnson_table():
    rows = []

    # group files by float WMO with a nitrate-like variable
    groups = {}
    for p in sorted(glob.glob(os.path.join(DIR, "*.nc"))):
        try:
            with xr.open_dataset(p) as ds:
                NO3, _ = _resolve_var(ds, NO3_CANDIDATES)
                if NO3 is not None:
                    wmo = _wmo_from_ds(ds, p)
                    groups.setdefault(wmo, []).append(p)
        except Exception:
            pass

    if not groups:
        print("[ANCP] No files with nitrate-like variable found.")
        return pd.DataFrame(columns=[
            "float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN"
        ])

    for wmo, files in groups.items():
        print(f"wmo:{wmo}")
        ds, dsets = _concat_time(files)
        if ds is None:
            continue

        NO3, no3_base = _resolve_var(ds, NO3_CANDIDATES)
        P, _ = _resolve_var(ds, ["PRES","PRESSURE"])
        S, _ = _resolve_var(ds, ["PSAL","SP","SAL","SALINITY"])
        T, _ = _resolve_var(ds, ["TEMP","TEMP_CTD","TEMPERATURE"])

        if ("TIME" not in ds) or (NO3 is None):
            for d in dsets: d.close()
            continue

        ukind = _units_kind(NO3.attrs.get("units",""))
        need_density = (ukind == "umol_kg")
        if need_density and (P is None or S is None or T is None):
            if PRINT_QA:
                print(f"[ANCP][{wmo}] Skipping: nitrate units µmol/kg but P/S/T missing.")
            for d in dsets: d.close()
            continue

        # QC + screens
        NO3 = _apply_qc(ds, NO3, no3_base).where(NO3 >= 0)
        if P is not None: P = _apply_qc(ds, P, "PRES")
        if S is not None: S = _apply_qc(ds, S, "PSAL")
        if T is not None: T = _apply_qc(ds, T, "TEMP")

        tt = _decode_time(ds)
        n = int(NO3.sizes["TIME"])
        prof = []

        # Build per-profile stats (native depths, no interpolation)
        for i in range(n):
            lat = _lat_at(ds, i)
            p_i = np.asarray(P.isel(TIME=i).values, float) if P is not None else None
            s_i = np.asarray(S.isel(TIME=i).values, float) if S is not None else None
            t_i = np.asarray(T.isel(TIME=i).values, float) if T is not None else None
            no3_i = np.asarray(NO3.isel(TIME=i).values, float)

            z_i = _depth_from_p(p_i, lat) if p_i is not None else np.full_like(no3_i, np.nan)

            # unit conversion
            if need_density:
                if p_i is None or s_i is None or t_i is None:
                    continue
                v = _to_mmol_m3_from_umol_kg(no3_i, p_i, s_i, t_i, lat)
            else:
                v = no3_i

            if SAL_NORMALIZE and (s_i is not None):
                v = v * (S_REF / s_i)

            cm = _col_mean_0_200(z_i, v)
            integ, npts = _integral_0_200(z_i, v)

            if PRINT_QA and npts > 0 and npts < SPARSITY_WARN_N:
                print(f"[{wmo}] {tt[i].date()} sparse depth points in 0–{ZMAX}m: n={npts}")

            prof.append(dict(
                time=tt[i], year=tt[i].year, month=tt[i].month,
                col_mean=cm, integral=integ, npts=npts
            ))

        df = pd.DataFrame(prof).sort_values("time")
        if df.empty:
            for d in dsets: d.close()
            continue

        years = sorted(df["year"].unique())
        print(f"years:{len(years)}")
        print(f"years:{years}")
        for y in years:
            # Winter: Aug–Nov (year y) — pick month that contains profile with max column-mean
            Wcand = df[(df["year"]==y) & (df["month"].isin([8,9,10,11]))].copy()
            if Wcand.empty or (Wcand["col_mean"].all()==np.NaN):
                if wmo=="5904397":
                    print("wcand empty")
                continue
            iw = Wcand["col_mean"].idxmax()
            mw = int(Wcand.loc[iw, "month"])
            Wm = Wcand[Wcand["month"] == mw]

            # Summer: Dec(y)–Mar(y+1) — pick month that contains profile with min column-mean
            Scand = df[((df["year"]==y) & (df["month"]==12)) | ((df["year"]==y+1) & (df["month"].isin([1,2,3])))]
            if Scand.empty:
                if wmo=="5904397":
                    print("scand empty")
                continue
            is_ = Scand["col_mean"].idxmin()
            ms = int(Scand.loc[is_, "month"])
            Sm = Scand[Scand["month"] == ms]

            if Wm.empty or Sm.empty:
                if wmo=="5904397":
                    print("wm,sm empty")
                continue

            # Month "means" = average of per-profile integrals (no vertical regridding)
            Wints = Wm["integral"].to_numpy(dtype=float)
            Sints = Sm["integral"].to_numpy(dtype=float)

            # keep only finite integrals
            Wints = Wints[np.isfinite(Wints)]
            Sints = Sints[np.isfinite(Sints)]

            if (Wints.size == 0) or (Sints.size == 0):
                if PRINT_QA:
                    print(f"[{wmo} {y}] skip: no finite integrals in winter or summer (W={Wints.size}, S={Sints.size})")
                continue

            Wmean = float(np.nanmean(Wints))
            Smean = float(np.nanmean(Sints))
            print(f"Year:{y}\nWmean={Wmean},Smean={Smean}")
            Ninv = Wmean - Smean                        # mmol N m^-2
            ancp_mmolC = 6.6 * Ninv                     # mmol C m^-2
            ancp_molC  = ancp_mmolC / 1000.0            # mol C m^-2

            rows.append(dict(
                float_wmo=wmo,
                season_calendar_year=y,
                winter_month=mw, summer_month=ms,
                ancp_molC_m2=ancp_molC,
                ancp_mmolC_m2=ancp_mmolC,
                n_prof_winter=int(Wm.shape[0]),
                n_prof_summer=int(Sm.shape[0]),
                warn_winter_lowN=int((Wm["npts"] < SPARSITY_WARN_N).sum()),
                warn_summer_lowN=int((Sm["npts"] < SPARSITY_WARN_N).sum())
            ))

        for d in dsets:
            try: d.close()
            except Exception: pass

    cols = ["float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN"]

    if not rows:
        print("[ANCP] No valid winter/summer pairs computed.")
        ancp = pd.DataFrame(columns=cols)
    else:
        ancp = (pd.DataFrame(rows)
                .sort_values(["float_wmo","season_calendar_year"])
                .reset_index(drop=True))

    print(f"[ANCP] Built {len(ancp)} rows from {ancp['float_wmo'].nunique() if not ancp.empty else 0} floats.")
    if not ancp.empty and PRINT_QA:
        show = ancp[["float_wmo","season_calendar_year","winter_month","summer_month",
                     "ancp_molC_m2","n_prof_winter","n_prof_summer",
                     "warn_winter_lowN","warn_summer_lowN"]]
        print(show.to_string(index=False))
    if OUT_CSV and not ancp.empty:
        ancp.to_csv(OUT_CSV, index=False)
        print("Wrote:", OUT_CSV)
    return ancp

# ---- run ----
ancp_table = ancp_johnson_table()
```

```python
# === ANCP (Johnson et al., 2017) — "official" version ===
# - No interpolation / gap filling / smoothing
# - Integrate only finite points with np.trapz
# - Pick winter/summer months by column-mean nitrate (0–200 m) rule
# - Month "mean" is the mean of per-profile integrals (no vertical regridding)
# - Outputs a tidy table and optional CSV

import os, glob, re, warnings
import numpy as np
import pandas as pd
import xarray as xr
import gsw

# ---------------- USER CONFIG ----------------
DIR = "/glade/u/home/vyaasv/NO3_Floats"
OUT_CSV = os.path.join(DIR, "ancp_johnson_0_200m.csv")  # set to None to skip writing
ZMAX = 200.0
SPARSITY_WARN_N = 5            # warn if fewer than this many finite depth points used by np.trapz
SAL_NORMALIZE = False          # Johnson+2017 did not sal-normalize; keep False unless needed
S_REF = 35.0                   # only used if SAL_NORMALIZE=True
PRINT_QA = True
# ---------------------------------------------

warnings.filterwarnings("ignore", message="Mean of empty slice")

NO3_CANDIDATES = ["NTAW", "NITRATE", "NO3"]

def _resolve_var(ds, base_names):
    if isinstance(base_names, str):
        base_names = [base_names]
    for b in base_names:
        an = b + "_ADJUSTED"
        if an in ds.variables:
            return ds[an], b
    for b in base_names:
        if b in ds.variables:
            return ds[b], b
    return None, None

def _apply_qc(ds, da, base):
    for suf in ("_ADJUSTED_QC","_QC"):
        qn = base + suf
        if qn in ds.variables and tuple(ds[qn].dims) == tuple(da.dims):
            qc = ds[qn].astype("U1")
            return da.where(qc.isin(list("128")))
    return da

def _units_kind(units_attr):
    if not units_attr:
        return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    if "mmol/m3" in u or "mmol m-3" in u or "umol/l" in u or "µmol l-1" in u:
        return "mmol_m3"
    return "mmol_m3"

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        return t.to_index()
    except Exception:
        return pd.to_datetime(t.values)

def _lat_at(ds, i):
    if "LATITUDE" not in ds: return np.nan
    da = ds["LATITUDE"]
    if da.ndim == 0: return float(da.values)
    d0 = da.dims[0]
    return float(da.isel({d0: min(i, da.sizes[d0]-1)}).values)

def _depth_from_p(pres_1d, lat):
    p = np.asarray(pres_1d, float)
    try:
        return -gsw.z_from_p(p, float(lat))
    except Exception:
        return np.full_like(p, np.nan, float)

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lat):
    SA = gsw.SA_from_SP(psal, pres, 0.0, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    return (no3_umolkg * rho) / 1000.0  # µmol/kg → µmol/m^3 → mmol/m^3

# def _col_mean_0_200(z, v):
#     z = np.asarray(z, float); v = np.asarray(v, float)
#     m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
#     return float(np.nanmean(v[m])) if m.sum() >= 2 else np.nan

def _col_mean_0_200(z, v):
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= 100)
    if m.sum() < 2:
        return np.nan
    zz = z[m]; vv = v[m]
    o = np.argsort(zz)
    intgrl = np.trapz(vv[o], zz[o])  # mmol N m^-2 if v is mmol N m^-3
    return float(intgrl / ZMAX)       # mmol N m^-3 depth-weighted mean over 0–200 m


# def _integral_0_200(z, v):
#     """Return (integral mmol N m^-2, npoints_used)."""
#     z = np.asarray(z, float); v = np.asarray(v, float)
#     m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
#     if m.sum() < 2:
#         return np.nan, int(m.sum())
#     # IMPORTANT: we still integrate on the native (z[m], v[m]) without filling.
#     return float(np.trapz(v[m], z[m])), int(m.sum())

def _integral_0_200(z, v):
    """Return (integral mmol N m^-2, npoints_used)."""
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= 100)
    if m.sum() < 2:
        return np.nan, int(m.sum())
    # IMPORTANT: we still integrate on the native (z[m], v[m]) without filling.
    return float(np.trapz(v[m], z[m])), int(m.sum())



def _wmo_from_ds(ds, fname):
    if "PLATFORM_NUMBER" in ds.variables:
        try:
            raw = ds["PLATFORM_NUMBER"].values
            wmo = "".join(np.atleast_1d(raw)[0].astype(str)).strip()
            return re.sub(r"\s+", "", wmo)
        except Exception:
            pass
    m = re.search(r"(\d{6,})", os.path.basename(fname))
    return m.group(1) if m else os.path.basename(fname)

def _concat_time(files):
    """Concatenate along TIME with graceful fallback."""
    dsets = []
    for p in files:
        try: dsets.append(xr.open_dataset(p))
        except Exception: pass
    if not dsets:
        return None, []
    if len(dsets) == 1:
        return dsets[0], dsets
    try:
        ds = xr.concat(dsets, dim="TIME", data_vars="minimal", coords="minimal",
                       compat="override", join="outer")
        return ds, dsets
    except Exception:
        ds = dsets[0]
        for ex in dsets[1:]:
            try:
                ds = xr.concat([ds, ex], dim="TIME", data_vars="minimal",
                               coords="minimal", compat="override", join="outer")
            except Exception:
                pass
        return ds, dsets

def ancp_johnson_table():
    rows = []

    # group files by float WMO with a nitrate-like variable
    groups = {}
    for p in sorted(glob.glob(os.path.join(DIR, "*.nc"))):
        try:
            with xr.open_dataset(p) as ds:
                NO3, _ = _resolve_var(ds, NO3_CANDIDATES)
                if NO3 is not None:
                    wmo = _wmo_from_ds(ds, p)
                    groups.setdefault(wmo, []).append(p)
        except Exception:
            pass

    if not groups:
        print("[ANCP] No files with nitrate-like variable found.")
        return pd.DataFrame(columns=[
            "float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN"
        ])

    for wmo, files in groups.items():
        print(f"wmo:{wmo}")
        ds, dsets = _concat_time(files)
        if ds is None:
            continue

        NO3, no3_base = _resolve_var(ds, NO3_CANDIDATES)
        P, _ = _resolve_var(ds, ["PRES","PRESSURE"])
        S, _ = _resolve_var(ds, ["PSAL","SP","SAL","SALINITY"])
        T, _ = _resolve_var(ds, ["TEMP","TEMP_CTD","TEMPERATURE"])

        if ("TIME" not in ds) or (NO3 is None):
            for d in dsets: d.close()
            continue

        ukind = _units_kind(NO3.attrs.get("units",""))
        need_density = (ukind == "umol_kg")
        if need_density and (P is None or S is None or T is None):
            if PRINT_QA:
                print(f"[ANCP][{wmo}] Skipping: nitrate units µmol/kg but P/S/T missing.")
            for d in dsets: d.close()
            continue

        # QC + screens
        NO3 = _apply_qc(ds, NO3, no3_base).where(NO3 >= 0)
        if P is not None: P = _apply_qc(ds, P, "PRES")
        if S is not None: S = _apply_qc(ds, S, "PSAL")
        if T is not None: T = _apply_qc(ds, T, "TEMP")

        tt = _decode_time(ds)
        n = int(NO3.sizes["TIME"])
        prof = []

        # Build per-profile stats (native depths, no interpolation)
        for i in range(n):
            lat = _lat_at(ds, i)
            p_i = np.asarray(P.isel(TIME=i).values, float) if P is not None else None
            s_i = np.asarray(S.isel(TIME=i).values, float) if S is not None else None
            t_i = np.asarray(T.isel(TIME=i).values, float) if T is not None else None
            no3_i = np.asarray(NO3.isel(TIME=i).values, float)

            z_i = _depth_from_p(p_i, lat) if p_i is not None else np.full_like(no3_i, np.nan)

            # unit conversion
            if need_density:
                if p_i is None or s_i is None or t_i is None:
                    continue
                v = _to_mmol_m3_from_umol_kg(no3_i, p_i, s_i, t_i, lat)
            else:
                v = no3_i

            if SAL_NORMALIZE and (s_i is not None):
                v = v * (S_REF / s_i)

            cm = _col_mean_0_200(z_i, v)
            integ, npts = _integral_0_200(z_i, v)

            if PRINT_QA and npts > 0 and npts < SPARSITY_WARN_N:
                print(f"[{wmo}] {tt[i].date()} sparse depth points in 0–200 m: n={npts}")

            prof.append(dict(
                time=tt[i], year=tt[i].year, month=tt[i].month,
                col_mean=cm, integral=integ, npts=npts
            ))

        df = pd.DataFrame(prof).sort_values("time")
        if df.empty:
            for d in dsets: d.close()
            continue

        years = sorted(df["year"].unique())
        print(f"years:{len(years)}")
        print(f"years:{years}")

        for y in years:
            # ---- Winter: Aug–Nov (y), month-weighted ----
            winter_month_means = []
            for mw in (8, 9, 10, 11):
                W = df[(df["year"] == y) & (df["month"] == mw)]
                if W.empty:
                    continue
                wints = W["integral"].to_numpy(dtype=float)
                wints = wints[np.isfinite(wints)]
                if wints.size:
                    winter_month_means.append(float(np.nanmean(wints)))

            # ---- Summer: Dec (y) and Jan–Mar (y+1), month-weighted ----
            summer_month_means = []
            # December of year y
            D = df[(df["year"] == y) & (df["month"] == 12)]
            if not D.empty:
                dints = D["integral"].to_numpy(dtype=float)
                dints = dints[np.isfinite(dints)]
                if dints.size:
                    summer_month_means.append(float(np.nanmean(dints)))
            # Jan–Mar of year y+1
            for ms in (1, 2, 3):
                S = df[(df["year"] == y+1) & (df["month"] == ms)]
                if S.empty:
                    continue
                sints = S["integral"].to_numpy(dtype=float)
                sints = sints[np.isfinite(sints)]
                if sints.size:
                    summer_month_means.append(float(np.nanmean(sints)))

            if not winter_month_means or not summer_month_means:
                if PRINT_QA:
                    print(f"[{wmo} {y}] skip: missing month means (W={len(winter_month_means)}, S={len(summer_month_means)})")
                continue

            Wmean = float(np.nanmean(winter_month_means))
            Smean = float(np.nanmean(summer_month_means))
            Ninv = Wmean - Smean            # mmol N m^-2
            ancp_mmolC = 6.6 * Ninv         # mmol C m^-2
            ancp_molC  = ancp_mmolC / 1000. # mol C m^-2

            rows.append(dict(
                float_wmo=wmo,
                season_calendar_year=y,
                ancp_molC_m2=ancp_molC,
                ancp_mmolC_m2=ancp_mmolC,
            ))


        for d in dsets:
            try: d.close()
            except Exception: pass

    cols = ["float_wmo","season_calendar_year",
            "ancp_molC_m2","ancp_mmolC_m2"]

    if not rows:
        print("[ANCP] No valid winter/summer pairs computed.")
        ancp = pd.DataFrame(columns=cols)
    else:
        ancp = (pd.DataFrame(rows)
                .sort_values(["float_wmo","season_calendar_year"])
                .reset_index(drop=True))

    print(f"[ANCP] Built {len(ancp)} rows from {ancp['float_wmo'].nunique() if not ancp.empty else 0} floats.")
    if not ancp.empty and PRINT_QA:
        show = ancp[["float_wmo","season_calendar_year",
                     "ancp_molC_m2"]]
        print(show.to_string(index=False))
    if OUT_CSV and not ancp.empty:
        ancp.to_csv(OUT_CSV, index=False)
        print("Wrote:", OUT_CSV)
    return ancp

# ---- run ----
ancp_table = ancp_johnson_table()
```

```python
# === ANCP (Johnson et al., 2017) — "official" version ===
# - No interpolation / gap filling / smoothing
# - Integrate only finite points with np.trapz
# - Pick winter/summer months by column-mean nitrate (0–200 m) rule
# - Month "mean" is the mean of per-profile integrals (no vertical regridding)
# - Outputs a tidy table and optional CSV

import os, glob, re, warnings
import numpy as np
import pandas as pd
import xarray as xr
import gsw

# ---------------- USER CONFIG ----------------
DIR = "/glade/u/home/vyaasv/NO3_Floats"
OUT_CSV = os.path.join(DIR, "ancp_johnson_0_200m.csv")  # set to None to skip writing
ZMAX = 200.0
SPARSITY_WARN_N = 5            # warn if fewer than this many finite depth points used by np.trapz
SAL_NORMALIZE = False          # Johnson+2017 did not sal-normalize; keep False unless needed
S_REF = 35.0                   # only used if SAL_NORMALIZE=True
PRINT_QA = True
# ---------------------------------------------

warnings.filterwarnings("ignore", message="Mean of empty slice")

NO3_CANDIDATES = ["NTAW", "NITRATE", "NO3"]

def _resolve_var(ds, base_names):
    if isinstance(base_names, str):
        base_names = [base_names]
    for b in base_names:
        an = b + "_ADJUSTED"
        if an in ds.variables:
            return ds[an], b
    for b in base_names:
        if b in ds.variables:
            return ds[b], b
    return None, None

def _apply_qc(ds, da, base):
    for suf in ("_ADJUSTED_QC","_QC"):
        qn = base + suf
        if qn in ds.variables and tuple(ds[qn].dims) == tuple(da.dims):
            qc = ds[qn].astype("U1")
            return da.where(qc.isin(list("128")))
    return da

def _units_kind(units_attr):
    if not units_attr:
        return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    if "mmol/m3" in u or "mmol m-3" in u or "umol/l" in u or "µmol l-1" in u:
        return "mmol_m3"
    return "mmol_m3"

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        return t.to_index()
    except Exception:
        return pd.to_datetime(t.values)

def _lat_at(ds, i):
    if "LATITUDE" not in ds: return np.nan
    da = ds["LATITUDE"]
    if da.ndim == 0: return float(da.values)
    d0 = da.dims[0]
    return float(da.isel({d0: min(i, da.sizes[d0]-1)}).values)

def _depth_from_p(pres_1d, lat):
    p = np.asarray(pres_1d, float)
    try:
        return -gsw.z_from_p(p, float(lat))
    except Exception:
        return np.full_like(p, np.nan, float)

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lat):
    SA = gsw.SA_from_SP(psal, pres, 0.0, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    return (no3_umolkg * rho) / 1000.0  # µmol/kg → µmol/m^3 → mmol/m^3

def _col_mean_0_200(z, v):
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
    return float(np.nanmean(v[m])) if m.sum() >= 2 else np.nan

def _integral_0_200(z, v):
    """Return (integral mmol N m^-2, npoints_used)."""
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
    if m.sum() < 2:
        return np.nan, int(m.sum())
    # IMPORTANT: we still integrate on the native (z[m], v[m]) without filling.
    return float(np.trapz(v[m], z[m])), int(m.sum())

def _wmo_from_ds(ds, fname):
    if "PLATFORM_NUMBER" in ds.variables:
        try:
            raw = ds["PLATFORM_NUMBER"].values
            wmo = "".join(np.atleast_1d(raw)[0].astype(str)).strip()
            return re.sub(r"\s+", "", wmo)
        except Exception:
            pass
    m = re.search(r"(\d{6,})", os.path.basename(fname))
    return m.group(1) if m else os.path.basename(fname)

def _concat_time(files):
    """Concatenate along TIME with graceful fallback."""
    dsets = []
    for p in files:
        try: dsets.append(xr.open_dataset(p))
        except Exception: pass
    if not dsets:
        return None, []
    if len(dsets) == 1:
        return dsets[0], dsets
    try:
        ds = xr.concat(dsets, dim="TIME", data_vars="minimal", coords="minimal",
                       compat="override", join="outer")
        return ds, dsets
    except Exception:
        ds = dsets[0]
        for ex in dsets[1:]:
            try:
                ds = xr.concat([ds, ex], dim="TIME", data_vars="minimal",
                               coords="minimal", compat="override", join="outer")
            except Exception:
                pass
        return ds, dsets

def ancp_johnson_table():
    rows = []

    # group files by float WMO with a nitrate-like variable
    groups = {}
    for p in sorted(glob.glob(os.path.join(DIR, "*.nc"))):
        try:
            with xr.open_dataset(p) as ds:
                NO3, _ = _resolve_var(ds, NO3_CANDIDATES)
                if NO3 is not None:
                    wmo = _wmo_from_ds(ds, p)
                    groups.setdefault(wmo, []).append(p)
        except Exception:
            pass

    if not groups:
        print("[ANCP] No files with nitrate-like variable found.")
        return pd.DataFrame(columns=[
            "float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN"
        ])

    for wmo, files in groups.items():
        print(f"wmo:{wmo}")
        ds, dsets = _concat_time(files)
        if ds is None:
            continue

        NO3, no3_base = _resolve_var(ds, NO3_CANDIDATES)
        P, _ = _resolve_var(ds, ["PRES","PRESSURE"])
        S, _ = _resolve_var(ds, ["PSAL","SP","SAL","SALINITY"])
        T, _ = _resolve_var(ds, ["TEMP","TEMP_CTD","TEMPERATURE"])

        if ("TIME" not in ds) or (NO3 is None):
            for d in dsets: d.close()
            continue

        ukind = _units_kind(NO3.attrs.get("units",""))
        need_density = (ukind == "umol_kg")
        if need_density and (P is None or S is None or T is None):
            if PRINT_QA:
                print(f"[ANCP][{wmo}] Skipping: nitrate units µmol/kg but P/S/T missing.")
            for d in dsets: d.close()
            continue

        # QC + screens
        NO3 = _apply_qc(ds, NO3, no3_base).where(NO3 >= 0)
        if P is not None: P = _apply_qc(ds, P, "PRES")
        if S is not None: S = _apply_qc(ds, S, "PSAL")
        if T is not None: T = _apply_qc(ds, T, "TEMP")

        tt = _decode_time(ds)
        n = int(NO3.sizes["TIME"])
        prof = []

        # Build per-profile stats (native depths, no interpolation)
        for i in range(n):
            lat = _lat_at(ds, i)
            p_i = np.asarray(P.isel(TIME=i).values, float) if P is not None else None
            s_i = np.asarray(S.isel(TIME=i).values, float) if S is not None else None
            t_i = np.asarray(T.isel(TIME=i).values, float) if T is not None else None
            no3_i = np.asarray(NO3.isel(TIME=i).values, float)

            z_i = _depth_from_p(p_i, lat) if p_i is not None else np.full_like(no3_i, np.nan)

            # unit conversion
            if need_density:
                if p_i is None or s_i is None or t_i is None:
                    continue
                v = _to_mmol_m3_from_umol_kg(no3_i, p_i, s_i, t_i, lat)
            else:
                v = no3_i

            if SAL_NORMALIZE and (s_i is not None):
                v = v * (S_REF / s_i)

            cm = _col_mean_0_200(z_i, v)
            integ, npts = _integral_0_200(z_i, v)

            if PRINT_QA and npts > 0 and npts < SPARSITY_WARN_N:
                print(f"[{wmo}] {tt[i].date()} sparse depth points in 0–200 m: n={npts}")

            prof.append(dict(
                time=tt[i], year=tt[i].year, month=tt[i].month,
                col_mean=cm, integral=integ, npts=npts
            ))

        df = pd.DataFrame(prof).sort_values("time")
        if df.empty:
            for d in dsets: d.close()
            continue

        years = sorted(df["year"].unique())
        print(f"years:{len(years)}")
        print(f"years:{years}")
        for y in years:
            # Winter: Aug–Nov (year y) — pick month that contains profile with max column-mean
            Wcand = df[(df["year"]==y) & (df["month"].isin([8,9,10,11]))].copy()
            Wcand = Wcand[np.isfinite(Wcand["col_mean"])]  # <— drop NaNs
            if Wcand.empty:
                if wmo=="5904397":
                    print("wcand empty")
                continue
            iw = Wcand["col_mean"].idxmax()
            mw = int(Wcand.loc[iw, "month"])
            Wm = Wcand[Wcand["month"] == mw]

            # Summer: Dec(y)–Mar(y+1) — pick month that contains profile with min column-mean
            Scand = df[((df["year"]==y) & (df["month"]==12)) | ((df["year"]==y+1) & (df["month"].isin([1,2,3])))]
            Scand = Scand[np.isfinite(Scand["col_mean"])]  # <— drop NaNs
            if Scand.empty:
                if wmo=="5904397":
                    print("scand empty")
                continue
            is_ = Scand["col_mean"].idxmin()
            ms = int(Scand.loc[is_, "month"])
            Sm = Scand[Scand["month"] == ms]

            if Wm.empty or Sm.empty:
                if wmo=="5904397":
                    print("wm,sm empty")
                continue

            # Month "means" = average of per-profile integrals (no vertical regridding)
            Wints = Wm["integral"].to_numpy(dtype=float)
            Sints = Sm["integral"].to_numpy(dtype=float)

            # keep only finite integrals
            Wints = Wints[np.isfinite(Wints)]
            Sints = Sints[np.isfinite(Sints)]

            if (Wints.size == 0) or (Sints.size == 0):
                if PRINT_QA:
                    print(f"[{wmo} {y}] skip: no finite integrals in winter or summer (W={Wints.size}, S={Sints.size})")
                continue

            Wmean = float(np.nanmean(Wints))
            Smean = float(np.nanmean(Sints))
            print(f"Year:{y}\nWmean={Wmean},Smean={Smean}")
            Ninv = Wmean - Smean                        # mmol N m^-2
            ancp_mmolC = 6.6 * Ninv                     # mmol C m^-2
            ancp_molC  = ancp_mmolC / 1000.0            # mol C m^-2

            rows.append(dict(
                float_wmo=wmo,
                season_calendar_year=y,
                winter_month=mw, summer_month=ms,
                ancp_molC_m2=ancp_molC,
                ancp_mmolC_m2=ancp_mmolC,
                n_prof_winter=int(Wm.shape[0]),
                n_prof_summer=int(Sm.shape[0]),
                warn_winter_lowN=int((Wm["npts"] < SPARSITY_WARN_N).sum()),
                warn_summer_lowN=int((Sm["npts"] < SPARSITY_WARN_N).sum())
            ))

        for d in dsets:
            try: d.close()
            except Exception: pass

    cols = ["float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN"]

    if not rows:
        print("[ANCP] No valid winter/summer pairs computed.")
        ancp = pd.DataFrame(columns=cols)
    else:
        ancp = (pd.DataFrame(rows)
                .sort_values(["float_wmo","season_calendar_year"])
                .reset_index(drop=True))

    print(f"[ANCP] Built {len(ancp)} rows from {ancp['float_wmo'].nunique() if not ancp.empty else 0} floats.")
    if not ancp.empty and PRINT_QA:
        show = ancp[["float_wmo","season_calendar_year","winter_month","summer_month",
                     "ancp_molC_m2","n_prof_winter","n_prof_summer",
                     "warn_winter_lowN","warn_summer_lowN"]]
        print(show.to_string(index=False))
    if OUT_CSV and not ancp.empty:
        ancp.to_csv(OUT_CSV, index=False)
        print("Wrote:", OUT_CSV)
    return ancp

# ---- run ----
ancp_table = ancp_johnson_table()
```

```python
# === ANCP (Johnson et al., 2017) — "official" version with rich diagnostics ===
# - No vertical interpolation / gap-filling / smoothing
# - Integrate only finite points with np.trapz
# - Pick winter/summer months by column-mean nitrate (0–200 m) rule
# - Month "mean" is the mean of per-profile *integrals* (no vertical regridding)
# - Adds detailed diagnostics + optional per-profile CSV

import os, glob, re, warnings
import numpy as np
import pandas as pd
import xarray as xr
import gsw
from collections import Counter

# ---------------- USER CONFIG ----------------
DIR = "/glade/u/home/vyaasv/NO3_Floats"
OUT_CSV = os.path.join(DIR, "ancp_johnson_0_200m.csv")                 # final season table
OUT_PROF_CSV = os.path.join(DIR, "ancp_profile_diagnostics_0_200m.csv") # per-profile QA (optional)
WRITE_PROF_DIAG = True           # write per-profile diagnostics CSV
ZMAX = 200.0
SPARSITY_WARN_N = 5             # warn if fewer than this many finite depth points used by np.trapz
SAL_NORMALIZE = False           # Johnson+2017 did not sal-normalize; keep False unless needed
S_REF = 35.0                    # only used if SAL_NORMALIZE=True
PRINT_QA = True                 # concise QA
PRINT_DETAILED = True           # season-by-season detailed QA per float
# ---------------------------------------------

warnings.filterwarnings("ignore", message="Mean of empty slice")

NO3_CANDIDATES = ["NTAW", "NITRATE", "NO3"]

def _resolve_var(ds, base_names):
    if isinstance(base_names, str):
        base_names = [base_names]
    for b in base_names:
        an = b + "_ADJUSTED"
        if an in ds.variables:
            return ds[an], b
    for b in base_names:
        if b in ds.variables:
            return ds[b], b
    return None, None

def _apply_qc(ds, da, base):
    for suf in ("_ADJUSTED_QC","_QC"):
        qn = base + suf
        if qn in ds.variables and tuple(ds[qn].dims) == tuple(da.dims):
            qc = ds[qn].astype("U1")
            return da.where(qc.isin(list("0128")))
    return da

def _units_kind(units_attr):
    if not units_attr:
        return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    if "mmol/m3" in u or "mmol m-3" in u or "umol/l" in u or "µmol l-1" in u:
        return "mmol_m3"
    return "mmol_m3"

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        return t.to_index()
    except Exception:
        return pd.to_datetime(t.values)

def _lat_at(ds, i):
    if "LATITUDE" not in ds:
        return np.nan
    da = ds["LATITUDE"]
    if da.ndim == 0:
        return float(da.values)
    d0 = da.dims[0]
    return float(da.isel({d0: min(i, da.sizes[d0]-1)}).values)

def _depth_from_p(pres_1d, lat):
    p = np.asarray(pres_1d, float)
    try:
        return -gsw.z_from_p(p, float(lat))  # positive down
    except Exception:
        return np.full_like(p, np.nan, float)

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lat):
    SA = gsw.SA_from_SP(psal, pres, 0.0, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    return (no3_umolkg * rho) / 1000.0  # µmol/kg → µmol/m^3 → mmol/m^3

def _col_mean_0_200(z, v):
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
    return float(np.nanmean(v[m])) if m.sum() >= 2 else np.nan

def _integral_0_200(z, v):
    """Return (integral mmol N m^-2, npoints_used)."""
    z = np.asarray(z, float); v = np.asarray(v, float)
    m = np.isfinite(z) & np.isfinite(v) & (z >= 0) & (z <= ZMAX)
    if m.sum() < 2:
        return np.nan, int(m.sum())
    return float(np.trapz(v[m], z[m])), int(m.sum())

def _wmo_from_ds(ds, fname):
    if "PLATFORM_NUMBER" in ds.variables:
        try:
            raw = ds["PLATFORM_NUMBER"].values
            wmo = "".join(np.atleast_1d(raw)[0].astype(str)).strip()
            return re.sub(r"\s+", "", wmo)
        except Exception:
            pass
    m = re.search(r"(\d{6,})", os.path.basename(fname))
    return m.group(1) if m else os.path.basename(fname)

def _concat_time(files):
    """Concatenate along TIME with graceful fallback."""
    dsets = []
    for p in files:
        try: dsets.append(xr.open_dataset(p))
        except Exception: pass
    if not dsets:
        return None, []
    if len(dsets) == 1:
        return dsets[0], dsets
    try:
        ds = xr.concat(dsets, dim="TIME", data_vars="minimal", coords="minimal",
                       compat="override", join="outer")
        return ds, dsets
    except Exception:
        ds = dsets[0]
        for ex in dsets[1:]:
            try:
                ds = xr.concat([ds, ex], dim="TIME", data_vars="minimal",
                               coords="minimal", compat="override", join="outer")
            except Exception:
                pass
        return ds, dsets

def ancp_johnson_table():
    rows = []
    prof_diag_rows = []  # optional per-profile diagnostics

    # group files by float WMO with a nitrate-like variable
    groups = {}
    for p in sorted(glob.glob(os.path.join(DIR, "*.nc"))):
        try:
            with xr.open_dataset(p) as ds:
                NO3, _ = _resolve_var(ds, NO3_CANDIDATES)
                if NO3 is not None:
                    wmo = _wmo_from_ds(ds, p)
                    groups.setdefault(wmo, []).append(p)
        except Exception:
            pass

    if not groups:
        print("[ANCP] No files with nitrate-like variable found.")
        return pd.DataFrame(columns=[
            "float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN",
            "winter_integral_mean","summer_integral_mean",
            "winter_npts_min","winter_npts_med","winter_npts_max",
            "summer_npts_min","summer_npts_med","summer_npts_max"
        ])

    for wmo, files in groups.items():
        ds, dsets = _concat_time(files)
        if ds is None:
            continue

        NO3, no3_base = _resolve_var(ds, NO3_CANDIDATES)
        P, _ = _resolve_var(ds, ["PRES","PRESSURE"])
        S, _ = _resolve_var(ds, ["PSAL","SP","SAL","SALINITY"])
        T, _ = _resolve_var(ds, ["TEMP","TEMP_CTD","TEMPERATURE"])

        if ("TIME" not in ds) or (NO3 is None):
            for d in dsets:
                d.close()
            continue

        ukind = _units_kind(NO3.attrs.get("units",""))
        need_density = (ukind == "umol_kg")
        if need_density and (P is None or S is None or T is None):
            if PRINT_QA:
                print(f"[ANCP][{wmo}] Skipping: nitrate units µmol/kg but P/S/T missing.")
            for d in dsets:
                d.close()
            continue

        # QC + screens
        NO3 = _apply_qc(ds, NO3, no3_base).where(NO3 >= 0)
        if P is not None:
            P = _apply_qc(ds, P, "PRES")
        if S is not None:
            S = _apply_qc(ds, S, "PSAL")
        if T is not None:
            T = _apply_qc(ds, T, "TEMP")

        tt = _decode_time(ds)
        n = int(NO3.sizes["TIME"])
        prof = []

        # Build per-profile stats (native depths, no vertical interpolation)
        for i in range(n):
            lat = _lat_at(ds, i)
            p_i = np.asarray(P.isel(TIME=i).values, float) if P is not None else None
            s_i = np.asarray(S.isel(TIME=i).values, float) if S is not None else None
            t_i = np.asarray(T.isel(TIME=i).values, float) if T is not None else None
            no3_i = np.asarray(NO3.isel(TIME=i).values, float)

            z_i = _depth_from_p(p_i, lat) if p_i is not None else np.full_like(no3_i, np.nan)

            # unit conversion
            if need_density:
                if p_i is None or s_i is None or t_i is None:
                    continue
                v = _to_mmol_m3_from_umol_kg(no3_i, p_i, s_i, t_i, lat)
            else:
                v = no3_i

            if SAL_NORMALIZE and (s_i is not None):
                v = v * (S_REF / s_i)

            cm = _col_mean_0_200(z_i, v)
            integ, npts = _integral_0_200(z_i, v)

            if PRINT_QA and npts > 0 and npts < SPARSITY_WARN_N:
                print(f"[{wmo}] {tt[i].date()} sparse depth points in 0–200 m: n={npts}")

            prof.append(dict(
                wmo=wmo,
                time=tt[i],
                year=tt[i].year,
                month=tt[i].month,
                col_mean=cm,
                integral=integ,
                npts=npts
            ))

        df = pd.DataFrame(prof).sort_values("time")
        if WRITE_PROF_DIAG and not df.empty:
            prof_diag_rows.append(df)

        if df.empty:
            for d in dsets:
                d.close()
            continue

        years = sorted(df["year"].unique())
        if PRINT_DETAILED and years:
            # compact coverage summary
            counts = df.groupby(["year","month"]).size().reset_index(name="nprof")
            months_by_year = (
                counts.groupby("year")
                      .apply(lambda x: ", ".join(f"{int(m)}(n={int(n)})" for m, n in zip(x["month"], x["nprof"])))
            )
            print(f"[diag] Coverage for {wmo}:")
            for y in years:
                lines = months_by_year.get(y, "")
                print(f"  Year {y}: months present -> {lines}")

        for y in years:
            # Winter: Aug–Nov (year y) — pick the month containing profile with max column-mean
            Wcand = df[(df["year"]==y) & (df["month"].isin([8,9,10,11]))].copy()
            if Wcand.empty:
                continue
            print(df.head())
            iw = Wcand["col_mean"].idxmax()
            mw = int(Wcand.loc[iw, "month"])
            Wm = Wcand[Wcand["month"] == mw]

            # Summer: Dec(y)–Mar(y+1) — pick month containing profile with min column-mean
            Scand = df[((df["year"]==y) & (df["month"]==12)) | ((df["year"]==y+1) & (df["month"].isin([1,2,3])))]
            if Scand.empty:
                continue
            is_ = Scand["col_mean"].idxmin()
            ms = int(Scand.loc[is_, "month"])
            Sm = Scand[Scand["month"] == ms]

            if Wm.empty or Sm.empty:
                continue

            # Month "means" = average of per-profile integrals (no vertical regridding)
            Wints = Wm["integral"].to_numpy(dtype=float); Wnpts = Wm["npts"].to_numpy(dtype=float)
            Sints = Sm["integral"].to_numpy(dtype=float); Snpts = Sm["npts"].to_numpy(dtype=float)

            Wints = Wints[np.isfinite(Wints)]; Wnpts = Wnpts[np.isfinite(Wnpts)]
            Sints = Sints[np.isfinite(Sints)]; Snpts = Snpts[np.isfinite(Snpts)]

            if (Wints.size == 0) or (Sints.size == 0):
                if PRINT_QA:
                    print(f"[{wmo} {y}] skip: no finite integrals in winter or summer (W={Wints.size}, S={Sints.size})")
                continue

            Wmean = float(np.nanmean(Wints)); Smean = float(np.nanmean(Sints))
            Ninv = Wmean - Smean                        # mmol N m^-2
            ancp_mmolC = 6.6 * Ninv                     # mmol C m^-2
            ancp_molC  = ancp_mmolC / 1000.0            # mol C m^-2

            # --- detailed print for this season
            if PRINT_DETAILED:
                def _stat_str(a):
                    if a.size == 0: return "n=0"
                    return f"n={a.size}, min={np.nanmin(a):.0f}, med={np.nanmedian(a):.0f}, max={np.nanmax(a):.0f}"
                print(f"[diag] {wmo} season {y}: winter month={mw} (profiles={len(Wm)}), summer month={ms} (profiles={len(Sm)})")
                print(f"       winter npts (used by trapz): {_stat_str(Wnpts)};  summer npts: {_stat_str(Snpts)}")
                print(f"       winter mean integral={Wmean:.1f}  summer mean integral={Smean:.1f}  ΔN={Ninv:.1f} mmol N m^-2  ANCP={ancp_molC:.3f} mol C m^-2")

            rows.append(dict(
                float_wmo=wmo,
                season_calendar_year=y,
                winter_month=mw, summer_month=ms,
                ancp_molC_m2=ancp_molC,
                ancp_mmolC_m2=ancp_mmolC,
                n_prof_winter=int(Wm.shape[0]),
                n_prof_summer=int(Sm.shape[0]),
                warn_winter_lowN=int((Wm["npts"] < SPARSITY_WARN_N).sum()),
                warn_summer_lowN=int((Sm["npts"] < SPARSITY_WARN_N).sum()),
                winter_integral_mean=Wmean,
                summer_integral_mean=Smean,
                winter_npts_min=int(np.nanmin(Wnpts)) if Wnpts.size else np.nan,
                winter_npts_med=float(np.nanmedian(Wnpts)) if Wnpts.size else np.nan,
                winter_npts_max=int(np.nanmax(Wnpts)) if Wnpts.size else np.nan,
                summer_npts_min=int(np.nanmin(Snpts)) if Snpts.size else np.nan,
                summer_npts_med=float(np.nanmedian(Snpts)) if Snpts.size else np.nan,
                summer_npts_max=int(np.nanmax(Snpts)) if Snpts.size else np.nan
            ))

        for d in dsets:
            try: d.close()
            except Exception: pass

    # ---- finalize outputs ----
    cols = ["float_wmo","season_calendar_year","winter_month","summer_month",
            "ancp_molC_m2","ancp_mmolC_m2","n_prof_winter","n_prof_summer",
            "warn_winter_lowN","warn_summer_lowN",
            "winter_integral_mean","summer_integral_mean",
            "winter_npts_min","winter_npts_med","winter_npts_max",
            "summer_npts_min","summer_npts_med","summer_npts_max"]

    if not rows:
        print("[ANCP] No valid winter/summer pairs computed.")
        ancp = pd.DataFrame(columns=cols)
    else:
        ancp = (pd.DataFrame(rows)
                .sort_values(["float_wmo","season_calendar_year"])
                .reset_index(drop=True))

    print(f"[ANCP] Built {len(ancp)} rows from {ancp['float_wmo'].nunique() if not ancp.empty else 0} floats.")
    if not ancp.empty and PRINT_QA:
        show = ancp[["float_wmo","season_calendar_year","winter_month","summer_month",
                     "ancp_molC_m2","n_prof_winter","n_prof_summer",
                     "winter_npts_med","summer_npts_med",
                     "warn_winter_lowN","warn_summer_lowN"]].head(12)
        print(show.to_string(index=False))

    if OUT_CSV and not ancp.empty:
        ancp.to_csv(OUT_CSV, index=False)
        print("Wrote season table:", OUT_CSV)

    if WRITE_PROF_DIAG and prof_diag_rows:
        prof_all = pd.concat(prof_diag_rows, ignore_index=True)
        prof_all = prof_all.sort_values(["wmo","time"]).reset_index(drop=True)
        prof_all.to_csv(OUT_PROF_CSV, index=False)
        if PRINT_QA:
            print("Wrote per-profile diagnostics:", OUT_PROF_CSV)

    return ancp

# ---- run ----
ancp_table = ancp_johnson_table()
```

```python
# ===== Winter MLD vs following-season productivity (Chl, POC, ANCP) =====
# - Winter = months in WINTER_MONTHS for calendar year y (default: Jun–Sep)
# - Summer (following season for the same y) = Dec(y)–Mar(y+1)
# - MLD from N^2 maximum depth (same convention used before), magnitude in meters
# - Integrated Chl, POC: native depths, np.trapz to ZMAX (no regridding)
# - Optional join with ANCP CSV (season_calendar_year = y)
# ------------------------------------------------------------------------

import os, glob, re, warnings
import numpy as np
import pandas as pd
import xarray as xr
import gsw
import matplotlib.pyplot as plt

# ---------------- USER CONFIG ----------------
DIR = "/glade/u/home/vyaasv/NO3_Floats"           # directory with your per-float .nc files
ANCP_CSV = "/glade/u/home/vyaasv/NO3_Floats/ancp_johnson_0_200m.csv"  # set None to skip
WINTER_MONTHS = [6,7,8,9]                     # Jun–Sep
ZMAX = 200.0
K_POC_BBP700 = 13000.0                        # mg m^-3 per m^-1
QC_OK = ("0","1","2","8")                     # relaxed (optics-friendly)
PRINT_QA = True
# ---------------------------------------------

def _pick(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else (ds[base] if base in ds.variables else None)

def _qc_mask(ds, base, ok=QC_OK):
    for suf in ("_ADJUSTED_QC","_QC"):
        n = base + suf
        if n in ds.variables:
            return ds[n].astype("U1").isin(list(ok))
    return None

def _decode_time(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    try:
        return t.to_index()
    except Exception:
        return pd.to_datetime(t.values)

def _pressure_to_depth(p, lat):
    p = np.asarray(p, float)
    try:
        return -gsw.z_from_p(p, float(lat))   # +down (m)
    except Exception:
        return np.full_like(p, np.nan, float)

def _integrate0_200(v, z, zmax=ZMAX, clip_nonneg=True):
    v = np.asarray(v, float); z = np.asarray(z, float)
    m = np.isfinite(v) & np.isfinite(z) & (z >= 0) & (z <= zmax)
    if m.sum() < 2:
        return np.nan
    vv = np.clip(v[m], 0, np.inf) if clip_nonneg else v[m]
    zz = z[m]
    return float(np.trapz(vv, zz))

def _get_wmo_from_ds(ds, fname):
    if "PLATFORM_NUMBER" in ds.variables:
        try:
            raw = ds["PLATFORM_NUMBER"].values
            wmo = "".join(np.atleast_1d(raw)[0].astype(str)).strip()
            return re.sub(r"\s+","", wmo)
        except Exception:
            pass
    m = re.search(r"(\d{6,})", os.path.basename(fname))
    return m.group(1) if m else os.path.basename(fname)

def _compute_profile_MLD_N2_max(sp, tp, pp, lat, lon):
    # Conservative Temp & Absolute Salinity
    SA = gsw.SA_from_SP(sp, pp, lon, lat)
    CT = gsw.CT_from_t(SA, tp, pp)
    N2, pmid = gsw.Nsquared(SA, CT, pp)   # N^2 on mid-pressures
    if not np.isfinite(N2).any():
        return np.nan
    i = int(np.nanargmax(N2))
    zmid = -gsw.z_from_p(pmid, lat)       # +depth (m)
    return float(zmid[i]) if np.isfinite(zmid[i]) else np.nan

def _per_file_records(path):
    """Return DataFrame with per-profile time, month, year, |MLD|, int_chl, int_poc, and WMO."""
    out = []
    with xr.open_dataset(path) as ds:
        P   = _pick(ds, "PRES")
        T   = _pick(ds, "TEMP")
        S   = _pick(ds, "PSAL")
        LAT = ds["LATITUDE"]  if "LATITUDE"  in ds else None
        LON = ds["LONGITUDE"] if "LONGITUDE" in ds else None
        # Chl: prefer CPHL, else CHLA
        CHL = _pick(ds, "CPHL") if _pick(ds, "CPHL") is not None else _pick(ds, "CHLA")
        BBP = _pick(ds, "BBP700")

        if any(v is None for v in (P,T,S,LAT,LON,CHL,BBP)) or "TIME" not in ds:
            return pd.DataFrame([])

        # QC (relaxed by default)
        for da, base in ((P,"PRES"),(T,"TEMP"),(S,"PSAL"),
                         (CHL, CHL.name.split("_")[0]), (BBP,"BBP700")):
            qm = _qc_mask(ds, base, ok=QC_OK)
            if (qm is not None) and (tuple(qm.dims) == tuple(da.dims)):
                da[:] = da.where(qm)
        CHL = CHL.where(CHL >= 0)
        BBP = BBP.where(BBP >= 0)

        time = _decode_time(ds)
        nprof = int(P.sizes["TIME"])

        # find the non-TIME dim name (“DEPTH”/“N_LEVELS”/etc.)
        dname = [d for d in P.dims if d != "TIME"][0]

        for i in range(nprof):
            ti = time[i]
            year = int(ti.year); month = int(ti.month)
            lat = float(LAT.isel({LAT.dims[0]: i}).values)
            lon = float(LON.isel({LON.dims[0]: i}).values)

            pp  = np.asarray(P.isel(TIME=i).values, float)
            tp  = np.asarray(T.isel(TIME=i).values, float)
            sp  = np.asarray(S.isel(TIME=i).values, float)
            chl = np.asarray(CHL.isel(TIME=i).values, float)
            bbp = np.asarray(BBP.isel(TIME=i).values, float)

            z = _pressure_to_depth(pp, lat)

            # MLD (|depth|) using your N^2-max proxy
            mld = _compute_profile_MLD_N2_max(sp, tp, pp, lat, lon)
            mld_abs = float(np.abs(mld)) if np.isfinite(mld) else np.nan

            # Integrals to 200 m
            poc = ((3.12*(10**4)*bbp)+3.04)/12
            int_chl = _integrate0_200(chl, z, ZMAX, clip_nonneg=True)
            int_poc = _integrate0_200(poc, z, ZMAX, clip_nonneg=True)

            out.append(dict(
                wmo=_get_wmo_from_ds(ds, path),
                time=ti, year=year, month=month,
                mld_abs=mld_abs,
                int_chl=int_chl,
                int_poc=int_poc
            ))
    return pd.DataFrame(out)

# --------- build seasonal table (winter MLD vs following summer outcomes) ---------
def build_winterMLD_vs_outcomes():
    recs = []
    files = sorted(glob.glob(os.path.join(DIR, "*.nc")))
    if PRINT_QA:
        print(f"[winter→summer] scanning {len(files)} files in {DIR}")

    for p in files:
        df = _per_file_records(p)
        if df.empty:
            if PRINT_QA: print(f"  SKIP (no usable vars): {os.path.basename(p)}")
            continue
        wmo = df["wmo"].iloc[0]

        # group by winter year y (calendar year)
        years = sorted(df["year"].unique())
        for y in years:
            # WINTER window: months are in current year y
            W = df[(df["year"]==y) & (df["month"].isin(WINTER_MONTHS))]
            if W.empty or not np.isfinite(W["mld_abs"]).any():
                continue
            winter_mld_mean = float(np.nanmean(W["mld_abs"]))
            winter_mld_n    = int(np.isfinite(W["mld_abs"]).sum())

            # SUMMER (following season for winter y): Dec(y)–Mar(y+1)
            S = pd.concat([
                df[(df["year"]==y)   & (df["month"]==12)],
                df[(df["year"]==y+1) & (df["month"].isin([1,2,3]))]
            ], ignore_index=True)
            if S.empty:
                continue

            # keep finite integrals
            ch = S["int_chl"].to_numpy(float)
            pc = S["int_poc"].to_numpy(float)
            ch = ch[np.isfinite(ch)]
            pc = pc[np.isfinite(pc)]

            if ch.size==0 and pc.size==0:
                continue

            out_row = dict(
                float_wmo=wmo,
                winter_year=y,
                winter_mld_mean=winter_mld_mean,
                winter_mld_n=winter_mld_n,
                summer_chl_mean=float(np.nanmean(ch)) if ch.size else np.nan,
                summer_chl_max=float(np.nanmax(ch))   if ch.size else np.nan,
                summer_poc_mean=float(np.nanmean(pc)) if pc.size else np.nan,
                summer_poc_max=float(np.nanmax(pc))   if pc.size else np.nan,
                summer_nprof=int(S.shape[0])
            )
            recs.append(out_row)

    tab = pd.DataFrame(recs).sort_values(["float_wmo","winter_year"]).reset_index(drop=True)

    # ---- optional ANCP join (season_calendar_year = winter_year) ----
    if ANCP_CSV and os.path.exists(ANCP_CSV) and not tab.empty:
        # read with float_wmo as string so we don't lose leading zeros etc.
        ancp = pd.read_csv(ANCP_CSV, dtype={"float_wmo": str})
        keep = [c for c in ["float_wmo","season_calendar_year","ancp_molC_m2","ancp_mmolC_m2"] if c in ancp.columns]
        ancp = ancp[keep].rename(columns={"season_calendar_year":"winter_year"})

        # make sure merge keys match dtypes on BOTH sides
        tab["float_wmo"]   = tab["float_wmo"].astype(str)
        tab["winter_year"] = pd.to_numeric(tab["winter_year"], errors="coerce").astype("Int64")
        ancp["winter_year"] = pd.to_numeric(ancp["winter_year"], errors="coerce").astype("Int64")

        tab = tab.merge(ancp, on=["float_wmo","winter_year"], how="left")
    return tab


table = build_winterMLD_vs_outcomes()
print(f"[winter→summer] built {len(table)} rows from {table['float_wmo'].nunique() if not table.empty else 0} floats")
print(table.head(12).to_string(index=False))

# --------- quick correlations & an example plot with regression line ---------
def quick_correlations(df):
    cols = [
        ("summer_chl_mean", "Chl (integrated) mean"),
        ("summer_chl_max",  "Chl (integrated) max"),
        ("summer_poc_mean", "POC (integrated) mean"),
        ("summer_poc_max",  "POC (integrated) max"),
        ("ancp_molC_m2",    "ANCP (mol C m$^{-2}$)")
    ]
    for c, label in cols:
        if c not in df.columns:
            continue
        m = np.isfinite(df["winter_mld_mean"]) & np.isfinite(df[c])
        x = df.loc[m, "winter_mld_mean"].to_numpy(float)
        y = df.loc[m, c].to_numpy(float)
        if x.size >= 3:
            r = np.corrcoef(x, y)[0,1]
            print(f"r={r:.2f}, R^2={r**2:.2f}  (winter MLD vs {label}, n={x.size})")
        else:
            print(f"(winter MLD vs {label}) — not enough paired points.")

def scatter_with_reg(df, ycol, title=None):
    m = np.isfinite(df["winter_mld_mean"]) & np.isfinite(df[ycol])
    if m.sum() < 3:
        print(f"Not enough points to plot {ycol}")
        return
    x = df.loc[m, "winter_mld_mean"].to_numpy(float)
    y = df.loc[m, ycol].to_numpy(float)

    # regression
    p = np.polyfit(x, y, 1)
    xx = np.linspace(np.nanmin(x), np.nanmax(x), 200)
    yy = p[0]*xx + p[1]
    r = np.corrcoef(x, y)[0,1]

    fig, ax = plt.subplots(figsize=(6,4.6))
    ax.scatter(x, y, facecolors='none', edgecolors='C0', s=36, label="seasons")  # hollow markers
    ax.plot(xx, yy, lw=1.8, label=f"y = {p[0]:.3g} x + {p[1]:.3g}")
    ax.set_xlabel("Winter mean MLD (m, Jun–Sep)")
    ax.set_ylabel(ycol.replace("_"," "))
    ax.grid(alpha=0.3)
    ax.set_title(title or f"Winter MLD vs {ycol}  (R²={r**2:.2f}, n={x.size})")
    ax.legend()
    plt.tight_layout()
    plt.show()

# Run some quick numbers and one example figure:
if not table.empty:
    quick_correlations(table)
    # Example: winter MLD vs summer integrated Chl mean
    scatter_with_reg(table, "summer_chl_mean", title="Winter MLD vs Summer Integrated Chl (mean)")
else:
    print("No rows built — check variable availability and QC settings.")

to_plot = [
    ("summer_chl_mean", "Winter MLD vs Summer Integrated Chl"),
    ("summer_chl_max",  "Winter MLD vs Summer Max Chl"),
    ("summer_poc_mean", "Winter MLD vs Summer Integrated POC"),
    ("summer_poc_max",  "Winter MLD vs Summer Max POC"),
    ("ancp_molC_m2",    "Winter MLD vs ANCP (mol C m$^{-2}$)"),
]
print("plotting")
for col, ttl in to_plot:
    if col in table.columns:
        scatter_with_reg(table, col, title=ttl)
```

Code for plots that remove outliers

```python
# ---------- Outlier filters + cleaned plotting ----------

import numpy as np
import pandas as pd

def remove_outliers(
    df,
    ycol,
    xcol="winter_mld_mean",
    method="iqr",       # "iqr" or "percentile"
    k=1.5,              # IQR multiplier for "iqr"
    lo=1.0, hi=99.0,    # percentiles for "percentile"
    on="both"           # "y", "x", or "both"
):
    """
    Return a filtered copy of df with outliers removed.
    Finite-mask is applied first, then the chosen method.
    """
    d = df.copy()

    # start with finite mask
    m = np.isfinite(d[xcol].to_numpy(float)) & np.isfinite(d[ycol].to_numpy(float))

    def _mask_iqr(series):
        q1 = np.nanpercentile(series, 25)
        q3 = np.nanpercentile(series, 75)
        iqr = q3 - q1
        lo_b = q1 - k * iqr
        hi_b = q3 + k * iqr
        return (series >= lo_b) & (series <= hi_b)

    def _mask_pct(series):
        lo_b = np.nanpercentile(series, lo)
        hi_b = np.nanpercentile(series, hi)
        return (series >= lo_b) & (series <= hi_b)

    if method not in ("iqr", "percentile"):
        raise ValueError("method must be 'iqr' or 'percentile'")

    # choose which columns to filter on
    targets = []
    if on in ("y", "both"):   targets.append(ycol)
    if on in ("x", "both"):   targets.append(xcol)
    if on not in ("y", "x", "both"):
        raise ValueError("on must be 'y', 'x', or 'both'")

    # build masks
    for col in targets:
        vals = d.loc[m, col].to_numpy(float)
        if vals.size < 4:
            # too few points to form a sensible outlier rule; skip
            continue
        if method == "iqr":
            keep = _mask_iqr(vals)
        else:
            keep = _mask_pct(vals)
        # update global mask only at the rows already marked True
        idx_true = np.where(m)[0]
        m[idx_true] &= keep

    out = d.loc[m].copy()
    dropped = int(len(d) - len(out))
    print(f"[outlier-filter] kept {len(out)} / {len(d)} rows (dropped {dropped}) using {method} on={on}")
    return out

def scatter_with_reg_clean(
    df, ycol, xcol="winter_mld_mean",
    method="iqr", k=1.5, lo=1.0, hi=99.0, on="both",
    title=None
):
    """
    Filter outliers, then call your existing scatter_with_reg.
    """
    dclean = remove_outliers(df, ycol, xcol=xcol, method=method, k=k, lo=lo, hi=hi, on=on)
    if dclean.empty:
        print(f"No points left after filtering for {ycol}.")
        return
    scatter_with_reg(dclean, ycol, title=title)

# ---------- Examples: make the same figures, but outlier-filtered ----------

to_plot = [
    ("summer_chl_mean", "Winter MLD vs Summer Integrated Chl  — IQR filtered"),
    ("summer_chl_max",  "Winter MLD vs Summer Max Chl  — IQR filtered"),
    ("summer_poc_mean", "Winter MLD vs Summer Integrated POC  — IQR filtered"),
    ("summer_poc_max",  "Winter MLD vs Summer Max POC  — IQR filtered"),
    ("ancp_molC_m2",    "Winter MLD vs ANCP (mol C m$^{-2}$) — IQR filtered"),
]

for col, ttl in to_plot:
    if col in table.columns:
        # IQR on both axes (k=1.5 is Tukey's rule). Change 'on' to "y" if you only want to clip y.
        scatter_with_reg_clean(
            table, col, xcol="winter_mld_mean",
            method="iqr", k=1.5, on="both",
            title=ttl
        )

# --- If you prefer percentile trimming instead (e.g., keep 1st–99th percentile on both axes):
for col, ttl in to_plot:
    if col in table.columns:
        scatter_with_reg_clean(
            table, col, xcol="winter_mld_mean",
            method="percentile", lo=1.0, hi=99.0, on="both",
            title=ttl.replace("IQR", "P1–P99")
        )
```

```python
# Melt onset + Growth Initiation (GI) detector — strict in-ML chlorophyll
# with diagnostics, signal-strength tie-break for melt, and plotting helpers

import os, glob
import numpy as np
import xarray as xr
import pandas as pd
import gsw
from scipy.signal import argrelextrema as lmm
import scipy.signal as signal
import matplotlib.pyplot as plt

# ---------------- small helpers ---------------- #
def safe_interp_1d(arr):
    out = np.asarray(arr, dtype=float).copy()
    nans = np.isnan(out)
    good = ~nans
    if good.sum() >= 2:
        idx = np.flatnonzero
        out[nans] = np.interp(idx(nans), idx(good), out[good])
    return out

def safe_filtfilt(B, A, arr):
    a = np.asarray(arr, dtype=float)
    if np.all(np.isfinite(a)) and a.size > 6:
        try:
            return signal.filtfilt(B, A, a, method="pad")
        except Exception:
            return a
    return a

def pick_var(ds, base):
    adj = base + "_ADJUSTED"
    return ds[adj] if adj in ds.variables else ds[base]

def good_qc_mask(ds, base, qc_ok=("0","1","2","8")):
    qname = base + "_ADJUSTED_QC" if base + "_ADJUSTED_QC" in ds.variables else base + "_QC"
    if qname not in ds.variables:
        return None
    qc = ds[qname].astype("U1")
    return qc.isin(list(qc_ok))

def get_time_index(ds):
    dsd = xr.decode_cf(ds)
    t = dsd["TIME"]
    ti = t.to_index() if hasattr(t, "to_index") else pd.to_datetime(t.values)
    if hasattr(ti, "to_datetimeindex"):
        return ti.to_datetimeindex()
    return pd.DatetimeIndex(ti)

# fallback GI when extrema fail

def gi_fallback_from_increase(chls):
    a = np.asarray(chls, float)
    if np.isfinite(a).sum() < 5:
        return None
    a = safe_interp_1d(a)
    if len(a) >= 3:
        kernel = np.ones(3)/3.0
        a_sm = np.convolve(a, kernel, mode="same")
    else:
        a_sm = a.copy()
    d = np.diff(a_sm)
    finite = np.isfinite(a_sm)
    if finite.sum() >= 2:
        std = np.nanstd(a_sm)
        amin = np.nanmin(a_sm)
        thresh = amin + 0.5*std if np.isfinite(std) else amin
    else:
        thresh = np.nanmin(a_sm[finite]) if finite.any() else 0.0
    for i in range(len(d)-1):
        if d[i] > 0 and d[i+1] > 0 and a_sm[i+1] >= thresh:
            return i+1
    for i in range(len(d)):
        if d[i] > 0:
            return i+1
    return None

# ---------------- main detector ---------------- #

def save_meltbloom2(
    rtc=0.1,                   # freezing-point margin for under-ice (paper default)
    bl=0,                      # levels below MLD for local averaging window center
    DIR="",
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    chl_min_points_in_ml=3,    # require at least N points within ML for chl average
    ml_top_pad_m=0.0,          # allow small pad above surface if vertical grid is coarse
    debug=True,
    collect_diag=True
):
    def frac_finite(arr):
        arr = np.asarray(arr)
        return 100.0 * np.isfinite(arr).sum() / arr.size if arr.size else 0.0

    flist = glob.glob(os.path.join(DIR, "*.nc"))
    print("No. of files:", len(flist))

    bloomdays  = np.array([]); bloomweeks = np.array([]); bloomweeks_m = np.array([])
    bloomyears = np.array([]); meltweeks  = np.array([]); meltdays     = np.array([])
    lons = np.array([]); lats = np.array([]); files = np.array([]); mld_gi = np.array([])
    mld_chl_pos = np.array([]); timing_pg = np.array([]); grth_rate_gm = np.array([])
    peak_chl = np.array([]); len_ts = np.array([]); gia_week = np.array([])

    diag = {}

    # filters
    N = 1; Wn, Wn2 = 0.1, 0.2
    B, A  = signal.butter(N, Wn,  output="ba")
    B2, A2 = signal.butter(N, Wn2, output="ba")

    for f, path in enumerate(flist):
        print(f"\n=== File {f} {os.path.basename(path)} ===")
        ds = xr.open_dataset(path)
        if not all(v in ds.variables for v in ("TIME","LONGITUDE","LATITUDE")):
            print("  STOP: missing TIME/LONGITUDE/LATITUDE"); ds.close(); continue

        # pick variables (CF/Argo)
        t  = pick_var(ds, "TEMP")
        s  = pick_var(ds, "PSAL")
        pr = pick_var(ds, "PRES")
        if ("CPHL" in ds.variables) or ("CPHL_ADJUSTED" in ds.variables):
            chl = pick_var(ds, "CPHL")
        else:
            print("  STOP: no CPHL/CPHL_ADJUSTED"); ds.close(); continue

        lon = ds["LONGITUDE"]; lat = ds["LATITUDE"]
        time = get_time_index(ds)

        lon_dim = lon.dims[0]; lat_dim = lat.dims[0]
        if "TIME" not in t.sizes:
            print("  STOP: TEMP has no TIME dim"); ds.close(); continue
        nprof = int(t.sizes["TIME"])
        if lon.sizes[lon_dim] != nprof or lat.sizes[lat_dim] != nprof:
            print("  STOP: lon/lat length mismatch vs TIME"); ds.close(); continue

        # QC and screens
        for da, base in [(t,"TEMP"),(s,"PSAL"),(pr,"PRES"),(chl,"CPHL")]:
            m = good_qc_mask(ds, base, qc_ok=qc_ok)
            if m is not None and da.dims == m.dims:
                da[:] = da.where(m)
        s   = s.where(s > 28.0)
        chl = chl.where(chl >= 0)

        # duration and gaps
        if nprof >= 2:
            step_days = np.diff(time.values) / np.timedelta64(1, "D")
            tsmax = float(np.nanmax(step_days)) if len(step_days) else np.nan
            span_days = float((time[-1] - time[0]).days)
        else:
            tsmax = np.nan; span_days = np.nan
        print(f"  profiles: {nprof}, time span days: {span_days:.1f}, max gap days: {tsmax if np.isfinite(tsmax) else np.nan:.1f}")

        long_enough = (
            (np.isfinite(tsmax) and tsmax > 20 and nprof > 48) or
            (np.isfinite(tsmax) and tsmax <= 20 and nprof >= 24)
        )
        if not long_enough:
            print("  STOP: record too short or too gappy for seasonal analysis"); ds.close(); continue

        # series
        rl = [None] * nprof
        pi = np.zeros(nprof)
        ts = np.array([]); ss = np.array([]); chls = np.array([]); mlds = np.array([])
        tdiff_series = []; tns_series = []; sns_series = []; tf_ns_series = []

        used_in_ml = used_shallow_fallback = used_any_deep = used_no_chl = 0

        for p in range(nprof):
            lonp = float(lon.isel({lon_dim: p}).values)
            latp = float(lat.isel({lat_dim: p}).values)
            pp   = np.asarray(pr.isel(TIME=p).values, dtype=float)
            chlp = np.asarray(chl.isel(TIME=p).values, dtype=float)
            tp   = np.asarray(t.isel(TIME=p).values, dtype=float)
            sp   = np.asarray(s.isel(TIME=p).values, dtype=float)

            dp = gsw.z_from_p(pp, latp); zpos = -dp
            if np.isnan(np.nanmean(dp)):
                dp = gsw.z_from_p(pp, float(lat.mean().values)); zpos = -dp

            spa = gsw.SA_from_SP(sp, pp, lonp, latp)
            if np.isnan(np.nanmean(spa)):
                spa = gsw.SA_from_SP(sp, pp, float(lon.mean().values), float(lat.mean().values))
            ct  = gsw.CT_from_t(spa, tp, pp)
            n2, pmid = gsw.Nsquared(spa, ct, pp)
            dmid = gsw.z_from_p(pmid, latp)
            if np.isnan(np.nanmean(dmid)):
                dmid = gsw.z_from_p(pmid, float(lat.mean().values))

            tfreeze = gsw.t_freezing(spa, 0.0, 0.0)  # p=0 dbar
            try:
                i0 = int(np.nanargmin(pp))
            except Exception:
                i0 = 0

            # MLD pick from max N2 level
            try:
                mld = dmid[int(np.nanargmax(n2))]  # negative depth
                mlds = np.append(mlds, mld)

                # near-surface averages for T and S (10 dbar window, fallback 20 dbar)
                ddl = dp[i0] - 10.0
                ddi = int(np.nanargmin(np.abs(dp - ddl)))
                lo = min(ddi, i0); hi = max(ddi, i0) + 1
                vT = np.nanmean(tp[lo:hi]); vS = np.nanmean(sp[lo:hi])
                if np.isnan(vT) or np.isnan(vS):
                    p0 = pp[i0]
                    near_idx = np.where((pp - p0) <= 20.0)[0]
                    if near_idx.size:
                        if np.isnan(vT): vT = np.nanmean(tp[near_idx])
                        if np.isnan(vS): vS = np.nanmean(sp[near_idx])
                ts = np.append(ts, vT); ss = np.append(ss, vS)
                tns_series.append(vT);   sns_series.append(vS)

                # strict in-ML chlorophyll averaging
                mld_mag = float(np.abs(mld))
                ml_mask = (zpos >= 0.0 - ml_top_pad_m) & (zpos <= mld_mag)
                ml_idx = np.where(ml_mask & np.isfinite(chlp))[0]
                vC = np.nan
                if ml_idx.size >= chl_min_points_in_ml:
                    vC = float(np.nanmean(chlp[ml_idx]))
                    used_in_ml += 1
                else:
                    # shallow fallback: use the shallowest finite chl point if ML is empty
                    shallow_idx = np.where(np.isfinite(chlp) & (zpos <= 100.0))[0]
                    if shallow_idx.size > 0:
                        vC = float(np.nanmean(chlp[shallow_idx]))
                        used_shallow_fallback += 1
                    else:
                        deep_idx = np.where(np.isfinite(chlp))[0]
                        if deep_idx.size > 0:
                            vC = float(np.nanmean(chlp[deep_idx]))
                            used_any_deep += 1
                        else:
                            used_no_chl += 1
                chls = np.append(chls, vC)
            except Exception:
                mlds = np.append(mlds, np.nan)
                ts   = np.append(ts, np.nan)
                ss   = np.append(ss, np.nan)
                chls = np.append(chls, np.nan)
                tns_series.append(np.nan); sns_series.append(np.nan)

            # T - Tf at near-surface
            tf_ns = float(tfreeze[i0]) if np.isfinite(tfreeze[i0]) else np.nan
            tf_ns_series.append(tf_ns)
            try:
                tdiff = float(vT - tf_ns)
            except Exception:
                tdiff = np.inf
            tdiff_series.append(tdiff)

            if np.isfinite(tdiff) and (tdiff <= rtc):
                rl[p] = 1; pi[p] = 1
            else:
                rl[p] = 0

        # seasonal axes
        week = pd.Index(time).isocalendar().week.to_numpy()
        day  = pd.Index(time).dayofyear.to_numpy()
        year = pd.Index(time).year.to_numpy()
        week = np.where(week <= 13, week + 52, week)
        day  = np.where(day  <= 89,  day  + 365, day)

        print(f"  built series len: ts={len(ts)} ss={len(ss)} chl={len(chls)}  finite% ts:{frac_finite(ts):.1f} ss:{frac_finite(ss):.1f} chl:{frac_finite(chls):.1f}")
        print(f"  CHL gate counts: in-ML:{used_in_ml} shallow_fallback<=100m:{used_shallow_fallback} deep_any:{used_any_deep} no_chl:{used_no_chl}")

        # safe interp + filters
        ts_i   = safe_interp_1d(ts);   ss_i   = safe_interp_1d(ss);   chls_i = safe_interp_1d(chls)
        chlsf = safe_filtfilt(B, A, chls_i)
        tsf   = safe_filtfilt(B, A, ts_i)
        ssf   = safe_filtfilt(B, A, ss_i)
        ssf2  = safe_filtfilt(B2, A2, ss_i)

        # ============= melt onset ============= #
        mis = np.array([], dtype=int)
        uice_ct = int(np.sum(pi == 1))
        print(f"  under-ice profiles (T-Tf <= {rtc}): {uice_ct}")
        if uice_ct >= 3:
            dpi = pi[1:] - pi[0:-1]
            mi  = np.where(dpi == -1)[0]
            print(f"  melt candidates (under-ice→open): {len(mi)}")
            # require at least two prior under-ice profiles
            keep = []
            for idx in mi:
                nundi = int(np.sum(pi[max(0, idx - 15): idx] == 1))
                if nundi >= 2:
                    keep.append(idx)
            mi = np.array(keep, dtype=int)
            print(f"   after '>=2 prior under-ice' check: {len(mi)}")

            # compute salinity and temperature gradients following each candidate
            if mi.size >= 1:
                dsalta = []
                dtempa = []
                for k, m_idx in enumerate(mi):
                    # one month after candidate
                    salt_seg = ssf2[m_idx:m_idx+4]
                    temp_seg = tsf[m_idx:m_idx+4]
                    dsalt = np.nanmean(salt_seg[1:] - salt_seg[0:-1])
                    dtemp = np.nanmean(temp_seg[1:] - temp_seg[0:-1])
                    dsalta.append(dsalt)
                    dtempa.append(dtemp)
                dsalta = np.asarray(dsalta)
                dtempa = np.asarray(dtempa)

                # seasonal spacing filter first
                if mi.size > 1:
                    dmi = np.abs(mi[1:] - mi[0:-1])
                else:
                    dmi = np.array([])

                # base melt acceptance: seasonal spacing and salinity drop
                base_accept = []
                for nm, m_idx in enumerate(mi):
                    dsalt = dsalta[nm]
                    cond = (nm < len(dmi) and dmi[nm] > 12 and dsalt < 0.0) or (nm == len(dmi) and dsalt < 0.0)
                    if cond:
                        base_accept.append(m_idx)
                mis = np.array(base_accept, dtype=int)
                print(f"   after salinity-drop & spacing check: {len(mis)}")

                # resolve consecutive candidates closer than 12 using signal strength
                # define signal = (-dsalt) * max(dtemp, 0) so larger is stronger if salinity drops and temperature rises
                if mis.size > 1:
                    mis_sorted = np.sort(mis)
                    # compute signals for accepted candidates
                    sig = []
                    for m_idx in mis_sorted:
                        nm = int(np.where(mi == m_idx)[0][0])
                        dsalt = dsalta[nm]
                        dtemp = dtempa[nm]
                        sgnl = max(-dsalt, 0.0) * max(dtemp, 0.0)
                        sig.append(sgnl)
                    sig = np.asarray(sig)
                    # walk through, dropping the weaker one when consecutive are < 12 apart
                    keep_inds = [0]
                    for j in range(1, len(mis_sorted)):
                        if (mis_sorted[j] - mis_sorted[j-1]) < 12:
                            # compare signals
                            if sig[j] >= sig[j-1]:
                                # drop j-1 (replace last keep with j)
                                keep_inds[-1] = j
                            else:
                                # keep j-1, drop j
                                pass
                        else:
                            keep_inds.append(j)
                    mis = mis_sorted[keep_inds]
                print(f"  FINAL melt picks: {len(mis)}")
        else:
            print("  NOTE: fewer than 3 under-ice profiles -> melt detection disabled")

        # ============= GI detection ============= #
        finite_chls = np.isfinite(chls_i)
        ch_std = np.nanstd(chls_i) if finite_chls.sum() >= 2 else 0.0
        chl_anom = chls_i - (ch_std if np.isfinite(ch_std) else 0.0)

        lmax = lmm(chlsf, np.greater)[0]
        lmin = lmm(chlsf, np.less)[0]
        print(f"  GI extrema counts: lmin={len(lmin)} lmax={len(lmax)}  finite% chl(after interp)={frac_finite(chls_i):.1f}")

        bloomday_f = np.array([]); bloomweek_f = np.array([]); bloomweek_m_f = np.array([])
        blm_inds = np.array([]); mld_a = np.array([]); mld_fg = np.array([]); pgweek = np.array([]); gia_a = np.array([])
        gi_found_segments = 0

        if len(lmin) > 0 and len(lmax) > 0:
            try:
                dlmin = len(chlsf) - lmin[-1]
            except IndexError:
                dlmin, lmin = lmax[0], np.array([0])
            try:
                dlmax = lmax[0]
            except IndexError:
                dlmax, lmax = len(chlsf)-1, np.array([len(chlsf)-1])

            nlmax, nlmin = len(lmax), len(lmin)
            if nlmin > nlmax and dlmin < 6: lmin = np.delete(lmin, -1)
            elif nlmin > nlmax and dlmin >= 6:
                chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe)
            elif nlmin < nlmax and dlmax < 9: lmax = np.delete(lmax, 0)
            elif nlmin < nlmax and dlmax >= 9: lmin = np.concatenate(([0], lmin))
            elif nlmin == nlmax and dlmin >= 6:
                dlmm = lmax - lmin
                if np.sum(dlmm < 0) > 0:
                    chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                    lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe); lmax = np.delete(lmax, 0)
            elif nlmin == nlmax and dlmin < 6:
                lmax = np.delete(lmax, 0); lmin = np.delete(lmin, -1)

            for l in range(len(lmax)):
                imax, imin = int(lmax[l]), int(lmin[l])
                if (imin < imax) and (abs(imax - imin) > 3):
                    chlmm = chls_i[imin:imax]
                    peak_chl = np.append(peak_chl, np.nanmax(chlmm))
                    weekmm = week[imin:imax]; daymm = day[imin:imax]
                    dchl = chlmm[1:] - chlmm[0:-1]
                    dchl = np.where(dchl > 0.0, dchl, np.nan)
                    medchl = np.nanmedian(dchl)
                    blmind = int(np.nanargmax(dchl > medchl))
                    blmind_m = int(np.nanargmax(dchl))
                    chl_fi = int(np.nanargmax(dchl > 0.0))

                    blm_inds = np.append(blm_inds, imin + blmind)
                    bloomweek_f = np.append(bloomweek_f, weekmm[blmind])
                    bloomweek_m_f = np.append(bloomweek_m_f, weekmm[blmind_m])
                    bloomday_f  = np.append(bloomday_f,  daymm[blmind])

                    mldmm = mlds[imin:imax]
                    mld_a = np.append(mld_a, mldmm[blmind])
                    mld_fg = np.append(mld_fg, mldmm[chl_fi])
                    pgweek = np.append(pgweek, weekmm[chl_fi])

                    try:
                        gia_a = np.append(gia_a, weekmm[np.min(np.where(chl_anom[imin:imax] > 0))])
                    except ValueError:
                        gia_a = np.append(gia_a, np.nan)

                    gi_found_segments += 1

        used_fallback = False
        if gi_found_segments == 0:
            gi_idx = gi_fallback_from_increase(chls_i)
            used_fallback = gi_idx is not None
            print(f"  GI fallback used: {used_fallback}  idx:{gi_idx if gi_idx is not None else 'None'}")
            if gi_idx is not None:
                bloomday_f   = np.append(bloomday_f,  day[gi_idx])
                bloomweek_f  = np.append(bloomweek_f, week[gi_idx])
                bloomweek_m_f = np.append(bloomweek_m_f, week[gi_idx])
                mld_a = np.append(mld_a, mlds[gi_idx] if len(mlds) > gi_idx else np.nan)
                pgweek = np.append(pgweek, week[gi_idx])
                blm_inds = np.append(blm_inds, gi_idx)
                gia_week = np.append(gia_week, week[gi_idx])
        print(f"  GI picks: {len(bloomweek_f)}  (segments:{gi_found_segments}, fallback:{used_fallback})")

        # ============= pairing and export ============= #
        if (mis.size == 0) and gi_only_if_no_ice and (len(bloomweek_f) > 0):
            print("  Exporting GI-only rows (no melt detected)")
            for e in range(len(bloomweek_f)):
                bloomdays  = np.append(bloomdays,  bloomday_f[e])
                bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                bloomyears = np.append(bloomyears, year[0])
                meltweeks  = np.append(meltweeks, np.nan)
                meltdays   = np.append(meltdays,  np.nan)
                idx_for_loc = int(blm_inds[e])
                lons = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                lats = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                files = np.append(files, f)
                mld_gi = np.append(mld_gi, mld_a[e])
                mld_chl_pos = np.append(mld_chl_pos, np.nan)
                timing_pg  = np.append(timing_pg, pgweek[e] if len(pgweek) > e else np.nan)
        else:
            meltday = np.array([day[m] for m in mis.astype(int)])
            pair_ct = 0
            for e in range(min(mis.size, len(bloomday_f))):
                if abs(float(meltday[e]) - float(bloomday_f[e])) <= 90:
                    pair_ct += 1
            print(f"  Potential GI–melt pairs within 90 days: {pair_ct}")

            for e in range(min(len(meltday), len(bloomday_f))):
                dd = abs(meltday[e] - bloomday_f[e])
                if dd <= 90:
                    bloomdays  = np.append(bloomdays,  bloomday_f[e])
                    bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                    bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                    bloomyears = np.append(bloomyears, year[0])
                    meltweeks  = np.append(meltweeks,  int(np.array([week[m] for m in mis.astype(int)])[e]))
                    meltdays   = np.append(meltdays,   meltday[e])
                    idx_for_loc = int(blm_inds[e])
                    lons = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                    lats = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                    files = np.append(files, f)
                    mld_gi = np.append(mld_gi, mld_a[e])
                    mld_chl_pos = np.append(mld_chl_pos, np.nan)
                    timing_pg   = np.append(timing_pg,   pgweek[e] if e < len(pgweek) else np.nan)

        len_ts = np.append(len_ts, len(chls_i))

        if collect_diag:
            diag[f] = {
                "path": path,
                "time": time,
                "week": week, "day": day, "year": year,
                "lon": lon.values.astype(float),
                "lat": lat.values.astype(float),
                "ts_raw": ts, "ss_raw": ss, "chl_raw": chls,
                "ts_f": tsf, "ss_f": ssf, "chl_f": chlsf,
                "mld": mlds,
                "pi": pi.astype(int),
                "tdiff": np.asarray(tdiff_series, float),
                "tf_ns": np.asarray(tf_ns_series, float),
                "melt_idx": np.asarray(mis, int),
                "gi_idx": np.asarray(blm_inds, int) if len(blm_inds) else np.array([], dtype=int),
                "notes": {
                    "rtc": float(rtc),
                    "chl_min_points_in_ml": int(chl_min_points_in_ml),
                    "used_in_ml": int(used_in_ml),
                    "used_shallow_fallback": int(used_shallow_fallback),
                    "used_any_deep": int(used_any_deep),
                    "used_no_chl": int(used_no_chl),
                }
            }

        ds.close()

    print("\n=== Summary across files ===")
    print("Avg. length of time series:", np.nanmean(len_ts) if len(len_ts) else np.nan)
    print("Avg. peak chlorophyll:",     np.nanmean(peak_chl) if len(peak_chl) else np.nan)
    print("Avg. MLD at GI:",            np.nanmean(mld_gi) if len(mld_gi) else np.nan)
    print("Avg. Timing of First Positive Chl:", np.nanmean(timing_pg) if len(timing_pg) else np.nan)

    return (
        bloomweeks, bloomdays, bloomweeks_m, bloomyears,
        meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag
    )

# convenience padding and write-out helper

def _pad_to_len(a, L):
    a = np.asarray(a); n = a.size
    if n == L: return a
    if n > L:  return a[:L]
    pad = np.full(L - n, np.nan)
    return np.concatenate([a, pad])


def write_out_nc(DIR, res):
    (bloomweeks, bloomdays, bloomweeks_m, bloomyears,
     meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag) = res

    all_arrays = {
        "bloomweeks":     bloomweeks,
        "bloomdays":      bloomdays,
        "bloomweeks_max": bloomweeks_m,
        "bloomyears":     bloomyears,
        "meltweeks":      meltweeks,
        "meltdays":       meltdays,
        "mld_gi":         mld_gi,
        "lons":           lons,
        "lats":           lats,
        "files":          files,
        "gia_week":       gia_week,
    }
    L = max(len(v) for v in all_arrays.values()) if all_arrays else 0
    all_arrays = {k: _pad_to_len(v, L) for k, v in all_arrays.items()}

    dsb = xr.Dataset({k: (["e"], v) for k, v in all_arrays.items()})
    out_path = os.path.join(DIR, "out.nc")
    dsb.to_netcdf(out_path, format="NETCDF4_CLASSIC")
    print("Wrote:", out_path)

# plotting helpers from your previous version can be reused unchanged


# ---------------- diagnostics + plotting ---------------- #

def list_diag_summary(diag_dict):
    """Quick table of which floats have picks."""
    rows = []
    for f, d in diag_dict.items():
        rows.append((f, os.path.basename(d["path"]), len(d["gi_idx"]), len(d["melt_idx"])) )
    rows = sorted(rows, key=lambda x: x[0])
    print("f | file | GI_picks | Melt_picks")
    for r in rows:
        print(f"{r[0]:2d} | {r[1]} | {r[2]:8d} | {r[3]:10d}")

from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import matplotlib.dates as mdates

def _proxy_lines():
    gi_proxy   = Line2D([0], [0], lw=1.4, linestyle="-",  label="GI")
    melt_proxy = Line2D([0], [0], lw=1.4, linestyle="--", label="Melt")
    return gi_proxy, melt_proxy

def _add_event_lines(ax, t, gi_idx, melt_idx, color="C3"):
    for i in gi_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="-")
    for i in melt_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="--")

def _apply_year_ticks(ax, label="Year"):
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    ax.set_xlabel(label)

def plot_float_panels(diag_dict, f, title=None):
    """
    Make THREE SEPARATE FIGURES (not subplots), each with its own x-label:
      1) CHL (raw + filtered) with GI (solid) and Melt (dashed) markers
      2) Near-surface T and T_freeze, right axis T − Tf, under-ice shading
      3) Near-surface S (left) and |MLD| (right)
    """
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")

    d = diag_dict[f]
    t = pd.DatetimeIndex(d["time"])
    gi_idx  = np.asarray(d["gi_idx"], int)
    melt_idx = np.asarray(d["melt_idx"], int)
    rtc = float(d["notes"].get("rtc", 0.1))
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"
    plt.rcParams.update({"figure.dpi": 130})

    # Fig 1: CHL
    fig1 = plt.figure(figsize=(12, 4))
    ax1 = fig1.gca()
    l_chl_f, = ax1.plot(t, d["chl_f"], lw=1.4, label="CHL (filtered)")
    s_chl_r  = ax1.scatter(t, d["chl_raw"], s=12, alpha=0.7, label="CHL (raw)")
    _add_event_lines(ax1, t, gi_idx, melt_idx)
    gi_proxy, melt_proxy = _proxy_lines()
    ax1.set_ylabel("Chlorophyll (a.u.)")
    ax1.set_title(f"Mixed-layer CHL with GI (solid) and melt (dashed) — {title_stub}")
    ax1.grid(alpha=0.3)
    ax1.legend([l_chl_f, s_chl_r, gi_proxy, melt_proxy],
               [h.get_label() for h in [l_chl_f, s_chl_r, gi_proxy, melt_proxy]],
               loc="upper left")
    _apply_year_ticks(ax1)
    fig1.tight_layout(); plt.show()

    # Fig 2: Temperature / T_freeze with T − Tf
    fig2 = plt.figure(figsize=(12, 4))
    axT = fig2.gca()
    l_T,  = axT.plot(t, d["ts_f"], lw=1.4, label="T (near-sfc, filtered)")
    l_Tf, = axT.plot(t, d["tf_ns"], lw=1.2, label="T_freeze (near-sfc)")
    tdiff = np.asarray(d["tdiff"], float)
    under = np.where(np.isfinite(tdiff) & (tdiff <= rtc))[0]
    shade_handle = None
    if under.size > 0:
        for i in under:
            t0, t1 = t[i], t[min(i + 1, len(t) - 1)]
            axT.axvspan(t0, t1, alpha=0.15)
        shade_handle = Patch(alpha=0.15, label=f"Under-ice (T−Tf ≤ {rtc:g})")
    axTd = axT.twinx()
    l_dT, = axTd.plot(t, tdiff, lw=1.0, label="T − Tf")
    _add_event_lines(axT, t, gi_idx, melt_idx)
    axT.set_ylabel("Temperature (°C)"); axTd.set_ylabel("T − Tf (°C)")
    axT.set_title(f"Near-surface T and freezing T with under-ice shading — {title_stub}")
    axT.grid(alpha=0.3)
    left_h, left_l   = axT.get_legend_handles_labels()
    right_h, right_l = axTd.get_legend_handles_labels()
    gi_proxy, melt_proxy = _proxy_lines()
    handles = left_h + right_h + [gi_proxy, melt_proxy]
    labels  = left_l + right_l + [gi_proxy.get_label(), melt_proxy.get_label()]
    if shade_handle is not None:
        handles.append(shade_handle); labels.append(shade_handle.get_label())
    axT.legend(handles, labels, loc="upper left")
    _apply_year_ticks(axT)
    fig2.tight_layout(); plt.show()

    # Fig 3: Salinity / |MLD|
    fig3 = plt.figure(figsize=(12, 4))
    axS = fig3.gca()
    l_S,  = axS.plot(t, d["ss_f"], lw=1.4, label="S (near-sfc, filtered)")
    axM = axS.twinx()
    mld_mag = np.abs(np.asarray(d["mld"], float))
    l_MLD, = axM.plot(t, mld_mag, lw=1.0, label="MLD |depth|")
    _add_event_lines(axS, t, gi_idx, melt_idx)
    axS.set_ylabel("Salinity"); axM.set_ylabel("MLD (m)")
    axS.set_title(f"Near-surface S with mixed-layer depth magnitude — {title_stub}")
    axS.grid(alpha=0.3)
    left_h, left_l   = axS.get_legend_handles_labels()
    right_h, right_l = axM.get_legend_handles_labels()
    gi_proxy, melt_proxy = _proxy_lines()
    axS.legend(left_h + right_h + [gi_proxy, melt_proxy],
               left_l + right_l + [gi_proxy.get_label(), melt_proxy.get_label()],
               loc="upper left")
    _apply_year_ticks(axS)
    fig3.tight_layout(); plt.show()

# -------- quick summary + A–F figures (call after writing out.nc) -------- #

def _arr(ds, name):
    return np.asarray(ds[name].values, float)

def _valid_mask(*arrays):
    m = np.ones_like(arrays[0], dtype=bool)
    for a in arrays:
        m &= np.isfinite(a)
    return m

def _to_lon180(lon):
    lon = lon.copy()
    lon = np.where(lon > 180, lon - 360, lon)
    lon = np.where(lon < -180, lon + 360, lon)
    return lon


# ---------------------------------------------------------------------

DIR = "/glade/u/home/vyaasv/Floats"
res = save_meltbloom2(rtc=0.1, bl=0, DIR=DIR,
                       qc_ok=("0","1","2","8"),
                       gi_only_if_no_ice=True,
                       chl_min_points_in_ml=3)
write_out_nc(DIR, res)
bloomweeks, bloomdays, bloomweeks_m, bloomyears, meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag = res
list_diag_summary(diag)

# Summary figures A–E from out.nc
import xarray as xr
ds_out = xr.open_dataset(os.path.join(DIR, "out.nc"))
bloomweeks = _arr(ds_out, "bloomweeks"); bloomdays = _arr(ds_out, "bloomdays")
meltweeks  = _arr(ds_out, "meltweeks");  meltdays  = _arr(ds_out, "meltdays")
mld_gi     = _arr(ds_out, "mld_gi");     lons      = _arr(ds_out, "lons")
lats       = _arr(ds_out, "lats")
plt.rcParams.update({"figure.dpi": 130})
# Fig A — Map of GI locations colored by GI week
mA = _valid_mask(bloomweeks, lons, lats)
if np.any(mA):
    lon180 = _to_lon180(lons[mA])
    plt.figure(figsize=(8,4.5))
    sc = plt.scatter(lon180, lats[mA], c=bloomweeks[mA], s=18, cmap="viridis", edgecolor="none")
    plt.colorbar(sc, label="GI week (ISO week shifted)")
    plt.xlabel("Longitude (°)"); plt.ylabel("Latitude (°)")
    plt.title("GI event locations colored by GI week"); plt.grid(alpha=0.3); plt.show()
else:
    print("Fig A: No valid GI+location data to plot.")
# Fig B — GI week vs Latitude (with linear fit)
mB = _valid_mask(bloomweeks, lats); xB, yB = lats[mB], bloomweeks[mB]
plt.figure(figsize=(6,4.5)); plt.scatter(xB, yB, s=18, alpha=0.8)
if xB.size >= 2:
    p = np.polyfit(xB, yB, 1); xx = np.linspace(np.nanmin(xB), np.nanmax(xB), 100)
    yy = p[0]*xx + p[1]; plt.plot(xx, yy, lw=2, alpha=0.8)
    plt.text(0.02, 0.98, f"slope={p[0]:.2f} wk/°", transform=plt.gca().transAxes, ha="left", va="top")
plt.xlabel("Latitude (°)"); plt.ylabel("GI week"); plt.title("GI timing vs latitude"); plt.grid(alpha=0.3); plt.show()
# Fig C — GI week vs Melt week (1:1)
mC = _valid_mask(bloomweeks, meltweeks); xC, yC = meltweeks[mC], bloomweeks[mC]
plt.figure(figsize=(5.2,5.2)); plt.scatter(xC, yC, s=22, alpha=0.85)
if xC.size:
    mn = np.nanmin(np.concatenate([xC, yC])); mx = np.nanmax(np.concatenate([xC, yC]))
    plt.plot([mn, mx], [mn, mx], ls="--", lw=1.2)
plt.xlabel("Melt week"); plt.ylabel("GI week"); plt.title("GI vs Melt week (1:1 dashed)")
plt.grid(alpha=0.3); plt.axis("equal"); plt.show()
# Fig D — Δt = GI − Melt (days) histogram
mD = _valid_mask(bloomdays, meltdays)
if np.any(mD):
    dd = bloomdays[mD] - meltdays[mD]
    plt.figure(figsize=(7,4)); bins = np.arange(-90, 91, 7)
    plt.hist(dd, bins=bins, alpha=0.85); plt.axvline(0, color="k", ls="--", lw=1)
    plt.xlabel("GI − Melt (days)"); plt.ylabel("Count"); plt.title("Distribution of GI–Melt timing difference")
    plt.grid(axis="y", alpha=0.3); plt.show()
else:
    print("Fig D: No paired GI–melt events to plot.")
# Fig E — MLD at GI histogram (abs depth, meters)
mE = np.isfinite(mld_gi)
if np.any(mE):
    mld_mag = np.abs(mld_gi[mE])
    plt.figure(figsize=(7,4)); plt.hist(mld_mag, bins=30, alpha=0.85)
    plt.xlabel("MLD at GI (m)"); plt.ylabel("Count"); plt.title("Distribution of mixed-layer depth at GI")
    plt.grid(axis="y", alpha=0.3); plt.show()
else:
    print("Fig E: No finite MLD-at-GI values to plot.")
# Fig F — choose a float with the most total picks, then plot panels
if isinstance(diag, dict) and len(diag) > 0:
    picks = [(f, len(d["gi_idx"]) + len(d["melt_idx"])) for f, d in diag.items()]
    if picks:
        best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
        print(f"Plotting Fig F for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
        plot_float_panels(diag, f=best_f)
    else:
        print("Fig F: diag exists but no picks found.")
else:
    print("Fig F: Skipped (no `diag` in memory).")
```

```python
# ===== NCS (start of nitrate consumption) — add-on to your existing pipeline =====
# This cell:
#   1) Re-runs our melt+GI detector but ALSO extracts an upper-ocean nitrate series (same gate as CHL)
#   2) Detects NCS per float: first sustained decline after a local NO3 maximum, heading to the next minimum
#   3) Stores NO3 raw/filtered + NCS indices in `diag`
#   4) Adds a nitrate figure with NCS dash-dot lines to the per-float panels
#
# Notes:
# - It reuses your helpers: safe_interp_1d, safe_filtfilt, pick_var, good_qc_mask, get_time_index, gi_fallback_from_increase
# - CHL gate logic stays identical; nitrate uses the SAME indices as CHL averaging for fairness
# - No changes to our earlier figures A–E; only the per-float time-series gets an extra nitrate figure

import numpy as np
import xarray as xr
import pandas as pd
import gsw
from scipy.signal import argrelextrema as lmm
import scipy.signal as signal
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import os

# ---------- nitrate helpers ----------
def _resolve_no3(ds):
    for base in ("NTAW", "NITRATE", "NO3"):
        adj = base + "_ADJUSTED"
        if adj in ds.variables:
            return ds[adj], base
        if base in ds.variables:
            return ds[base], base
    return None, None

def _no3_units_kind(units_attr):
    if not units_attr:
        return "mmol_m3"
    u = units_attr.replace("µ","u").lower()
    if "umol/kg" in u or "umol kg" in u or "umolkg" in u:
        return "umol_kg"
    # treat mmol/m3, umol/l, etc. as per-volume
    return "mmol_m3"

def _to_mmol_m3_from_umol_kg(no3_umolkg, pres, psal, temp, lon, lat):
    SA = gsw.SA_from_SP(psal, pres, lon, lat)
    CT = gsw.CT_from_t(SA, temp, pres)
    rho = gsw.rho(SA, CT, pres)  # kg m^-3
    # µmol/kg * kg/m^3 = µmol/m^3 -> mmol/m^3
    return (no3_umolkg * rho) / 1000.0

# ---------- NCS detection (works on a single smoothed series) ----------
def _find_ncs_indices(no3f):
    ncs_idx = []
    if np.isfinite(no3f).sum() < 5:
        return np.array(ncs_idx, dtype=int)
    try:
        n_lmax = lmm(no3f, np.greater)[0]
        n_lmin = lmm(no3f, np.less)[0]
    except Exception:
        n_lmax, n_lmin = np.array([], int), np.array([], int)

    def _first_consumption_index(f, imax, imin):
        if not (0 <= imax < len(f)) or not (0 <= imin < len(f)) or imin <= imax:
            return None
        seg = f[imax:imin+1]
        d   = np.diff(seg)
        R   = seg[0] - seg[-1]
        pre = f[max(0, imax-5):imax+1]
        sigma = np.nanstd(pre) if np.isfinite(pre).sum() >= 2 else np.nanstd(seg)
        thresh = max(0.03 * (R if np.isfinite(R) else 0.0), 0.5 * (sigma if np.isfinite(sigma) else 0.0))
        for j in range(len(d)-1):
            if (d[j] < 0.0) and (d[j+1] < 0.0) and (seg[j+1] <= seg[0] - thresh):
                return imax + (j+1)
        # fallback: first negative step after imax
        for j in range(len(d)):
            if d[j] < 0.0:
                return imax + j + 1
        return None

    # pair each max with the next min
    for imax in n_lmax:
        nxt = n_lmin[n_lmin > imax]
        if nxt.size == 0:
            continue
        imin = int(nxt[0])
        k = _first_consumption_index(no3f, int(imax), int(imin))
        if k is not None:
            ncs_idx.append(int(k))

    # global fallback if nothing found
    if not ncs_idx:
        gmax = int(np.nanargmax(no3f))
        d = np.diff(no3f[gmax:])
        for j in range(len(d)):
            if d[j] < 0.0:
                ncs_idx.append(gmax + j + 1)
                break
    return np.array(ncs_idx, dtype=int)

# ---------- Detector with NO3 add-on ----------
def save_meltbloom2_with_ncs(
    rtc=0.5,
    bl=0,
    DIR="",
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
):
    # reuse your filter design
    N = 1; Wn, Wn2 = 0.1, 0.2
    B, A  = signal.butter(N, Wn,  output="ba")
    B2, A2 = signal.butter(N, Wn2, output="ba")

    flist = glob.glob(os.path.join(DIR, "*.nc"))
    print("No. of files:", len(flist))

    # outputs (same names as yours)
    bloomdays  = np.array([]); bloomweeks = np.array([]); bloomweeks_m = np.array([])
    bloomyears = np.array([]); meltweeks  = np.array([]); meltdays     = np.array([])
    lons = np.array([]); lats = np.array([]); files = np.array([]); mld_gi = np.array([])
    mld_chl_pos = np.array([]); timing_pg = np.array([]); grth_rate_gm = np.array([])
    peak_chl = np.array([]); len_ts = np.array([]); melt_week = np.array([]); chl_sog = np.array([])
    gia_week = np.array([])

    diag = {}

    for f, path in enumerate(flist):
        print(f"\n=== File {f} { os.path.basename(path) } ===")
        ds = xr.open_dataset(path)

        # basic vars (reuse your helpers)
        if not all(v in ds.variables for v in ("TIME","LONGITUDE","LATITUDE")):
            print("  STOP: missing TIME/LONGITUDE/LATITUDE"); ds.close(); continue
        t  = pick_var(ds, "TEMP")
        s  = pick_var(ds, "PSAL")
        pr = pick_var(ds, "PRES")
        if ("CPHL" in ds.variables) or ("CPHL_ADJUSTED" in ds.variables):
            chl = pick_var(ds, "CPHL")
        else:
            print("  STOP: no CPHL/CPHL_ADJUSTED"); ds.close(); continue

        # nitrate (optional)
        no3, no3_base = _resolve_no3(ds)
        have_no3 = no3 is not None

        lon = ds["LONGITUDE"]; lat = ds["LATITUDE"]
        time = get_time_index(ds)

        lon_dim = lon.dims[0]; lat_dim = lat.dims[0]
        if "TIME" not in t.sizes:
            print("  STOP: TEMP has no TIME dim"); ds.close(); continue
        nprof = int(t.sizes["TIME"])
        if lon.sizes[lon_dim] != nprof or lat.sizes[lat_dim] != nprof:
            print("  STOP: lon/lat length mismatch vs TIME"); ds.close(); continue

        # QC (reuse your QC mask)
        for da, base in [(t,"TEMP"),(s,"PSAL"),(pr,"PRES"),(chl,"CPHL")]:
            m = good_qc_mask(ds, base, qc_ok=qc_ok)
            if m is not None and da.dims == m.dims:
                da[:] = da.where(m)
        s   = s.where(s > 28.0)
        chl = chl.where(chl >= 0)

        if have_no3:
            m_no3 = good_qc_mask(ds, no3_base, qc_ok=("1","2","8","0"))
            if m_no3 is not None and no3.dims == m_no3.dims:
                no3 = no3.where(m_no3)

        # timing diag
        if nprof >= 2:
            step_days = np.diff(time.values) / np.timedelta64(1,"D")
            tsmax = float(np.nanmax(step_days)) if len(step_days) else np.nan
            span_days = float((time[-1] - time[0]).days)
        else:
            tsmax = np.nan; span_days = np.nan
        print(f"  profiles: {nprof}, time span days: {span_days:.1f}, max gap days: {tsmax if np.isfinite(tsmax) else np.nan:.1f}")

        # quick series & containers
        rl = [None] * nprof
        pi = np.zeros(nprof)
        ts = np.array([]); ss = np.array([]); chls = np.array([]); mlds = np.array([])
        tdiff_series = []; tns_series = []; sns_series = []; tf_ns_series = []
        no3s = np.array([])

        used_shallow = used_deep = no_chl = 0

        # per profile loop (same as yours; nitrate uses same indices as CHL gate)
        for p in range(nprof):
            lonp = float(lon.isel({lon_dim: p}).values)
            latp = float(lat.isel({lat_dim: p}).values)
            pp   = np.asarray(pr.isel(TIME=p).values, dtype=float)
            chlp = np.asarray(chl.isel(TIME=p).values, dtype=float)

            dp = gsw.z_from_p(pp, latp); zpos = -dp  # +depth
            if np.isnan(np.nanmean(dp)):
                dp = gsw.z_from_p(pp, float(lat.mean().values)); zpos = -dp

            tp = np.asarray(t.isel(TIME=p).values, dtype=float)
            sp = np.asarray(s.isel(TIME=p).values, dtype=float)
            spa = gsw.SA_from_SP(sp, pp, lonp, latp)
            if np.isnan(np.nanmean(spa)):
                spa = gsw.SA_from_SP(sp, pp, float(lon.mean().values), float(lat.mean().values))
            ct = gsw.CT_from_t(spa, tp, pp)
            n2, pmid = gsw.Nsquared(spa, ct, pp)
            dmid = gsw.z_from_p(pmid, latp)
            if np.isnan(np.nanmean(dmid)):
                dmid = gsw.z_from_p(pmid, float(lat.mean().values))

            tfreeze = gsw.t_freezing(spa, 0.0, 0.0)  # p=0 dbar
            try:
                i0 = int(np.nanargmin(pp))
            except Exception:
                i0 = 0

            try:
                mld = dmid[int(np.nanargmax(n2))]
                mlds = np.append(mlds, mld)

                # near-sfc windows (as in your code)
                ddl = dp[i0] - 10.0
                ddi = int(np.nanargmin(np.abs(dp - ddl)))
                lo = min(ddi, i0); hi = max(ddi, i0) + 1
                seg = slice(lo, hi)
                vT = np.nanmean(tp[seg]); vS = np.nanmean(sp[seg])
                if np.isnan(vT) or np.isnan(vS):
                    p0 = pp[i0]; near_idx = np.where((pp - p0) <= 20.0)[0]
                    if near_idx.size:
                        if np.isnan(vT): vT = np.nanmean(tp[near_idx])
                        if np.isnan(vS): vS = np.nanmean(sp[near_idx])
                ts = np.append(ts, vT); ss = np.append(ss, vS)
                tns_series.append(vT); sns_series.append(vS)

                # CHL gate (unchanged)
                valid_idx = np.where(~np.isnan(chlp))[0]
                if valid_idx.size == 0:
                    chls = np.append(chls, np.nan); no_chl += 1
                    # nitrate stays NaN when CHL gate cannot be formed
                    no3s = np.append(no3s, np.nan)
                else:
                    shallow_idx = valid_idx[zpos[valid_idx] <= chl_gate_depth_m]
                    if shallow_idx.size > 0:
                        ic = int(shallow_idx.max()); used_shallow += 1
                    else:
                        deep_idx = valid_idx[zpos[valid_idx] <= deep_gate_depth_m]
                        ic = int((deep_idx if deep_idx.size else valid_idx).max()); used_deep += 1
                    ddic = int(np.nanargmin(np.abs(dp - mld)))
                    start = max(0, min(ddic - bl, ic)); stop = min(len(chlp), max(ddic - bl, ic) + 1)
                    vC = np.nanmean(chlp[start:stop])
                    chls = np.append(chls, vC)

                    # nitrate using SAME start:stop window
                    if have_no3:
                        no3p = np.asarray(no3.isel(TIME=p).values, dtype=float)
                        uk = _no3_units_kind(no3.attrs.get("units",""))
                        if uk == "umol_kg":
                            no3v = _to_mmol_m3_from_umol_kg(no3p, pp, sp, tp, lonp, latp)
                        else:
                            no3v = no3p
                        vN = np.nanmean(no3v[start:stop])
                        no3s = np.append(no3s, vN)
                    else:
                        no3s = np.append(no3s, np.nan)

            except Exception:
                # keep parity with your code in the exception path
                mlds = np.append(mlds, np.nan); ts = np.append(ts, np.nan); ss = np.append(ss, np.nan); chls = np.append(chls, np.nan)
                tns_series.append(np.nan); sns_series.append(np.nan); no3s = np.append(no3s, np.nan)

            # T - Tf at near-sfc
            try: tf_ns = float(tfreeze[i0])
            except Exception: tf_ns = np.nan
            tf_ns_series.append(tf_ns)
            try: tdiff = float(vT - tf_ns)
            except Exception: tdiff = np.inf
            tdiff_series.append(tdiff)

            # under-ice flag
            if np.isfinite(tdiff) and (tdiff <= rtc):
                rl[p] = 1; pi[p] = 1
            else:
                rl[p] = 0

        # seasonal axes / prints exactly as yours
        week = pd.Index(time).isocalendar().week.to_numpy()
        day  = pd.Index(time).dayofyear.to_numpy()
        year = pd.Index(time).year.to_numpy()
        week = np.where(week <= 13, week + 52, week)
        day  = np.where(day  <= 89,  day  + 365, day)

        def _ff(x): return f"{100.0 * np.isfinite(np.asarray(x)).sum() / (len(x) if len(x) else 1):.1f}"
        print(f"  built series len: ts={len(ts)} ss={len(ss)} chl={len(chls)} no3={len(no3s)}  finite% ts:{_ff(ts)} ss:{_ff(ss)} chl:{_ff(chls)} no3:{_ff(no3s)}")
        print(f"  CHL gating counts: shallow<= {int(chl_gate_depth_m)}m: {used_shallow}  deeper<= {int(deep_gate_depth_m)}m: {used_deep}  no-CHL: {no_chl}")

        # interp + filters (reuse your filters)
        ts_i   = safe_interp_1d(ts);   ss_i   = safe_interp_1d(ss);   chls_i = safe_interp_1d(chls);  no3_i = safe_interp_1d(no3s)
        chlsf = safe_filtfilt(B, A, chls_i); tsf = safe_filtfilt(B, A, ts_i); ssf = safe_filtfilt(B, A, ss_i); no3f = safe_filtfilt(B, A, no3_i)
        ssf2  = safe_filtfilt(B2, A2, ss_i)

        # ---- melt detection (unchanged)
        mis = np.array([], dtype=int)
        uice_ct = int(np.sum(pi == 1))
        if uice_ct >= 3:
            dpi = pi[1:] - pi[0:-1]
            mi  = np.where(dpi == -1)[0]
            keep = []
            for idx in mi:
                nundi = int(np.sum(pi[max(0, idx - 15): idx] == 1))
                if nundi >= 2: keep.append(idx)
            mi = np.array(keep, dtype=int)
            if mi.size > 1:
                dmi = np.abs(mi[1:] - mi[0:-1])
                kept2 = 0
                for nm, melts in enumerate(mi):
                    salt = ssf2[melts:melts + 4]
                    dsalt = np.nanmean(salt[1:] - salt[0:-1])
                    cond = (nm < len(dmi) and dmi[nm] > 12 and dsalt < 0.0) or (nm == len(dmi) and dsalt < 0.0)
                    if cond:
                        mis = np.append(mis, melts); kept2 += 1
                if mis.size > 1:
                    mis = np.sort(mis.astype(int))
                    dmi_n = np.abs(mis[1:] - mis[0:-1])
                    mis = np.delete(mis, np.where(dmi_n < 12)[0])
            elif mi.size == 1:
                m0 = int(mi[0])
                salt = ssf2[m0:m0+4]
                dsalt = np.nanmean(salt[1:] - salt[0:-1])
                if dsalt < 0.0:
                    mis = np.append(mis, m0)

        # ---- GI detection (unchanged core)
        finite_chls = np.isfinite(chls_i)
        ch_std = np.nanstd(chls_i) if finite_chls.sum() >= 2 else 0.0
        chl_anom = chls_i - (ch_std if np.isfinite(ch_std) else 0.0)
        lmax = lmm(chlsf, np.greater)[0]
        lmin = lmm(chlsf, np.less)[0]

        bloomday_f = np.array([]); bloomweek_f = np.array([]); bloomweek_m_f = np.array([])
        blm_inds = np.array([]); mld_a = np.array([]); mld_fg = np.array([]); pgweek = np.array([]); gia_a = np.array([])
        gi_found_segments = 0

        if len(lmin) > 0 and len(lmax) > 0:
            try:    dlmin = len(chlsf) - lmin[-1]
            except: dlmin, lmin = lmax[0], np.array([0])
            try:    dlmax = lmax[0]
            except: dlmax, lmax = len(chlsf)-1, np.array([len(chlsf)-1])

            nlmax, nlmin = len(lmax), len(lmin)
            if nlmin > nlmax and dlmin < 6: lmin = np.delete(lmin, -1)
            elif nlmin > nlmax and dlmin >= 6:
                chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe)
            elif nlmin < nlmax and dlmax < 9: lmax = np.delete(lmax, 0)
            elif nlmin < nlmax and dlmax >= 9: lmin = np.concatenate(([0], lmin))
            elif nlmin == nlmax and dlmin >= 6:
                dlmm = lmax - lmin
                if np.sum(dlmm < 0) > 0:
                    chlsm = chls_i.copy(); chlsm[0:lmin[-1]] = np.nan
                    lmaxe = np.nanargmax(chlsm); lmax = np.append(lmax, lmaxe); lmax = np.delete(lmax, 0)
            elif nlmin == nlmax and dlmin < 6:
                lmax = np.delete(lmax, 0); lmin = np.delete(lmin, -1)

            for l in range(len(lmax)):
                imax, imin = int(lmax[l]), int(lmin[l])
                if (imin < imax) and (abs(imax - imin) > 3):
                    chlmm = chls_i[imin:imax]
                    peak_chl = np.append(peak_chl, np.nanmax(chlmm))
                    weekmm = week[imin:imax]; daymm = day[imin:imax]
                    dchl = chlmm[1:] - chlmm[0:-1]
                    dchl = np.where(dchl > 0.0, dchl, np.nan)
                    medchl = np.nanmedian(dchl)
                    blmind = int(np.nanargmax(dchl > medchl))
                    blmind_m = int(np.nanargmax(dchl))
                    chl_fi = int(np.nanargmax(dchl > 0.0))

                    blm_inds = np.append(blm_inds, imin + blmind)
                    bloomweek_f = np.append(bloomweek_f, weekmm[blmind])
                    bloomweek_m_f = np.append(bloomweek_m_f, weekmm[blmind_m])
                    bloomday_f  = np.append(bloomday_f,  daymm[blmind])

                    mldmm = mlds[imin:imax]
                    mld_a = np.append(mld_a, mldmm[blmind])
                    mld_fg = np.append(mld_fg, mldmm[chl_fi])
                    pgweek = np.append(pgweek, weekmm[chl_fi])

                    try:
                        gia_a = np.append(gia_a, weekmm[np.min(np.where(chl_anom[imin:imax] > 0))])
                    except ValueError:
                        gia_a = np.append(gia_a, np.nan)

                    gi_found_segments += 1

        used_fallback = False
        if gi_found_segments == 0:
            gi_idx = gi_fallback_from_increase(chls_i)
            used_fallback = gi_idx is not None
            if gi_idx is not None:
                bloomday_f   = np.append(bloomday_f,  day[gi_idx])
                bloomweek_f  = np.append(bloomweek_f, week[gi_idx])
                bloomweek_m_f = np.append(bloomweek_m_f, week[gi_idx])
                if len(mlds) > gi_idx: mld_a = np.append(mld_a, mlds[gi_idx])
                else:                   mld_a = np.append(mld_a, np.nan)
                pgweek = np.append(pgweek, week[gi_idx])
                blm_inds = np.append(blm_inds, gi_idx)
                gia_week = np.append(gia_week, week[gi_idx])

        # ---- NCS detection on no3f
        ncs_idx = _find_ncs_indices(no3f) if have_no3 else np.array([], dtype=int)
        print(f"  NCS picks: {len(ncs_idx)}")

        # ---- pairing & export (unchanged)
        if (mis.size == 0) and gi_only_if_no_ice and (len(bloomweek_f) > 0):
            for e in range(len(bloomweek_f)):
                bloomdays  = np.append(bloomdays,  bloomday_f[e])
                bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                bloomyears = np.append(bloomyears, year[0])
                meltweeks  = np.append(meltweeks, np.nan)
                meltdays   = np.append(meltdays,  np.nan)
                idx_for_loc = int(blm_inds[e])
                lons  = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                lats  = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                files = np.append(files, f)
                mld_gi = np.append(mld_gi, mld_a[e])
                mld_chl_pos = np.append(mld_chl_pos, np.nan)
                timing_pg  = np.append(timing_pg, pgweek[e] if len(pgweek) > e else np.nan)
        else:
            meltday = np.array([day[m] for m in mis.astype(int)])
            for e in range(min(len(meltday), len(bloomday_f))):
                dd = abs(meltday[e] - bloomday_f[e])
                if dd <= 90:
                    bloomdays  = np.append(bloomdays,  bloomday_f[e])
                    bloomweeks = np.append(bloomweeks, bloomweek_f[e])
                    bloomweeks_m = np.append(bloomweeks_m, bloomweek_m_f[e])
                    bloomyears = np.append(bloomyears, year[0])
                    meltweeks  = np.append(meltweeks,  int(np.array([week[m] for m in mis.astype(int)])[e]))
                    meltdays   = np.append(meltdays,   meltday[e])
                    idx_for_loc = int(blm_inds[e])
                    lons = np.append(lons, float(lon.isel({lon_dim: idx_for_loc}).values))
                    lats = np.append(lats, float(lat.isel({lat_dim: idx_for_loc}).values))
                    files = np.append(files, f)
                    mld_gi = np.append(mld_gi, mld_a[e])
                    mld_chl_pos = np.append(mld_chl_pos, np.nan)
                    timing_pg   = np.append(timing_pg,   pgweek[e] if e < len(pgweek) else np.nan)

        len_ts = np.append(len_ts, len(chls_i))

        # ---- diagnostics (same as yours + NO3/NCS) ----
        if collect_diag:
            diag[f] = {
                "path": path,
                "time": time,
                "week": week, "day": day, "year": year,
                "lon": lon.values.astype(float),
                "lat": lat.values.astype(float),
                "ts_raw": ts, "ss_raw": ss, "chl_raw": chls, "no3_raw": no3s,
                "ts_f": tsf, "ss_f": ssf, "chl_f": chlsf, "no3_f": no3f,
                "mld": mlds,
                "pi": pi.astype(int),
                "tdiff": np.asarray(tdiff_series, float),
                "tf_ns": np.asarray(tf_ns_series, float),
                "melt_idx": np.asarray(mis, int),
                "gi_idx": np.asarray(blm_inds, int) if len(blm_inds) else np.array([], dtype=int),
                "ncs_idx": np.asarray(ncs_idx, int),
                "notes": {
                    "used_shallow": int(used_shallow),
                    "used_deep": int(used_deep),
                    "no_chl": int(no_chl),
                    "rtc": float(rtc),
                    "chl_gate_depth_m": float(chl_gate_depth_m),
                    "deep_gate_depth_m": float(deep_gate_depth_m),
                }
            }

        ds.close()

    print("\n=== Summary across files (same metrics as before) ===")
    print("Avg. length of time series:", np.nanmean(len_ts) if len(len_ts) else np.nan)
    print("Avg. peak chlorophyll:",     np.nanmean(peak_chl) if len(peak_chl) else np.nan)
    print("Avg. MLD at GI:",            np.nanmean(mld_gi) if len(mld_gi) else np.nan)

    return (
        bloomweeks, bloomdays, bloomweeks_m, bloomyears,
        meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag
    )

# ---------- Plot: add a nitrate panel with NCS lines (dash-dot) ----------
def _proxy_lines_ncs():
    return Line2D([0], [0], color="C0", lw=1.4, linestyle="-.", label="NCS")

def _add_event_lines_ncs(ax, t, ncs_idx, color="C0"):
    for i in np.asarray(ncs_idx, int):
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="-.")

def plot_float_panels_nitrate(diag_dict, f, title=None):
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag_dict[f]
    t = pd.DatetimeIndex(d["time"])
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []), int)
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"

    if ("no3_f" not in d) or ("no3_raw" not in d) or (len(d["no3_raw"]) == 0):
        print("NO3 not available for this float.")
        return

    plt.rcParams.update({"figure.dpi": 130})
    figN = plt.figure(figsize=(12, 4))
    axN = figN.gca()
    l_no3_f, = axN.plot(t, d["no3_f"], lw=1.4, label="NO3 (filtered)")
    s_no3_r  = axN.scatter(t, d["no3_raw"], s=12, alpha=0.7, label="NO3 (raw)")

    # melt dashed, NCS dash-dot
    from matplotlib.dates import DateFormatter, YearLocator
    for i in melt_idx:
        if 0 <= i < len(t): axN.axvline(t[i], color="C3", lw=1.2, ls="--")
    _add_event_lines_ncs(axN, t, ncs_idx)

    ncs_proxy = _proxy_lines_ncs()
    melt_proxy = Line2D([0], [0], color="C3", lw=1.2, linestyle="--", label="Melt")

    axN.set_ylabel("NO3 (mmol m$^{-3}$)")
    axN.set_title(f"Upper-ocean nitrate with NCS (dash-dot) and melt (dashed) — {title_stub}")
    axN.grid(alpha=0.3)
    axN.legend([l_no3_f, s_no3_r, ncs_proxy, melt_proxy],
               [h.get_label() for h in [l_no3_f, s_no3_r, ncs_proxy, melt_proxy]],
               loc="upper left")
    axN.xaxis.set_major_locator(mdates.YearLocator())
    axN.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    axN.set_xlabel("Year")
    figN.tight_layout()
    plt.show()

# ---------- RUN (same call signature as yours; only function name changed) ----------
DIR = "/glade/u/home/vyaasv/Floats"

(bloomweeks, bloomdays, bloomweeks_m, bloomyears,
 meltweeks, meltdays, lons, lats, files, mld_gi, gia_week, diag) = save_meltbloom2_with_ncs(
    rtc=0.3,
    bl=0,
    DIR=DIR,
    qc_ok=("0","1","2","8"),
    gi_only_if_no_ice=True,
    debug=True,
    chl_gate_depth_m=100.0,
    deep_gate_depth_m=200.0,
    collect_diag=True
)

# Quick per-float nitrate panel on the richest-picked float (like your Fig F logic)
if isinstance(diag, dict) and len(diag) > 0:
    picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", []))) for f, d in diag.items()]
    if picks:
        best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
        print(f"Plotting nitrate+NCS panel for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
        plot_float_panels_nitrate(diag, f=best_f)
    else:
        print("Nitrate panel: diag exists but no picks found.")
else:
    print("Nitrate panel: Skipped (no `diag`).")
```

```python
# === NCS summary plots A–D (winter-only) + optional CHL panel overlay with NCS ===
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import os

# Configure winter window (Aug–Jan)
WINTER_MONTHS = (8, 9, 10, 11, 12, 1)

# Helpers (reuse if present)
def _to_lon180(lon):
    lon = np.asarray(lon, float).copy()
    lon = np.where(lon > 180, lon - 360, lon)
    lon = np.where(lon < -180, lon + 360, lon)
    return lon

def _valid_mask(*arrays):
    m = np.ones_like(np.asarray(arrays[0], float), dtype=bool)
    for a in arrays:
        m &= np.isfinite(np.asarray(a, float))
    return m

def _collect_ncs_events(diag, months=WINTER_MONTHS, max_pair_days=90):
    """
    Walks diag and collects per-event NCS timing, location, and nearest melt timing.
    Returns dict of numpy arrays: ncsweeks, ncsdays, years, lons, lats, meltweeks, meltdays
    Only events whose datetime month is in `months` are kept.
    """
    months_set = set(int(m) for m in months)
    ncs_w = []; ncs_d = []; yrs = []; LON = []; LAT = []
    melt_w_at = []; melt_d_at = []

    for f, d in diag.items():
        # required fields
        if not {"time", "ncs_idx", "week", "day", "year", "lon", "lat"}.issubset(d.keys()):
            continue
        t = pd.DatetimeIndex(d["time"])
        ncs_idx = np.asarray(d.get("ncs_idx", []), dtype=int)
        melt_idx = np.asarray(d.get("melt_idx", []), dtype=int)

        # precompute seasonal coordinates
        week = np.asarray(d["week"], dtype=float)
        day  = np.asarray(d["day"],  dtype=float)
        year = np.asarray(d["year"], dtype=float)
        lon  = np.asarray(d["lon"],  dtype=float)
        lat  = np.asarray(d["lat"],  dtype=float)

        for i in ncs_idx:
            if not (0 <= i < len(t)):
                continue
            if int(t[i].month) not in months_set:
                continue

            # basic event info
            ncs_w.append(float(week[i]))
            ncs_d.append(float(day[i]))
            yrs.append(float(year[i]))
            LON.append(float(lon[i] if i < len(lon) else np.nan))
            LAT.append(float(lat[i] if i < len(lat) else np.nan))

            # pair with nearest melt within ±max_pair_days
            if melt_idx.size:
                # choose melt with minimum |Δday|
                mdays = np.asarray([day[j] for j in melt_idx if 0 <= j < len(day)], dtype=float)
                if mdays.size:
                    diffs = np.abs(mdays - day[i])
                    j = int(np.nanargmin(diffs))
                    if np.isfinite(diffs[j]) and diffs[j] <= max_pair_days:
                        melt_d_at.append(mdays[j])
                        # melt week at that same melt index
                        mweeks = np.asarray([week[jj] for jj in melt_idx if 0 <= jj < len(week)], dtype=float)
                        melt_w_at.append(mweeks[j])
                    else:
                        melt_d_at.append(np.nan); melt_w_at.append(np.nan)
                else:
                    melt_d_at.append(np.nan); melt_w_at.append(np.nan)
            else:
                melt_d_at.append(np.nan); melt_w_at.append(np.nan)

    out = dict(
        ncsweeks=np.asarray(ncs_w, float),
        ncsdays =np.asarray(ncs_d, float),
        years   =np.asarray(yrs, float),
        lons    =np.asarray(LON, float),
        lats    =np.asarray(LAT, float),
        meltweeks=np.asarray(melt_w_at, float),
        meltdays =np.asarray(melt_d_at, float),
    )
    return out

def _linear_fit_scatter(ax, x, y, xlabel, ylabel, title):
    ax.scatter(x, y, s=18, alpha=0.85)
    if np.isfinite(x).sum() >= 2 and np.isfinite(y).sum() >= 2:
        m = np.isfinite(x) & np.isfinite(y)
        p = np.polyfit(x[m], y[m], 1)
        xx = np.linspace(np.nanmin(x[m]), np.nanmax(x[m]), 100)
        yy = p[0]*xx + p[1]
        ax.plot(xx, yy, lw=2, alpha=0.9)
        ax.text(0.02, 0.98, f"slope={p[0]:.2f}", transform=ax.transAxes, ha="left", va="top")
    ax.set_xlabel(xlabel); ax.set_ylabel(ylabel); ax.set_title(title); ax.grid(alpha=0.3)

def plot_chl_panel_with_ncs(diag, f, title=None):
    """
    Re-creates your CHL figure (filtered + raw) and adds:
      - GI vertical lines (solid)
      - Melt vertical lines (dashed)
      - NCS vertical lines (dash-dot, winter-only)
    Uses only what is already stored in `diag`.
    """
    if f not in diag:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag[f]
    if not {"time","chl_f","chl_raw"}.issubset(d.keys()):
        print("CHL series not available in diag for this float.")
        return
    t = pd.DatetimeIndex(d["time"])
    gi_idx   = np.asarray(d.get("gi_idx", []), int)
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []), int)
    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"

    plt.rcParams.update({"figure.dpi": 130})
    fig = plt.figure(figsize=(12, 4))
    ax = fig.gca()
    l_chl_f, = ax.plot(t, d["chl_f"], lw=1.4, label="CHL (filtered)")
    s_chl_r  = ax.scatter(t, d["chl_raw"], s=12, alpha=0.7, label="CHL (raw)")

    # GI solid, Melt dashed
    for i in gi_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color="C3", lw=1.2, ls="-")
    for i in melt_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color="C3", lw=1.2, ls="--")

    # NCS winter-only dash-dot
    months_set = set(int(m) for m in WINTER_MONTHS)
    for i in ncs_idx:
        if 0 <= i < len(t) and int(t[i].month) in months_set:
            ax.axvline(t[i], color="C0", lw=1.2, ls="-.")

    from matplotlib.lines import Line2D
    gi_proxy   = Line2D([0],[0], color="C3", lw=1.2, ls="-",  label="GI")
    melt_proxy = Line2D([0],[0], color="C3", lw=1.2, ls="--", label="Melt")
    ncs_proxy  = Line2D([0],[0], color="C0", lw=1.2, ls="-.", label=f"NCS {WINTER_MONTHS}")

    ax.set_ylabel("Chlorophyll (a.u.)")
    ax.set_title(f"CHL with GI (solid), melt (dashed), winter-only NCS (dash-dot) — {title_stub}")
    ax.grid(alpha=0.3)
    ax.legend([l_chl_f, s_chl_r, gi_proxy, melt_proxy, ncs_proxy],
              [h.get_label() for h in [l_chl_f, s_chl_r, gi_proxy, melt_proxy, ncs_proxy]],
              loc="upper left")
    ax.xaxis.set_major_locator(mdates.YearLocator()); ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    ax.set_xlabel("Year")
    fig.tight_layout()
    plt.show()

# ===== Build and plot =====
if 'diag' not in globals() or not isinstance(diag, dict) or len(diag) == 0:
    raise RuntimeError("This cell expects a populated `diag` from your GI+melt(+NCS) run.")

evt = _collect_ncs_events(diag, months=WINTER_MONTHS, max_pair_days=90)

plt.rcParams.update({"figure.dpi": 130})

# Aₙ — Map of NCS locations colored by NCS week
mA = _valid_mask(evt["ncsweeks"], evt["lons"], evt["lats"])
if np.any(mA):
    lon180 = _to_lon180(evt["lons"][mA])
    plt.figure(figsize=(8,4.5))
    sc = plt.scatter(lon180, evt["lats"][mA], c=evt["ncsweeks"][mA], s=18, cmap="viridis", edgecolor="none")
    plt.colorbar(sc, label="NCS week (ISO week shifted)")
    plt.xlabel("Longitude (°)"); plt.ylabel("Latitude (°)")
    plt.title(f"NCS event locations (winter months {WINTER_MONTHS})")
    plt.grid(alpha=0.3); plt.show()
else:
    print("Aₙ: No valid NCS+location data to plot (winter-only).")

# Bₙ — NCS week vs Latitude
mB = _valid_mask(evt["lats"], evt["ncsweeks"])
if np.any(mB):
    fig = plt.figure(figsize=(6,4.5)); ax = fig.gca()
    _linear_fit_scatter(ax, evt["lats"][mB], evt["ncsweeks"][mB],
                        "Latitude (°)", "NCS week", "NCS timing vs latitude (winter-only)")
    plt.show()
else:
    print("Bₙ: No valid NCS week vs latitude data (winter-only).")

# Cₙ — NCS week vs Melt week (1:1)
mC = _valid_mask(evt["ncsweeks"], evt["meltweeks"])
xC, yC = evt["meltweeks"][mC], evt["ncsweeks"][mC]
if np.any(mC):
    fig = plt.figure(figsize=(5.2,5.2)); ax = fig.gca()
    ax.scatter(xC, yC, s=22, alpha=0.85)
    if xC.size:
        mn = np.nanmin(np.concatenate([xC, yC])); mx = np.nanmax(np.concatenate([xC, yC]))
        ax.plot([mn, mx], [mn, mx], ls="--", lw=1.2)
    ax.set_xlabel("Melt week"); ax.set_ylabel("NCS week")
    ax.set_title("NCS vs Melt week (winter-only NCS)"); ax.grid(alpha=0.3); ax.axis("equal")
    plt.show()
else:
    print("Cₙ: No paired NCS–melt weeks within ±90 days (winter-only NCS).")

# Dₙ — Δt = NCS − Melt (days) histogram
mD = _valid_mask(evt["ncsdays"], evt["meltdays"])
if np.any(mD):
    dd = evt["ncsdays"][mD] - evt["meltdays"][mD]
    plt.figure(figsize=(7,4))
    bins = np.arange(-90, 91, 7)  # weekly bins over ±90 days
    plt.hist(dd, bins=bins, alpha=0.85)
    plt.axvline(0, color="k", ls="--", lw=1)
    plt.xlabel("NCS − Melt (days)"); plt.ylabel("Count")
    plt.title("Distribution of NCS–Melt timing difference (winter-only NCS)")
    plt.grid(axis="y", alpha=0.3); plt.show()
else:
    print("Dₙ: No paired NCS–melt events to plot (winter-only NCS).")

# Optional: pick a float and overlay NCS on the CHL panel
# Choose the float with the most combined picks
picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", []))) for f, d in diag.items()]
if picks:
    best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
    print(f"Overlaying NCS on CHL panel for float index {best_f}: {os.path.basename(diag[best_f]['path'])}")
    plot_chl_panel_with_ncs(diag, f=best_f)
else:
    print("CHL+NCS panel: no floats with picks found.")
```

```python
# ===== NO3 plotting helper (raw + filtered), CHL-style =====
# Requires: diag[f]["no3_raw"], diag[f]["no3_f"], and "time", "gi_idx", "melt_idx"
# Optional:  diag[f]["ncs_idx"] for NCS markers
# Notes: Forces GI/Melt legend proxies to red ("C3") to match the plotted lines.

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.lines import Line2D

WINTER_MONTHS = (8, 9, 10, 11, 12, 1)

def _proxy_lines_no3():
    # Legend proxies that MATCH the plotted event lines (red)
    gi_proxy   = Line2D([0], [0], color="C3", lw=1.2, linestyle="-",  label="GI")
    melt_proxy = Line2D([0], [0], color="C3", lw=1.2, linestyle="--", label="Melt")
    return gi_proxy, melt_proxy

def _add_event_lines_local(ax, t, gi_idx, melt_idx, color="C3"):
    # Draw the actual lines in red to match legend proxies
    gi_idx   = np.asarray(gi_idx, int)
    melt_idx = np.asarray(melt_idx, int)
    for i in gi_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="-")
    for i in melt_idx:
        if 0 <= i < len(t): ax.axvline(t[i], color=color, lw=1.2, ls="--")

def _apply_year_ticks_local(ax, label="Year"):
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
    ax.set_xlabel(label)

def _add_ncs_lines(ax, t, ncs_idx, months=None, color="C0"):
    if ncs_idx is None:
        return None
    months_set = set(int(m) for m in months) if months is not None else None
    for i in np.asarray(ncs_idx, int):
        if 0 <= i < len(t):
            if (months_set is None) or (int(t[i].month) in months_set):
                ax.axvline(t[i], color=color, lw=1.2, ls="-.")
    # Legend proxy that matches the drawn NCS lines
    return Line2D([0], [0], color=color, lw=1.2, linestyle="-.", label="NCS")

def plot_float_panels_no3(
    diag_dict,
    f,
    title=None,
    show_ncs=True,
    winter_only_for_ncs=True
):
    """
    Single NO3 figure (CHL-style):
      - NO3 (filtered) line + NO3 (raw) scatter
      - GI vertical lines (solid red) and Melt vertical lines (dashed red)
      - Optional NCS vertical lines (dash-dot blue), winter-only if requested
    """
    if f not in diag_dict:
        raise ValueError(f"float index {f} not found in diagnostics")
    d = diag_dict[f]

    if ("no3_f" not in d) or ("no3_raw" not in d) or (len(d["no3_raw"]) == 0):
        print("NO3 not available in diag for this float. Run the NCS add-on that computes nitrate series.")
        return

    t = pd.DatetimeIndex(d["time"])
    gi_idx   = np.asarray(d.get("gi_idx", []),   int)
    melt_idx = np.asarray(d.get("melt_idx", []), int)
    ncs_idx  = np.asarray(d.get("ncs_idx", []),  int) if (show_ncs and ("ncs_idx" in d)) else None

    title_stub = title or f"{os.path.basename(d['path'])} (float #{f})"
    plt.rcParams.update({"figure.dpi": 130})

    fig = plt.figure(figsize=(12, 4))
    ax  = fig.gca()

    # NO3 raw + filtered
    l_no3_f, = ax.plot(t, d["no3_f"], lw=1.4, label="NO3 (filtered)")
    s_no3_r  = ax.scatter(t, d["no3_raw"], s=12, alpha=0.7, label="NO3 (raw)")

    # Event lines (always use our local red implementation)
    _add_event_lines_local(ax, t, gi_idx, melt_idx)
    gi_proxy, melt_proxy = _proxy_lines_no3()  # legend proxies in red

    # Optional NCS (blue dash-dot, winter-only if requested)
    ncs_proxy = None
    if show_ncs and (ncs_idx is not None) and (len(ncs_idx) > 0):
        months = WINTER_MONTHS if winter_only_for_ncs else None
        ncs_proxy = _add_ncs_lines(ax, t, ncs_idx, months=months, color="C0")

    ax.set_ylabel("NO3 (mmol m$^{-3}$)")
    ttl_extra = "winter-only NCS" if (show_ncs and winter_only_for_ncs) else ("NCS" if show_ncs else None)
    if ttl_extra:
        ax.set_title(f"Upper-ocean nitrate with GI (solid), melt (dashed), {ttl_extra} (dash-dot) — {title_stub}")
    else:
        ax.set_title(f"Upper-ocean nitrate with GI (solid) and melt (dashed) — {title_stub}")
    ax.grid(alpha=0.3)

    # Legend (proxies are explicitly colored to match the lines)
    handles = [l_no3_f, s_no3_r, gi_proxy, melt_proxy]
    labels  = [h.get_label() for h in handles]
    if ncs_proxy is not None:
        handles.append(ncs_proxy); labels.append(ncs_proxy.get_label())
    ax.legend(handles, labels, loc="upper left")

    _apply_year_ticks_local(ax, label="Year")
    fig.tight_layout()
    plt.show()

# Example (after your NCS-enabled run populated `diag`):
picks = [(f, len(d.get("gi_idx", [])) + len(d.get("melt_idx", [])) + len(d.get("ncs_idx", []))) for f, d in diag.items()]
best_f = sorted(picks, key=lambda x: x[1], reverse=True)[0][0]
plot_float_panels_no3(diag, f=best_f, show_ncs=True, winter_only_for_ncs=True)
```
